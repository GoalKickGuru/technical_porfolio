```python
# STEP 2: GROUP BY ANALYSIS (TEMPLATE)
# Replace with your grouping variables
def run_grouping_analysis(data, group_by_col, agg_col, metrics=['mean', 'count', 'std']):
    """Template function for grouping analysis"""
    result = data.groupby(group_by_col)[agg_col].agg(metrics)
    return result

# EXAMPLE USAGE:
# grouping_result = run_grouping_analysis(
#     df_sub,
#     group_by_col='YOUR_GROUP_COLUMN',
#     agg_col='YOUR_METRIC_COLUMN'
# )
# display(grouping_result)
```

```python
# STEP 3: BINNING TEMPLATE
def create_bins(data, column, bins_edges, bin_labels):
    """Create categorical bins from numerical data"""
    if len(bins_edges) != len(bin_labels) + 1:
        raise ValueError("bins_edges must have 1 more element than bin_labels")
    
    data[f'{column}_binned'] = pd.cut(
        data[column],
        bins=bins_edges,
        labels=bin_labels
    )
    return data

# EXAMPLE:
# df_binned = create_bins(
#     df_sub,
#     column='INCOME',
#     bins_edges=[0, 50000, 100000, 200000],
#     bin_labels=['Low', 'Medium', 'High']
# )
```

```python
# STEP 4: CLEANING TEMPLATE
def clean_dataset(data, remove_duplicates=True, handle_missing='drop'):
    """Apply standard cleaning operations"""
    
    # Track original size
    original_shape = data.shape
    
    # Remove duplicates
    if remove_duplicates:
        data = data.drop_duplicates()
    
    # Handle missing values
    if handle_missing == 'drop':
        data = data.dropna()
    elif handle_missing == 'fill_zero':
        data = data.fillna(0)
    elif handle_missing == 'fill_mean':
        numeric_cols = data.select_dtypes(include=[np.number]).columns
        data[numeric_cols] = data[numeric_cols].fillna(data[numeric_cols].mean())
    
    print(f"Original shape: {original_shape}")
    print(f"Cleaned shape: {data.shape}")
    print(f"Rows removed: {original_shape[0] - data.shape[0]}")
    
    return data
```

## 5. VISUALIZATION TEMPLATES

```python
# DISTRIBUTION PLOT TEMPLATE
def plot_distribution(data, column, title=None):
    """Plot distribution of numerical column"""
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))
    
    # Histogram
    ax[0].hist(data[column], bins=30, edgecolor='black', alpha=0.7)
    ax[0].set_title(title or f'Distribution of {column}')
    ax[0].set_xlabel(column)
    ax[0].set_ylabel('Frequency')
    
    # Box plot
    ax[1].boxplot(data[column], vert=True)
    ax[1].set_title(f'Box Plot - {column}')
    ax[1].set_ylabel(column)
    
    plt.tight_layout()
    plt.show()

# USAGE:
# plot_distribution(df_sub, 'YOUR_NUMERIC_COLUMN')
```

```python
# CATEGORY BAR CHART TEMPLATE
def plot_categorical_bar(data, column, top_n=10):
    """Plot top N categories as bar chart"""
    value_counts = data[column].value_counts().head(top_n)
    
    fig, ax = plt.subplots(figsize=(12, 6))
    value_counts.plot(kind='bar', ax=ax, legend=False)
    ax.set_title(f'Top {top_n} Categories - {column}')
    ax.set_xlabel(column)
    ax.set_ylabel('Count')
    ax.tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()
```
