```python
# ===================== CONFIG =====================
from pathlib import Path
DATA_PATH = Path("../data/marketing_campaign.csv")   # change to your file
GROUP_BY = "Kidhome"
VALUE_COL = "NumStorePurchases"
SORT_BY = "NumStorePurchases"
BINS = [0, 4, 8, 13]
BIN_LABELS = ["Low", "Moderate", "High"]
FILL_STRATEGY = "median"
FILL_COLUMNS = ["Income"]
# ==================================================
```

```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name in ("notebooks", "templates") else Path.cwd()
sys.path.insert(0, str(ROOT / "scripts"))

from data_preparation import (
    group_mean, sort_dataframe, categorize_numeric,
    missing_report, fill_missing, prepare_pipeline, print_audit
)

df = pd.read_csv(DATA_PATH)
print(f"Loaded {df.shape}")
df.head(2)
```

```python
# Grouping
if GROUP_BY in df.columns and VALUE_COL in df.columns:
    print(group_mean(df, GROUP_BY, VALUE_COL))
```

```python
# Sorting + binning
if SORT_BY in df.columns:
    display(sort_dataframe(df, SORT_BY, ascending=False).head(5))
    cats = categorize_numeric(df[SORT_BY], bins=BINS, labels=BIN_LABELS)
    print(cats.value_counts())
```

```python
# Missing values + pipeline
print(missing_report(df).query("missing_count > 0"))
cleaned, audit = prepare_pipeline(
    df,
    fill_strategy=FILL_STRATEGY,
    fill_columns=[c for c in FILL_COLUMNS if c in df.columns],
    sort_by=SORT_BY if SORT_BY in df.columns else None,
)
print_audit(audit)
cleaned.head(3)
```
