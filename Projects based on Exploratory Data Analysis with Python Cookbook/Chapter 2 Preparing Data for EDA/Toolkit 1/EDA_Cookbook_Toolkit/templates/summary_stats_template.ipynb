# Reusable Template – Summary Statistics for Any Dataset

Copy this notebook, change the **DATA_PATH**, **VALUE_COL** and optional **GROUP_COL**, then run all cells.

Built on the material from *Exploratory Data Analysis with Python Cookbook* Chapter 1, plus practical extensions.


```python
# ========== USER CONFIGURATION ==========
DATA_PATH   = '../data/covid-data.csv'   # <-- change to your file
VALUE_COL   = 'new_cases'                # <-- numeric column to analyse
GROUP_COL   = 'continent'                # <-- optional categorical column (or None)
PERCENTILES = [10, 25, 50, 75, 90, 95, 99]
# ========================================
```

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name in ('notebooks', 'templates') else Path.cwd()
sys.path.insert(0, str(ROOT / 'scripts'))
from summary_statistics import *

df = pd.read_csv(DATA_PATH)
print(f'Loaded {df.shape[0]} rows × {df.shape[1]} columns')
print('Columns:', list(df.columns))
assert VALUE_COL in df.columns, f'{VALUE_COL} not found'
series = df[VALUE_COL]
```

## Quick numeric overview (pandas describe)

```python
display(series.describe(percentiles=[p/100 for p in PERCENTILES]))
```

## Full custom summary

```python
summary = full_summary(series, percentiles=PERCENTILES)
print_summary_report(summary, f'Summary of `{VALUE_COL}`')
```

## Group-by (if configured)

```python
if GROUP_COL and GROUP_COL in df.columns:
    grouped = summary_by_group(df, VALUE_COL, GROUP_COL)
    display(grouped)
else:
    print('No GROUP_COL set or column missing – skipping.')
```

## Visual check

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(series.dropna(), bins=40, color='teal', edgecolor='white')
ax[0].axvline(summary['mean'], color='red', ls='--', label='mean')
ax[0].axvline(summary['median'], color='orange', ls='--', label='median')
ax[0].legend()
ax[0].set_title(VALUE_COL)

ax[1].boxplot(series.dropna(), vert=True)
ax[1].set_title('Boxplot')
plt.tight_layout()
plt.show()
```
