# Extended Exploratory Data Analysis (EDA) Pipeline
**Based on *Exploratory Data Analysis with Python Cookbook* (Oluleye, 2023)**

This notebook provides an automated, re-usable execution environment for dataset profiling, statistical summary generation, and distribution evaluation.

```python
import numpy as np
import pandas as pd
from scipy import stats
from eda_toolkit import ExtendedEDA

# Display configuration
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
```

## 1. Data Ingestion & Structural Inspection

```python
# Load sample or target COVID-19 dataset
data_path = 'covid-data.csv'

try:
    covid_data = pd.read_csv(data_path)
except FileNotFoundError:
    # Generating synthetic mock data matching book structure for demonstration
    np.random.seed(42)
    covid_data = pd.DataFrame({
        'iso_code': ['AFG']*500 + ['USA']*500,
        'continent': ['Asia']*500 + ['North America']*500,
        'location': ['Afghanistan']*500 + ['United States']*500,
        'date': pd.date_range('2020-01-01', periods=500).tolist() * 2,
        'total_cases': np.cumsum(np.random.poisson(lam=100, size=1000)),
        'new_cases': np.random.negative_binomial(n=1, p=0.001, size=1000)
    })

print('Dataset Shape:', covid_data.shape)
covid_data.head()
```

## 2. Automated Summary Statistics Execution

```python
# Initialize ExtendedEDA object
eda_pipeline = ExtendedEDA(covid_data, target_cols=['total_cases', 'new_cases'])

# Run statistical profiling
summary_matrix = eda_pipeline.to_dataframe()
summary_matrix
```

## 3. In-Depth Metric Inspection & Export

```python
# Inspect granular statistical profile for 'new_cases'
detailed_stats = eda_pipeline.generate_summary_statistics('new_cases')
print("Detailed Metric Profile for 'new_cases':")
for key, val in detailed_stats.items():
    print(f"  - {key}: {val}")

# Export JSON artifact
eda_pipeline.export_summary_json('covid_summary_profile.json')
```
