## Technical requirements
We will be leveraging the `pandas`, `numpy`, and `scipy` libraries in Python for this chapter.

```
import numpy as np
import pandas as pd
from scipy import stats
```

## Load and inspect the data
Load the `.csv` into a dataframe using `read_csv`, then subset the dataframe to include only relevant columns.

```
covid_data = pd.read_csv("data/covid-data.csv")
covid_data = covid_data[['iso_code','continent','location','date','total_cases','new_cases']]
```

Take a quick look at the data. Check the first few rows, the data types, and the number of columns and rows.

```
covid_data.head(5)
```

```
covid_data.dtypes
```

```
covid_data.shape
```

## Analyzing the mean of a dataset
The mean is the average of a dataset — sum of all data points divided by the number of data points. It is sensitive to outliers.

To analyze the mean, we use the `mean` method from the `numpy` library.

```
data_mean = np.mean(covid_data["new_cases"])
data_mean
```

**There's more...** The `mean` method in `pandas` can also be used to compute the mean of a dataset.

```
covid_data["new_cases"].mean()
```

## Checking the median of a dataset
The median is the middle value within a sorted dataset. Unlike the mean, the median is not sensitive to outliers.

To analyze the median, we use the `median` method from the `numpy` library.

```
data_median = np.median(covid_data["new_cases"])
data_median
```

**There's more...** The `median` method in `pandas` can also be used to compute the median of a dataset.

```
covid_data["new_cases"].median()
```

## Identifying the mode of a dataset
The mode is the value that occurs most frequently in the dataset. Unlike the mean and median, the mode can be applied to both numeric and non-numeric values.

To analyze the mode, we use the `mode` method from the `stats` module in the `scipy` library.

```
data_mode = stats.mode(covid_data["new_cases"])
data_mode
```

Subset the result to extract the mode value.

```
data_mode[0]
```

```
data_mode[0][0]
```

**There's more...** The mode can also be computed for a non-numeric column, such as `continent`, to view the most frequently occurring continent in the data.

```
stats.mode(covid_data["continent"])
```

## Checking the variance of a dataset
The variance gives us a sense of the spread, or variability, of a dataset.

To analyze the variance, we use the `var` method from the `numpy` library.

```
data_variance = np.var(covid_data["new_cases"])
data_variance
```

**There's more...** The `var` method in `pandas` can also be used to compute the variance of a dataset.

```
covid_data["new_cases"].var()
```

## Identifying the standard deviation of a dataset
The standard deviation is derived from the variance and is simply the square root of the variance. It is expressed in the same units as the dataset.

To analyze the standard deviation, we use the `std` method from the `numpy` library.

```
data_sd = np.std(covid_data["new_cases"])
data_sd
```

**There's more...** The `std` method in `pandas` can also be used to compute the standard deviation of a dataset.

```
covid_data["new_cases"].std()
```

## Generating the range of a dataset
The range is the difference between the minimum and maximum values within a dataset.

To analyze the range, we use the `max` and `min` methods from the `numpy` library.

```
data_max = np.max(covid_data["new_cases"])
data_min = np.min(covid_data["new_cases"])
print(data_max, data_min)
```

```
data_range = data_max - data_min
data_range
```

**There's more...** The `max` and `min` methods in `pandas` can also be used to compute the range of a dataset.

```
covid_data["new_cases"].max() - covid_data["new_cases"].min()
```

## Identifying the percentiles of a dataset
The percentile divides the dataset into 100 equal portions. The value of the 50th percentile equals the median.

To analyze the percentile, we use the `percentile` method from the `numpy` library. Here we compute the 60th percentile.

```
data_percentile = np.percentile(covid_data["new_cases"], 60)
data_percentile
```

**There's more...** The `quantile` method in `pandas` can also be used to compute the percentile of a dataset.

```
covid_data["new_cases"].quantile(0.60)
```

## Checking the quartiles of a dataset
The quartile divides the dataset into 4 equal portions. Percentiles and quartiles are both called quantiles.

To analyze the quartiles, we use the `quantile` method from the `numpy` library. Here we compute the third quartile (Q3).

```
data_quartile = np.quantile(covid_data["new_cases"], 0.75)
data_quartile
```

**There's more...** The `quantile` method in `pandas` can also be used to compute the quartile of a dataset.

```
covid_data["new_cases"].quantile(0.75)
```

## Analyzing the interquartile range (IQR) of a dataset
The IQR is the distance between the first and third quartiles. It measures the spread of the middle 50% of values and is not affected by outliers.

To analyze the IQR, we use the `iqr` method from the `stats` module in the `scipy` library.

```
data_IQR = stats.iqr(covid_data["new_cases"])
data_IQR
```

## Summary
This notebook walked through generating key summary statistics — mean, median, mode, variance, standard deviation, range, percentiles, quartiles, and IQR — using `numpy`, `pandas`, and `scipy`.
