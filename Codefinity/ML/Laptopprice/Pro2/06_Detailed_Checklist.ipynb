## Part 1 — This Project's Step-by-Step Checklist

### Step 1 — Load & Inspect
- [ ] Import pandas, numpy, matplotlib, seaborn
- [ ] Load `laptop_price.csv` into `df`
- [ ] Print `df.shape`, `df.head()`, `df.info()`
- [ ] Run `df.describe(include='all').T` for full summary stats
- [ ] Identify every object-typed column that is secretly numeric


### Step 2 — Data Quality Audit
- [ ] Build a dtype/nunique/missing audit table
- [ ] Check `df.duplicated().sum()`; inspect flagged rows before dropping
- [ ] Print `.unique()` for every categorical column; look for disguised missing values (`"Not Available"`, `"No"`)
- [ ] Check for inconsistent category casing (e.g. `brand`'s `acer` vs `ASUS`)


### Step 3 — Memory & Storage Features
- [ ] Loop over `['ram_gb', 'ssd', 'hdd', 'graphic_card_gb']`; strip `" GB"`, cast to int
- [ ] Create `total_storage = ssd + hdd`
- [ ] Create `has_ssd`, `has_hdd` boolean flags
- [ ] Create `ssd_ratio` with divide-by-zero protection
- [ ] Drop original `ssd`, `hdd` columns
- [ ] Verify with `.head()` on the new columns


### Step 4 — OS, Rating & Generation Cleaning
- [ ] Clean `os_bit`: `.astype(str)` → strip `"-bit"` → `.astype(int)`
- [ ] Clean `rating`: `.astype(str)` → strip `" stars"` and `" star"` → `.astype(int)`
- [ ] Clean `processor_gnrtn`: `.astype(str)` → replace `"Not Available"` with `"0"` → strip `"th"` → `.astype(int)`
- [ ] Confirm every conversion is idempotent (re-run the cell — it should not error)


### Step 5 — Domain-Derived Features
- [ ] `processor_tier` ordinal mapping across the processor lineup
- [ ] `is_premium_brand` flag
- [ ] `popularity_score` (log-transformed ratings + reviews)
- [ ] `is_gaming` flag from the `weight` column
- [ ] `performance_index` composite score
- [ ] One-sentence justification written for each new feature
- [ ] Checked whether any new feature is an exact linear combination of features you're keeping


### Step 6 — Advanced EDA
- [ ] Histogram + skewness of `Price`; create `log_price`
- [ ] Written decision: model raw `Price` or `log_price`, and why
- [ ] Correlation heatmap on numeric features; top-5 correlated-with-`Price` list
- [ ] VIF table computed; multicollinear features flagged
- [ ] Boxplots: `Price` by `ram_gb`, `graphic_card_gb`, `brand`
- [ ] IQR outlier detection on `Price`; explicit keep/cap/remove decision with justification


### Step 7 — Leakage-Safe Encoding
- [ ] Bucket categorical columns by cardinality (< 5 → one-hot, ≥ 5 → target-encode)
- [ ] Train/test split performed **before** any target-encoding statistic is computed
- [ ] Target-encoding map fit on `y_train`/`X_train` only
- [ ] Map applied to `X_test`; unseen categories filled with the training global mean
- [ ] Zero `NaN`s confirmed in both `X_train` and `X_test`


### Step 8 — Baseline Models
- [ ] Mean-predictor baseline scored (MAE/RMSE/R²/MAPE)
- [ ] `LinearRegression` scored
- [ ] `Ridge` and `Lasso` scored; coefficients compared to plain linear regression
- [ ] Linear models fit on a feature set with known-multicollinear composites excluded


### Step 9 — Tree-Ensemble Models
- [ ] `RandomForestRegressor` (default hyperparameters) scored
- [ ] `GradientBoostingRegressor` (default hyperparameters) scored
- [ ] (Stretch) `XGBRegressor` scored if available
- [ ] All models so far collected into one comparison table, sorted by MAE


### Step 10 — Cross-Validation & Tuning
- [ ] 5-fold `cross_val_score` (MAE) computed for the top 1–2 candidate models
- [ ] Mean ± std CV MAE reported (not just a point estimate)
- [ ] Hyperparameter distribution defined for the chosen model family
- [ ] `RandomizedSearchCV` (or `GridSearchCV`) run with `cv=5`
- [ ] Best params and best CV score reported
- [ ] Tuned model refit and evaluated once on the held-out test set


### Step 11 — Evaluation & Diagnostics
- [ ] MAE, RMSE, R², MAPE all reported together for the final model
- [ ] Predicted-vs-actual scatterplot with a `y=x` reference line
- [ ] Residuals-vs-predicted scatterplot checked for heteroscedasticity (funnel shape)
- [ ] Residual histogram checked for centering near zero
- [ ] Written note on where the model struggles most (e.g. high-price segment)


### Step 12 — Feature Importance
- [ ] Impurity-based `feature_importances_` top-10 plotted
- [ ] Permutation importance (test set, `n_repeats>=10`) top-10 plotted
- [ ] Agreement/disagreement between the two rankings discussed
- [ ] Any disagreement attributed to a specific, named cause (e.g. cardinality bias)


### Step 13 — Persistence & Inference
- [ ] Final model saved with `joblib.dump`
- [ ] Encoding maps + column order saved alongside the model (not just the model alone)
- [ ] `predict_price()` (or equivalent) function written, reproducing every training-time transform
- [ ] Function tested on 2–3 made-up laptops with plausible outputs


### Step 14 — Conclusions
- [ ] Final model choice stated with a one-line justification
- [ ] Expected error margin translated into plain-language terms (currency, %, or both)
- [ ] Top 3–5 price drivers named
- [ ] At least one limitation named
- [ ] At least one concrete next step named


## Part 2 — Transferable Checklist for Similar Tabular Regression Projects

### Phase 1 — Understand the Problem
- [ ] Target variable and task type identified (regression vs. classification)
- [ ] Stakeholder-relevant metric(s) identified (currency? percentage? pass/fail threshold?)
- [ ] Domain context understood (what drives price/outcome in *this* domain specifically)
- [ ] Cost asymmetry considered (is overprediction worse than underprediction, or vice versa?)


### Phase 2 — Data Acquisition & Audit
- [ ] Shape, dtypes, nulls, duplicates all checked
- [ ] Dtype audit table built (expected type vs. actual type per column)
- [ ] Every column needing unit/text-artifact stripping listed
- [ ] Sentinel values and disguised missing data identified
- [ ] Category-casing / spelling inconsistencies checked


### Phase 3 — Feature Engineering
- [ ] Unit-bearing strings converted to numerics, idempotently
- [ ] Missing-value strategy chosen per column (impute / sentinel / drop) and justified
- [ ] Composite features engineered where domain logic supports it (sums, ratios, flags)
- [ ] Every transformation documented for reproducibility
- [ ] Checked for accidentally-manufactured multicollinearity in engineered composites


### Phase 4 — EDA
- [ ] Correlation heatmap on the numeric subset
- [ ] Target's univariate distribution examined; skew handled if present
- [ ] Categorical-feature-vs-target plots (boxplots/violins)
- [ ] Multicollinearity checked via VIF, not just the correlation heatmap
- [ ] Leakage-prone columns identified (anything derived from or highly coupled to the target)


### Phase 5 — Encoding & Splitting
- [ ] Categorical columns partitioned by cardinality
- [ ] Low cardinality → one-hot; high cardinality → target/frequency encoding
- [ ] Split performed **before** any target-dependent computation
- [ ] Target encoding fit on train only; unseen test categories handled with a fallback
- [ ] `random_state` fixed everywhere for reproducibility


### Phase 6 — Modeling
- [ ] Trivial baseline established (mean predictor or similar) for comparison
- [ ] At least one linear model fit as a second baseline
- [ ] Random Forest fit as a first serious non-linear model
- [ ] At least one boosting model (Gradient Boosting/XGBoost/LightGBM) fit for comparison
- [ ] Train vs. test metrics compared to detect over/underfitting


### Phase 7 — Evaluation & Interpretation
- [ ] MAE, RMSE, and R² reported together (never R² alone)
- [ ] A scale-appropriate metric added if useful (MAPE, or a domain-specific metric)
- [ ] Predicted-vs-actual plot produced for residual diagnostics
- [ ] Feature importance computed via both an impurity-based AND a permutation method
- [ ] Top features sanity-checked against domain intuition


### Phase 8 — Iterate & Communicate
- [ ] Hyperparameter tuning attempted (`RandomizedSearchCV`/`GridSearchCV`)
- [ ] Cross-validation used for a robust performance estimate, not a single split
- [ ] At least one alternative model family compared against the leading candidate
- [ ] Findings summarized for a non-technical audience in plain language
- [ ] Model + preprocessing artifacts persisted together for reuse/deployment


## Part 3 — Common Pitfalls Checklist

- [ ] **Data leakage** — encoding/scaling statistics computed on the full dataset instead of train only
- [ ] **Non-idempotent cleaning** — rerunning a cell breaks because it assumes uncleaned input
- [ ] **Hidden strings** — values like `"Not Available"` or `"No"` silently breaking `.astype(int)`
- [ ] **Manufactured multicollinearity** — an engineered composite feature kept alongside its raw inputs, poisoning linear-model coefficients
- [ ] **Impurity-importance overtrust** — trusting `feature_importances_` alone for high-cardinality encoded features without a permutation-importance cross-check
- [ ] **Overreliance on R²** — a high R² reported without MAE/RMSE alongside it, hiding real error magnitude
- [ ] **Untracked randomness** — missing `random_state`, making results non-reproducible run to run
- [ ] **Suspiciously high test performance** — treated as a win instead of a leakage red flag worth re-checking
- [ ] **Test-set reuse for tuning** — hyperparameters tuned directly against the final test set instead of via cross-validation on training data
- [ ] **Outlier removal by default** — genuine extreme values removed just because they're statistically rare, without checking whether they're real
- [ ] **Model without its preprocessing** — a persisted model saved without the encoding maps/column order needed to actually use it on new data


## Part 4 — Definition-of-Done Checklist for the Whole Project

- [ ] Every raw column is either numeric, properly encoded, or intentionally dropped with a stated reason
- [ ] No `NaN` values remain anywhere in the final training/test feature matrices
- [ ] At least 4 distinct models trained and fairly compared on the identical test split
- [ ] Final model selected with a written justification (not just "best number")
- [ ] Cross-validated performance estimate reported alongside the single-split test metrics
- [ ] Two independent feature-importance methods agree on (most of) the top drivers
- [ ] Model and all preprocessing artifacts are persisted and reloadable
- [ ] An inference function exists that takes raw, uncleaned input and returns a prediction
- [ ] A plain-language summary exists that a non-technical stakeholder could act on
- [ ] At least one limitation and one next step are explicitly named
