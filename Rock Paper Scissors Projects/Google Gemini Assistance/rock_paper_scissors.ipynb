# Rock Paper Scissors Game
An interactive implementation of Rock Paper Scissors with refactored logic.

```python
import random
import time

MOVES = {'R': 'ROCK', 'P': 'PAPER', 'S': 'SCISSORS'}
BEATS_MAP = {'ROCK': 'SCISSORS', 'PAPER': 'ROCK', 'SCISSORS': 'PAPER'}

def play_notebook_rps():
    wins = losses = ties = 0
    print("=== ROCK, PAPER, SCISSORS ===")
    
    while True:
        print(f"\nScoreboard -> Wins: {wins} | Losses: {losses} | Ties: {ties}")
        choice = input("Enter (R)ock, (P)aper, (S)cissors, or (Q)uit: ").strip().upper()
        
        if choice == 'Q':
            print(f"\nGame Over! Final Score -> Wins: {wins} | Losses: {losses} | Ties: {ties}")
            break
        if choice not in MOVES:
            print("Invalid input! Try again.")
            continue
            
        player_move = MOVES[choice]
        print(f"\n{player_move} versus...")
        
        for i in range(1, 4):
            print(f"{i}...")
            time.sleep(0.2)
            
        computer_move = random.choice(list(MOVES.values()))
        print(f"{computer_move}!\n")
        
        if player_move == computer_move:
            print("It's a tie!")
            ties += 1
        elif BEATS_MAP[player_move] == computer_move:
            print("🎉 You win!")
            wins += 1
        else:
            print("💻 Computer wins!")
            losses += 1

play_notebook_rps()
```
