### Solution 11 – Sorting Descending & Top-K

`np.sort` always returns ascending order; reverse it with `[::-1]`.  
The slice `[:3]` keeps the first three elements (top-3).

````
import numpy as np
salaries = np.array([45000, 38000, 52000, 47000, 43000, 39000])

sorted_salaries = np.sort(salaries)[::-1]
print(sorted_salaries[:3])   # [52000 47000 45000]
````

### Solution 12 – Column-wise Descending Sort

- `np.sort(..., axis=0)` sorts each column ascending.  
- `[::-1]` then reverses the rows, turning every column into descending order.  
- `axis=None` flattens the array before sorting, giving a global ascending list.

````
import numpy as np
exam_scores = np.array([[75, 82, 90],
                        [92, 88, 78],
                        [60, 70, 85]])

top_scores_subject = np.sort(exam_scores, axis=0)[::-1]
sorted_scores      = np.sort(exam_scores, axis=None)

print(top_scores_subject)
print(sorted_scores)
````

### Solution 17 – Vectorised Percentage Growth

All arithmetic is element-wise.  
Parentheses guarantee the correct order of operations.  
`np.round(..., 2)` rounds every element.

````
import numpy as np
sales_data_2021 = np.array([[350, 420, 380, 410],
                            [270, 320, 290, 310]])
sales_data_2022 = np.array([[360, 440, 390, 430],
                            [280, 330, 300, 320]])

revenue_growth = (sales_data_2022 - sales_data_2021) / sales_data_2021 * 100
revenue_growth = np.round(revenue_growth, 2)
print(revenue_growth)
````

### Solution 18 – Dot Product / Weighted Scores

`np.dot(matrix, vector)` (or the `@` operator) computes the matrix-vector product,  
i.e. a weighted sum for every row.

````
import numpy as np
exams_scores  = np.array([[100, 82, 95],
                          [56,  70, 90],
                          [45,  98, 66]])
coefficients  = np.array([0.5, 0.3, 0.2])

final_scores = np.dot(exams_scores, coefficients)
# equivalently: final_scores = exams_scores @ coefficients
print(final_scores)
````

### Solution 19 – Broadcasting & Centering

`exam_scores.mean(axis=0)` produces a 1-D array of column means (shape `(3,)`).  
Subtracting it from the 2-D matrix triggers broadcasting: the 1-D array is applied to every row.

````
import numpy as np
exam_scores = np.array([[100, 82, 95],
                        [56,  70, 90],
                        [45,  98, 66]], dtype=float)

subject_means   = exam_scores.mean(axis=0)
centered_scores = exam_scores - subject_means

print("Subject means:", subject_means)
print("Centered matrix:\\n", centered_scores.round(2))
````

### Solution 20 – Mini Real-World Pipeline

1. `.copy()` protects the original data.  
2. `np.sort(...)[::-1]` gives descending order.  
3. `[:3]` extracts the top-3.  
4. `.mean()` computes their average.  
5. Boolean mask on the original array + `.sum()` counts how many salaries exceed that average.

````
import numpy as np
salaries = np.array([45000, 38000, 52000, 47000, 43000, 39000, 61000, 48000])

sorted_desc   = np.sort(salaries.copy())[::-1]
top3          = sorted_desc[:3]
top3_average  = top3.mean()
above_avg_count = (salaries > top3_average).sum()

print("Top-3 average :", top3_average)
print("Employees above that average:", above_avg_count)
````
