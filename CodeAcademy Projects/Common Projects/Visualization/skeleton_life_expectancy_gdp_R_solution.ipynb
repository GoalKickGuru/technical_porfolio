## 1. Theory: Quantiles in R

**Quantiles** divide data into groups of equal size.

- `quantile(x, probs = c(0.25, 0.5, 0.75))` computes Q1, Median, and Q3.
- R’s `quantile()` function supports 9 different algorithms (`type = 1` to `type = 9`). The default usually works well.
- Quartiles are very useful for understanding spread and comparing subgroups without being heavily influenced by outliers.

In this project we will:
1. Examine the overall distribution of Life Expectancy.
2. Split countries into Low vs High GDP groups.
3. Compare life expectancy between these groups using quantiles and plots.


## 2. Loading and Inspecting the Data

````r
library(tidyverse)
library(ggplot2)
library(patchwork)

# TODO: Load the data
data <- read_csv("/home/workdir/attachments/country_data.csv")

# TODO: Inspect the data
head(data)
glimpse(data)
summary(data)
````

## 3. Computing Quartiles of Life Expectancy

````r
# TODO: Get Life Expectancy as a vector
life_exp <- data$`Life Expectancy`

# TODO: Compute quartiles
life_quartiles <- quantile(life_exp, probs = c(0.25, 0.5, 0.75))
print(life_quartiles)

cat("\nQ1 (25%):", life_quartiles[1], "years\n")
cat("Median (50%):", life_quartiles[2], "years\n")
cat("Q3 (75%):", life_quartiles[3], "years\n")
````

## 4. Visualizing the Overall Distribution

````r
ggplot(data, aes(x = `Life Expectancy`)) +
  geom_histogram(bins = 20, fill = "#4FC3F7", color = "#0277BD", alpha = 0.8) +
  geom_vline(xintercept = life_quartiles, color = "red", linetype = "dashed", linewidth = 1) +
  labs(title = "Distribution of Life Expectancy Across Countries",
       x = "Life Expectancy (years)", y = "Number of Countries") +
  theme_minimal()
````

## 5. Splitting Countries by GDP

````r
# TODO: Get GDP column
gdp <- data$GDP

# TODO: Find median GDP
median_gdp <- quantile(gdp, 0.5)
cat("Median GDP:", median_gdp, "\n")

# Split into low and high GDP
low_gdp <- data %>% filter(GDP <= median_gdp)
high_gdp <- data %>% filter(GDP > median_gdp)

cat("Low GDP countries:", nrow(low_gdp), "\n")
cat("High GDP countries:", nrow(high_gdp), "\n")
````

## 6. Comparing Life Expectancy: Low vs High GDP

````r
# TODO: Quartiles for low GDP countries
low_quartiles <- quantile(low_gdp$`Life Expectancy`, c(0.25, 0.5, 0.75))
print("Low GDP - Life Expectancy Quartiles:")
print(low_quartiles)

# TODO: Quartiles for high GDP countries
high_quartiles <- quantile(high_gdp$`Life Expectancy`, c(0.25, 0.5, 0.75))
print("\nHigh GDP - Life Expectancy Quartiles:")
print(high_quartiles)

cat("\nKey Finding:\n")
cat("Median life expectancy (Low GDP):", low_quartiles[2], "years\n")
cat("Median life expectancy (High GDP):", high_quartiles[2], "years\n")
````

## 7. Visual Comparison

````r
ggplot() +
  geom_histogram(data = high_gdp, aes(x = `Life Expectancy`, fill = "High GDP"), 
                 bins = 15, alpha = 0.6, position = "identity") +
  geom_histogram(data = low_gdp, aes(x = `Life Expectancy`, fill = "Low GDP"), 
                 bins = 15, alpha = 0.6, position = "identity") +
  scale_fill_manual(values = c("High GDP" = "#81C784", "Low GDP" = "#E57373")) +
  geom_vline(xintercept = low_quartiles[2], color = "#C62828", linetype = "dashed", linewidth = 1.2) +
  geom_vline(xintercept = high_quartiles[2], color = "#2E7D32", linetype = "dashed", linewidth = 1.2) +
  labs(title = "Life Expectancy Distribution: High vs Low GDP Countries",
       x = "Life Expectancy (years)", y = "Number of Countries", fill = "Group") +
  theme_minimal()
````

## 8. 🎮 Simulation: Try Different GDP Thresholds

Modify the percentile below to see how results change.


````r
# ============== SIMULATION PARAMETERS ==============
GDP_THRESHOLD_PERCENTILE <- 0.5   # Try 0.25 or 0.75
# ========================================================

threshold <- quantile(data$GDP, GDP_THRESHOLD_PERCENTILE)
cat("GDP Threshold (", GDP_THRESHOLD_PERCENTILE*100, "th percentile):", threshold, "\n")

low_group <- data %>% filter(GDP <= threshold)
high_group <- data %>% filter(GDP > threshold)

low_q <- quantile(low_group$`Life Expectancy`, c(0.25, 0.5, 0.75))
high_q <- quantile(high_group$`Life Expectancy`, c(0.25, 0.5, 0.75))

cat("Low group median life expectancy :", low_q[2], "years (n =", nrow(low_group), ")\n")
cat("High group median life expectancy:", high_q[2], "years (n =", nrow(high_group), ")\n")
````

## 10. Advanced & ML Uses in R

- **Robust Scaling**: Use median and IQR instead of mean and sd.
- **Feature Engineering**: Create a binary `High_GDP` column using `ntile()` or `quantile()`.
- **Quantile Regression**: Use the `quantreg` package to model different percentiles of life expectancy.
- **Log Transformation**: GDP is heavily right-skewed. Taking `log(GDP)` often improves linear models.

These techniques are very common when working with economic and health data in R.
