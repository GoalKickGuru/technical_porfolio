# Data Visualization Practice: Audience-Aware Techniques

**Skeleton Notebook for Hands-on Learning**

This notebook provides step-by-step guided exercises to master key visualization techniques in Python using Matplotlib, Pandas, and Seaborn.

### Why This Matters (from the readings)
- **Data Literacy**: Highly data-literate audiences (engineers, analysts) can handle error bars, normalized histograms, and statistical concepts. Less data-literate audiences (e.g., from literature/fine arts backgrounds) benefit from simpler bars, lines, and clear explanations without overload.
- **Subject Knowledge**: Experts skip intros and prefer precise conventions (e.g., waterfall for finance); novices need highlighted insights and plain language.
- **Time Span**: C-level and busy readers need quick, high-signal visuals. Long-form readers can absorb more detail.

Your goal: Practice creating visualizations that are **appropriate for the intended audience** while mastering the technical methods listed.

4. Stacked bar graphs for composition/comparisons
5. Side-by-side (grouped) bar graphs
6. Formatted pie charts for proportional data
7. Frequency analysis with histograms (single + multiple)
8. Normalized histograms for fair distribution comparison

## Flowchart: Audience-Aware Visualization Workflow

This flowchart summarizes the decision process for creating effective visualizations, integrating concepts from the provided documents on audience analysis and data analysis report structure.

```mermaid
flowchart TD
    Start[Start: Define Purpose & Primary Audience] --> Audience[Analyze Audience]
    Audience --> DL{Data Literacy Level?}
    DL -->|High (Experts/Technicians)| Advanced[Advanced Viz: Error bars, Normalized hists, Detailed stats]
    DL -->|Low (Executives/Nonspecialists)| Simple[Simple Viz: Basic bars, Pies, Clear labels, Minimal jargon]
    Audience --> SK{Subject Knowledge?}
    SK -->|High| SkipIntro[Skip basic explanations, use domain conventions]
    SK -->|Low| Explain[Provide context, highlight insights, explain 'good/bad']
    Audience --> TS{Time Span Available?}
    TS -->|Short (C-level, skimmers)| Quick[Large fonts, 1 key insight per viz, annotations]
    TS -->|Long| Detail[More layers, multiple related viz, appendix details]
    Simple & Advanced --> Implement[Implement with Matplotlib / Pandas / Seaborn]
    SkipIntro & Explain --> Implement
    Quick & Detail --> Implement
    Implement --> Validate[Validate: Does this help THIS audience reach the insight quickly & accurately?]
    Validate --> Report[Place in Data Analysis Report: Body for main points, Appendix for technical depth]
    Report --> End[Deliver clear, audience-resonant message]
```

**Key Insight from readings**: A good visualization is not just technically correct — it is *adapted* to the reader's background, needs, and constraints so the message resonates and is understood without friction.

## 8. Normalized Histograms (Density) for Fair Group Comparison

### Learning Objective
Use `density=True` (or `stat='density'` in seaborn) so the area under each histogram equals 1. This allows fair visual comparison of **shape** when sample sizes differ.

### Audience Consideration
Critical when comparing groups of unequal size (e.g., 180 novices vs 140 experts). Raw counts would misleadingly make the larger group look more 'important'. Normalized view focuses on the *distribution shape* — exactly what analysts need.

### Step-by-Step
1. Add `density=True` to `plt.hist()`.
2. Or use `seaborn.histplot(..., stat='density', common_norm=False)`.
3. Compare the shapes: Where do the peaks sit? How spread out are they?
