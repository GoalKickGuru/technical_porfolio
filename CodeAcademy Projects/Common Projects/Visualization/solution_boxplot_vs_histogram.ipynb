## 1. Theory: Boxplots and Histograms

### Boxplot Components
- **Box**: Shows the Interquartile Range (IQR = Q3 - Q1)
  - Bottom of box = Q1 (25th percentile)
  - Line inside box = **Median** (50th percentile)
  - Top of box = Q3 (75th percentile)
- **Whiskers**: Extend to the smallest/largest values within **1.5 × IQR** from Q1/Q3
- **Outliers**: Points beyond the whiskers (shown as individual dots)

### When to Use Each Plot

**Boxplots are excellent for**:
- Comparing the spread and central tendency of **multiple groups** side-by-side
- Quickly identifying **outliers**
- Showing **median** and **IQR** (robust to skewness)
- Compact summary when you have many groups

**Histograms are excellent for**:
- Seeing the **full shape** of the distribution (unimodal, bimodal, skewed, etc.)
- Understanding **frequency** and density
- Detecting **multimodality** or gaps in data

**Trade-off**: Boxplots lose information about the exact shape and number of modes. Histograms can be harder to compare across many groups.


## 2. Generating the Five Datasets

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
np.random.seed(1)

# Generate five different distributions
a = np.random.normal(1, 1, 1000)      # Narrow, centered near 1
b = np.random.normal(0, 3, 1000)      # Wide spread
c = np.random.normal(2, 1.5, 1000)    # Medium spread, slightly right
d = np.random.normal(-4, 5, 1000)     # Very wide, left-centered
e = np.random.normal(5, 2, 1000)      # Right-centered, moderate spread

datasets = {'A': a, 'B': b, 'C': c, 'D': d, 'E': e}
df = pd.DataFrame(datasets)

print("Dataset shapes and basic stats:")
print(df.describe().round(2))
````

## 3. Creating the Boxplot

````python
plt.figure(figsize=(10, 6))
plt.boxplot([a, b, c, d, e], labels=['A', 'B', 'C', 'D', 'E'])
plt.title("Boxplot Comparison of Five Distributions")
plt.ylabel("Value")
plt.xlabel("Dataset")
plt.show()

print("Boxplot shows median, IQR, and outliers clearly for comparison across groups.")
````

## 4. Creating Individual Histograms

````python
fig, axes = plt.subplots(5, 1, figsize=(8, 10), sharex=True)

for i, (name, data) in enumerate(datasets.items()):
    axes[i].hist(data, bins=30, color="#4FC3F7", edgecolor="#0277BD", alpha=0.8)
    axes[i].set_title(f"Dataset {name}")
    axes[i].set_xlim([-20, 20])

plt.tight_layout()
plt.show()

print("Histograms reveal the full shape (spread, skewness, modality) of each distribution.")
````

## 5. Detailed Statistical Analysis & Comparison

Let's calculate key statistics for each dataset to support our visual interpretation.


````python
def get_stats(name, data):
    q1, median, q3 = np.percentile(data, [25, 50, 75])
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = data[(data < lower_bound) | (data > upper_bound)]
    return {
        'Dataset': name,
        'Mean': np.mean(data),
        'Median': median,
        'Std Dev': np.std(data),
        'IQR': iqr,
        'Outliers Count': len(outliers),
        'Min': np.min(data),
        'Max': np.max(data)
    }

stats_list = [get_stats(name, data) for name, data in datasets.items()]
stats_df = pd.DataFrame(stats_list)
print(stats_df.round(2).to_string(index=False))
````

## 6. Advantages & Disadvantages (Analysis)

**Boxplots**:
- **Advantages**: Excellent for comparing multiple groups side-by-side. Clearly shows median, IQR, and outliers. Compact.
- **Disadvantages**: Hides the shape of the distribution (cannot see if it's bimodal or skewed in detail). Loses information about frequency.

**Histograms**:
- **Advantages**: Shows the full shape and modality of the distribution. Easy to understand frequency.
- **Disadvantages**: Harder to compare many groups at once. Choice of bin size can affect interpretation. Outliers can be less obvious unless the range is adjusted.

**Conclusion from our 5 datasets**:
- Dataset **B** and **D** have much larger spread (visible in both plots).
- Dataset **A** is the most concentrated.
- Boxplot makes it easy to see that **D** has many outliers on both sides due to its large standard deviation.
- Histograms reveal that all distributions are roughly symmetric (normal), which wouldn't be as obvious from boxplots alone.


## 7. 🎮 Simulation: Modify Distributions and Observe Changes

Change the parameters below and re-run the cell to see how the visualizations and statistics update.


````python
# ============== SIMULATION PARAMETERS (MODIFY THESE) ==============
DATASET_TO_MODIFY = 'B'     # 'A', 'B', 'C', 'D', or 'E'
NEW_MEAN = 0
NEW_STD = 4
N_SAMPLES = 1000
# =====================================================================

np.random.seed(42)
modified_data = np.random.normal(NEW_MEAN, NEW_STD, N_SAMPLES)

print(f"Modified Dataset {DATASET_TO_MODIFY}: Mean={NEW_MEAN}, Std={NEW_STD}")

# Quick stats
q1, med, q3 = np.percentile(modified_data, [25, 50, 75])
print(f"New Q1={q1:.2f}, Median={med:.2f}, Q3={q3:.2f}, IQR={q3-q1:.2f}")

# Re-plot boxplot and histogram side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.boxplot(modified_data)
ax1.set_title(f"Boxplot - Modified {DATASET_TO_MODIFY}")
ax2.hist(modified_data, bins=30, color="#81C784", edgecolor="#2E7D32")
ax2.set_title(f"Histogram - Modified {DATASET_TO_MODIFY}")
plt.tight_layout()
plt.show()
````

## 8. Alternate & Improved Visualizations with Seaborn

````python
# Melt dataframe for seaborn
df_melted = df.melt(var_name='Dataset', value_name='Value')

# Seaborn boxplot (cleaner)
plt.figure(figsize=(10,6))
sns.boxplot(data=df_melted, x='Dataset', y='Value', palette="Set2")
plt.title("Seaborn Boxplot - Much Cleaner")
plt.show()

# Violin plot (combines boxplot + density)
plt.figure(figsize=(10,6))
sns.violinplot(data=df_melted, x='Dataset', y='Value', palette="Set2", inner="quartile")
plt.title("Violin Plot (Boxplot + Distribution Shape)")
plt.show()

# Side-by-side boxplot + swarmplot for one dataset (e.g. Dataset B)
plt.figure(figsize=(8,5))
sns.boxplot(y=b, color="#4FC3F7")
sns.swarmplot(y=b, color="black", alpha=0.4, size=2)
plt.title("Boxplot + Swarmplot for Dataset B (shows individual points)")
plt.show()
````
