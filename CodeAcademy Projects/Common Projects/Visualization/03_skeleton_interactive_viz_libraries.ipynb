# Interactive Visualization Libraries in Jupyter

**Skeleton Notebook — Guided Practice with Plotly**

In the previous notebooks we mastered **static** visualizations (bars, error bars, histograms, pies) using Matplotlib, Pandas, and Seaborn. Now we investigate **interactive** libraries that turn those same charts into explorable experiences.

### Why Interactive Visualizations Matter (Audience Perspective)
From the readings on audience analysis:
- **Executives & short time-span readers**: Interactive filters and hover details let them quickly focus on what matters to *them* without wading through everything.
- **Data-literate analysts & technical supervisors**: Hover tooltips, zooming, linked selections, and drill-down give them the depth they crave while keeping the overview clean.
- **Mixed or unknown audiences**: One interactive dashboard can serve multiple user types — each person explores at their own level.

Interactive viz also fits beautifully into modern **data analysis reports** (Jupyter, HTML exports, internal portals) because the reader becomes an active participant rather than a passive consumer.

### Libraries We Will Investigate
1. **Plotly** (primary focus) — Most popular for Jupyter. `plotly.express` gives Seaborn-like speed with built-in interactivity (hover, zoom, pan, legend toggle, export). Excellent error bar and distribution support.
2. **ipywidgets** (mentioned for completeness) — Adds sliders/dropdowns to any plot (including matplotlib). *Note: may require separate install in your environment.*
3. **Altair & Bokeh** (overview only) — Declarative alternatives. Concepts transfer across libraries.

**Goal of this notebook**: Learn to recreate and enhance the visualizations from the previous notebooks as **interactive** versions using Plotly, while considering audience needs.


## Flowchart: Choosing Between Static and Interactive Visualizations

```mermaid
flowchart TD
    Start[Define Audience + Key Questions from Data Analysis Report] --> Needs{What does the audience need?}
    Needs -->|Fixed story, print/PDF, executive summary| Static[Use Static Viz<br/>Matplotlib / Seaborn / Pandas plot<br/>Controlled, reproducible, publication-ready]
    Needs -->|Exploration, filtering, hover details, drill-down, self-service| Interactive[Use Interactive Viz<br/>Plotly (recommended in Jupyter)]
    Interactive --> Library{Which library?}
    Library -->|Fast, expressive, great Jupyter integration| Plotly[Plotly Express first → graph_objects for fine control]
    Library -->|Declarative grammar + rich interactivity| Altair[Altair (if available)]
    Library -->|Highly customizable dashboards| Bokeh[ Bokeh or Panel/Holoviews]
    Plotly --> Enhance[Add hover templates, error info, color encoding, facets, marginals]
    Enhance --> Test[Test with real audience: Can they discover insights themselves?]
    Test --> Deliver[Embed in Jupyter Notebook, HTML report, or lightweight dashboard]
    Static --> Deliver
```

**Key takeaway**: Interactive visualizations shine when your audience has varying questions or needs to explore uncertainty/details on their own. They complement (do not always replace) the static figures you place in the Body of a formal data analysis report.
