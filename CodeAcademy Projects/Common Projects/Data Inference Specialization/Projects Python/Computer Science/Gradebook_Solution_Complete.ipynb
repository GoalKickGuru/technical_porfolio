### Complete Working Code

```python
subjects = ["physics", "calculus", "poetry", "history"]
grades = [98, 97, 85, 88]

# Step 3: Manual 2D list
gradebook = [["physics", 98], ["calculus", 97], ["poetry", 85], ["history", 88]]
print("Initial gradebook:")
print(gradebook)

# Step 5: Append Computer Science
gradebook.append(["computer science", 100])

# Step 6: Append Visual Arts
gradebook.append(["visual arts", 93])
print("\nAfter adding new classes:")
print(gradebook)

# Step 7: Modify Visual Arts (+5)
for subject in gradebook:
    if subject[0] == "visual arts":
        subject[1] += 5
        break
print("\nAfter +5 to visual arts:")
print(gradebook)

# Step 8-9: Poetry to Pass/Fail
for subject in gradebook:
    if subject[0] == "poetry":
        subject.remove(subject[1])   # remove numeric grade
        subject.append("Pass")       # add Pass
        break
print("\nAfter changing poetry to Pass/Fail:")
print(gradebook)

# Step 10: Combine with last semester
last_semester_gradebook = [["biology", 92], ["algebra", 95], ["english", 88]]
full_gradebook = last_semester_gradebook + gradebook
print("\nFull Gradebook (both semesters):")
print(full_gradebook)
```

### Expanded: Analysis Functions

```python
def calculate_average(gradebook):
    numeric_grades = [entry[1] for entry in gradebook if isinstance(entry[1], (int, float))]
    if not numeric_grades:
        return 0
    return round(sum(numeric_grades) / len(numeric_grades), 2)

def find_highest_lowest(gradebook):
    numeric = [(entry[0], entry[1]) for entry in gradebook if isinstance(entry[1], (int, float))]
    if not numeric:
        return None, None
    highest = max(numeric, key=lambda x: x[1])
    lowest = min(numeric, key=lambda x: x[1])
    return highest, lowest

print("\n=== Analysis ===")
print("Average grade:", calculate_average(gradebook))
high, low = find_highest_lowest(gradebook)
print("Highest:", high)
print("Lowest:", low)
```

### Simulation Section - Modify Values!

```python
import random

print("\n=== SIMULATION: Random Grade Scenarios ===")
num_simulations = 5          # MODIFY this
possible_subjects = ["math", "science", "art", "music", "PE"]

for i in range(num_simulations):
    sim_gradebook = []
    for subj in random.sample(possible_subjects, 4):
        score = random.randint(70, 100)
        sim_gradebook.append([subj, score])
    avg = calculate_average(sim_gradebook)
    high, low = find_highest_lowest(sim_gradebook)
    print(f"Simulation {i+1}: Avg={avg}, Highest={high}, Lowest={low}")
```

### Alternate 1: Dictionary Version (Easier lookups)

```python
gradebook_dict = {
    "physics": 98,
    "calculus": 97,
    "poetry": "Pass",
    "history": 88,
    "computer science": 100,
    "visual arts": 98   # already +5
}

def dict_average(g_dict):
    nums = [v for v in g_dict.values() if isinstance(v, (int, float))]
    return round(sum(nums)/len(nums), 2) if nums else 0

print("\nDictionary version average:", dict_average(gradebook_dict))
```

### Alternate 2: Using zip() to build gradebook

```python
# Build from two separate lists
subjects2 = ["physics", "calculus", "poetry", "history"]
grades2 = [98, 97, 85, 88]
gradebook_from_zip = list(zip(subjects2, grades2))
print("\nBuilt with zip():", gradebook_from_zip)
```
