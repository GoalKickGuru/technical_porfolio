### Complete Working Code


````python
import random

# Variables
name = "Juanazo"          # Change this
question = "Will I succeed with data visualization projects?"  # Change this
answer = ""

# Generate random number (expanded to 12)
random_number = random.randint(1, 12)
print("(Debug) Random number generated:", random_number)

# Control Flow - if/elif/else
if random_number == 1:
    answer = "Yes - definitely"
elif random_number == 2:
    answer = "It is decidedly so"
elif random_number == 3:
    answer = "Without a doubt"
elif random_number == 4:
    answer = "Reply hazy, try again"
elif random_number == 5:
    answer = "Ask again later"
elif random_number == 6:
    answer = "Better not tell you now"
elif random_number == 7:
    answer = "My sources say no"
elif random_number == 8:
    answer = "Outlook not so good"
elif random_number == 9:
    answer = "Very doubtful"
elif random_number == 10:
    answer = "Signs point to yes"   # New
elif random_number == 11:
    answer = "Cannot predict now" # New
elif random_number == 12:
    answer = "Concentrate and ask again" # New
else:
    answer = "Error - something went wrong"

# Formatted Output with name handling
if not name.strip():
    print("Question:", question)
else:
    print(name + " asks: " + question)
print("Magic 8-Ball's answer: " + answer)
````

### Alternate Code 1: Using List + Indexing (Cleaner)


````python
# Alternate using list
fortunes = [
    "Yes - definitely", "It is decidedly so", "Without a doubt",
    "Reply hazy, try again", "Ask again later", "Better not tell you now",
    "My sources say no", "Outlook not so good", "Very doubtful",
    "Signs point to yes", "Cannot predict now", "Concentrate and ask again"
]

random_number = random.randint(0, len(fortunes) - 1)
answer = fortunes[random_number]

if not name.strip():
    print("Question:", question)
else:
    print(name + " asks: " + question)
print("Magic 8-Ball's answer: " + answer)
````

### Alternate Code 2: Using Dictionary


````python
# Dictionary version
fortune_dict = {
    1: "Yes - definitely", 2: "It is decidedly so", 3: "Without a doubt",
    4: "Reply hazy, try again", 5: "Ask again later", 6: "Better not tell you now",
    7: "My sources say no", 8: "Outlook not so good", 9: "Very doubtful",
    10: "Signs point to yes", 11: "Cannot predict now", 12: "Concentrate and ask again"
}

random_number = random.randint(1, 12)
answer = fortune_dict.get(random_number, "Error")

if not name.strip():
    print("Question:", question)
else:
    print(name + " asks: " + question)
print("Magic 8-Ball's answer: " + answer)
````
