### Complete Working Program

````python
import random  # for simulation variety if needed

def calculate_shipping(weight):
    # Ground Shipping
    if weight <= 2:
        cost_ground = weight * 1.5 + 20
    elif weight <= 6:
        cost_ground = weight * 3.00 + 20
    elif weight <= 10:
        cost_ground = weight * 4.00 + 20
    else:
        cost_ground = weight * 4.75 + 20

    # Ground Shipping Premium
    cost_premium = 125.00

    # Drone Shipping (3x ground rates, no flat fee)
    if weight <= 2:
        cost_drone = weight * 4.5
    elif weight <= 6:
        cost_drone = weight * 9.00
    elif weight <= 10:
        cost_drone = weight * 12.00
    else:
        cost_drone = weight * 14.25

    # Find cheapest
    if cost_ground <= cost_premium and cost_ground <= cost_drone:
        method = "Ground Shipping"
        cost = cost_ground
    elif cost_premium <= cost_drone:
        method = "Ground Shipping Premium"
        cost = cost_premium
    else:
        method = "Drone Shipping"
        cost = cost_drone

    return method, round(cost, 2)

# Example tests from instructions
print("=== Example Tests ===")
method, cost = calculate_shipping(8.4)
print(f"8.4 lb → Cheapest: {method} ${cost}")  # Expected: Ground $53.6

method, cost = calculate_shipping(1.5)
print(f"1.5 lb → Cheapest: {method} ${cost}")  # Expected: Drone $6.75

method, cost = calculate_shipping(4.8)
print(f"4.8 lb → Cheapest: {method} ${cost}")

method, cost = calculate_shipping(41.5)
print(f"41.5 lb → Cheapest: {method} ${cost}")
````

### Alternate Code 1: Using Dictionaries for Rates (Cleaner & Scalable)

````python
def calculate_shipping_dict(weight):
    ground_rates = {2: 1.5, 6: 3.0, 10: 4.0, float('inf'): 4.75}
    drone_rates = {2: 4.5, 6: 9.0, 10: 12.0, float('inf'): 14.25}

    # Ground
    for max_w, rate in ground_rates.items():
        if weight <= max_w:
            cost_ground = weight * rate + 20
            break

    cost_premium = 125.0

    # Drone
    for max_w, rate in drone_rates.items():
        if weight <= max_w:
            cost_drone = weight * rate
            break

    # Compare (same logic as before)
    costs = {
        "Ground Shipping": cost_ground,
        "Ground Shipping Premium": cost_premium,
        "Drone Shipping": cost_drone
    }
    method = min(costs, key=costs.get)
    return method, round(costs[method], 2)

# Test
print(calculate_shipping_dict(8.4))
````

### Alternate Code 2: List of Tuples Approach

````python
# Even more compact using list of (max_weight, ground_rate, drone_rate)
shipping_tiers = [
    (2, 1.5, 4.5),
    (6, 3.0, 9.0),
    (10, 4.0, 12.0),
    (float('inf'), 4.75, 14.25)
]

def calculate_shipping_list(weight):
    for max_w, g_rate, d_rate in shipping_tiers:
        if weight <= max_w:
            cost_ground = weight * g_rate + 20
            cost_drone = weight * d_rate
            break
    cost_premium = 125.0

    costs = {"Ground": cost_ground, "Premium": cost_premium, "Drone": cost_drone}
    method = min(costs, key=costs.get)
    return method, round(costs[method], 2)

print(calculate_shipping_list(41.5))
````
