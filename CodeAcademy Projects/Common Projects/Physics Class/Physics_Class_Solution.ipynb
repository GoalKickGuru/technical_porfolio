## 0. Setup — Given Values

````python
train_mass = 22680          # kg
train_acceleration = 10     # m/s²
train_distance = 100        # m
bomb_mass = 1               # kg

print("Setup complete. Constants ready.")
print(f"train_mass = {train_mass} kg")
print(f"train_acceleration = {train_acceleration} m/s²")
print(f"train_distance = {train_distance} m")
print(f"bomb_mass = {bomb_mass} kg")
````

## 1. Turn Up the Temperature

### Task 1 & 2 — f_to_c and test with 100 °F


````python
def f_to_c(f_temp):
    c_temp = (f_temp - 32) * 5 / 9
    return c_temp

f100_in_celsius = f_to_c(100)
print(f"100 °F = {f100_in_celsius} °C")
print(f"Self-check 32 °F → {f_to_c(32)} °C (should be 0)")
````

### Task 3 & 4 — c_to_f and test with 0 °C


````python
def c_to_f(c_temp):
    f_temp = c_temp * (9 / 5) + 32
    return f_temp

c0_in_fahrenheit = c_to_f(0)
print(f"0 °C = {c0_in_fahrenheit} °F")
print(f"Self-check 100 °C → {c_to_f(100)} °F (should be 212)")
````

## 2. Use the Force

### Task 5 — get_force


````python
def get_force(mass, acceleration):
    return mass * acceleration
````

### Task 6 & 7 — train_force and sentence


````python
train_force = get_force(train_mass, train_acceleration)
print(train_force)
print(f"The GE train supplies {train_force} Newtons of force.")
````

### Task 8 — get_energy (with default c)


````python
def get_energy(mass, c=3*10**8):
    return mass * c ** 2
````

### Task 9 & 10 — bomb_energy and sentence


````python
bomb_energy = get_energy(bomb_mass)
print(bomb_energy)
print(f"A 1kg bomb supplies {bomb_energy} Joules.")
````

## 3. Do the Work

### Task 11 — get_work (composition)


````python
def get_work(mass, acceleration, distance):
    force = get_force(mass, acceleration)
    return force * distance
````

### Task 12 & 13 — train_work and final sentence


````python
train_work = get_work(train_mass, train_acceleration, train_distance)
print(train_work)
print(f"The GE train does {train_work} Joules of work over {train_distance} meters.")
````

### Core results at a glance


````python
print("=" * 50)
print("CORE RESULTS SUMMARY")
print("=" * 50)
print(f"100 °F          → {f100_in_celsius:.2f} °C")
print(f"0 °C            → {c0_in_fahrenheit:.1f} °F")
print(f"Train force     → {train_force:,} N")
print(f"Bomb energy     → {bomb_energy:.3e} J")
print(f"Train work      → {train_work:,} J  over {train_distance} m")
print("=" * 50)
````

## 4. Alternate Implementations

### 4a. Lambda versions


````python
f_to_c_lambda = lambda f_temp: (f_temp - 32) * 5 / 9
c_to_f_lambda = lambda c_temp: c_temp * (9 / 5) + 32
get_force_lambda = lambda m, a: m * a
get_energy_lambda = lambda m, c=3e8: m * c ** 2
get_work_lambda = lambda m, a, d: get_force_lambda(m, a) * d

print("Lambda check 100 °F →", f_to_c_lambda(100))
print("Lambda train force →", get_force_lambda(train_mass, train_acceleration))
print("Lambda train work  →", get_work_lambda(train_mass, train_acceleration, train_distance))
````

### 4b. One-liner get_work (no intermediate variable)


````python
def get_work_one_liner(mass, acceleration, distance):
    return get_force(mass, acceleration) * distance

print("One-liner work:", get_work_one_liner(train_mass, train_acceleration, train_distance))
````

### 4c. Using ** operator explicitly and scientific notation


````python
def get_energy_sci(mass, c=3e8):
    # Same physics, different literal style
    return mass * (c ** 2)

print(get_energy_sci(1))
````

## 5. More Practice — Solutions

### 5.1 Kelvin converters


````python
def c_to_k(c_temp):
    return c_temp + 273.15

def k_to_c(k_temp):
    return k_temp - 273.15

def f_to_k(f_temp):
    # reuse existing conversion
    return c_to_k(f_to_c(f_temp))

print(f"0 °C  → {c_to_k(0)} K")
print(f"273.15 K → {k_to_c(273.15)} °C")
print(f"32 °F → {f_to_k(32)} K")
print(f"100 °F → {f_to_k(100):.2f} K")
````

### 5.2 Kinetic energy & Power


````python
def get_kinetic_energy(mass, velocity):
    return 0.5 * mass * velocity ** 2

def get_power(work, time):
    return work / time

# Example numbers
ke = get_kinetic_energy(train_mass, 20)          # 20 m/s ≈ 72 km/h
power_10s = get_power(train_work, 10)            # if the work is performed in 10 s

print(f"Kinetic energy of train at 20 m/s : {ke:,.0f} J")
print(f"Average power if work done in 10 s: {power_10s:,.0f} W  ({power_10s/1e6:.2f} MW)")
````

### 5.3 Gravitational force


````python
G = 6.67430e-11   # m³ kg⁻¹ s⁻²

def get_gravitational_force(m1, m2, r):
    return G * m1 * m2 / r ** 2

earth_mass = 5.972e24
moon_mass  = 7.348e22
earth_moon_dist = 3.844e8

F_em = get_gravitational_force(earth_mass, moon_mass, earth_moon_dist)
print(f"Earth-Moon gravitational force ≈ {F_em:.3e} N")
````

## 6. Interactive Simulation Section

Change the values in the next cell and re-run the rest of the section to explore different physical scenarios.


````python
# ========== CHANGE THESE VALUES ==========
sim_train_mass = 22680       # kg
sim_train_accel = 10         # m/s²
sim_train_distance = 100     # m
sim_bomb_mass = 1            # kg
sim_c = 3e8                  # m/s

extra_scenarios = [
    {"name": "Light metro",   "mass": 5000,  "accel": 8,  "dist": 80},
    {"name": "Heavy freight", "mass": 40000, "accel": 3,  "dist": 150},
    {"name": "High-speed",    "mass": 18000, "accel": 12, "dist": 120},
]
````

### 6.1 Recompute with simulation parameters

````python
sim_force  = get_force(sim_train_mass, sim_train_accel)
sim_work   = get_work(sim_train_mass, sim_train_accel, sim_train_distance)
sim_energy = get_energy(sim_bomb_mass, c=sim_c)

print(f"Simulated force : {sim_force:,.0f} N")
print(f"Simulated work  : {sim_work:,.0f} J")
print(f"Simulated energy: {sim_energy:.3e} J")
````

### 6.2 Scenario comparison table

````python
print(f"{'Scenario':<16} {'Mass (kg)':>10} {'Accel':>7} {'Dist':>6} {'Force (N)':>12} {'Work (J)':>14}")
print("-" * 70)

# Original / current simulation
print(f"{'Simulated GE':<16} {sim_train_mass:>10,} {sim_train_accel:>7} {sim_train_distance:>6} "
      f"{sim_force:>12,.0f} {sim_work:>14,.0f}")

for sc in extra_scenarios:
    f = get_force(sc["mass"], sc["accel"])
    w = get_work(sc["mass"], sc["accel"], sc["dist"])
    print(f"{sc['name']:<16} {sc['mass']:>10,} {sc['accel']:>7} {sc['dist']:>6} "
          f"{f:>12,.0f} {w:>14,.0f}")
````

### 6.3 Visualisations

````python
import matplotlib.pyplot as plt

# --- Force comparison ---
names = ["Simulated GE"] + [sc["name"] for sc in extra_scenarios]
forces = [sim_force] + [get_force(sc["mass"], sc["accel"]) for sc in extra_scenarios]
works  = [sim_work]  + [get_work(sc["mass"], sc["accel"], sc["dist"]) for sc in extra_scenarios]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(names, forces, color=["#2c7bb6", "#fdae61", "#d7191c", "#1a9641"])
axes[0].set_ylabel("Force (N)")
axes[0].set_title("Force by Scenario")
axes[0].tick_params(axis='x', rotation=20)

axes[1].bar(names, works, color=["#2c7bb6", "#fdae61", "#d7191c", "#1a9641"])
axes[1].set_ylabel("Work (J)")
axes[1].set_title("Work by Scenario")
axes[1].tick_params(axis='x', rotation=20)

plt.suptitle("Physics Class Simulation — Force & Work", fontsize=13)
plt.tight_layout()
plt.show()
````

````python
# Temperature conversion curve (visual check of f_to_c / c_to_f)
import numpy as np

f_range = np.linspace(-40, 212, 100)
c_range = [f_to_c(f) for f in f_range]

plt.figure(figsize=(7, 4))
plt.plot(f_range, c_range, color="#2c7bb6", lw=2, label="F → C")
plt.axhline(0, color="gray", ls="--", lw=0.8)
plt.axvline(32, color="gray", ls="--", lw=0.8)
plt.scatter([32, 212], [0, 100], color="red", zorder=5, label="Key points (32 °F, 212 °F)")
plt.xlabel("Fahrenheit (°F)")
plt.ylabel("Celsius (°C)")
plt.title("Temperature Conversion: Fahrenheit → Celsius")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````
