# NPI Categorical Association Analysis

**Job-Ready EDA Practice: Chi-Square Test for Independence**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency

npi = pd.read_csv('npi_sample.csv')
print(npi.head())
```

## 1. Contingency Tables

```python
# Example: influence vs leader
contingency = pd.crosstab(npi['influence'], npi['leader'])
print(contingency)
```

## 2. Proportions & Marginal Proportions

```python
print(pd.crosstab(npi['influence'], npi['leader'], normalize='all'))
```

## 3. Chi-Square Test

```python
chi2, p, dof, expected = chi2_contingency(contingency)
print('Chi2:', chi2)
print('p-value:', p)
```
