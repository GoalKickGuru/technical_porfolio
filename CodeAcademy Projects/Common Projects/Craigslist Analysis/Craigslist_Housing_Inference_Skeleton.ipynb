**Role:** Data Scientist — Inference Specialist

**Goal:** Build, compare and validate nested linear regression models that predict monthly rental price (`price`) from housing characteristics listed on Craigslist.  Decide which model is best for *description* (in-sample fit) versus *prediction* (out-of-sample accuracy).

You will practice the classic nested-model workflow:

4. Explore residual diagnostics, multicollinearity and model extensions  
5. Run a simulation that varies the random seed and examines scenario predictions
