# Exploring Student Data - Full EDA Notebook

**Job-Ready Data Science Project**

In this notebook, we explore student performance, parental jobs, absences, and more.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

%matplotlib inline
sns.set_style('whitegrid')

students = pd.read_csv('students.csv')
print(students.head())
```

## Initial Exploration

```python
print('Shape:', students.shape)
print('\nSummary Stats:')
print(students.describe(include='all'))
```

## Math Grade Analysis

```python
print('Mean:', students['math_grade'].mean())
print('Median:', students['math_grade'].median())
print('Mode:', students['math_grade'].mode()[0])
print('Range:', students['math_grade'].max() - students['math_grade'].min())
print('Std Dev:', students['math_grade'].std())
print('MAD:', students['math_grade'].mad())
```

```python
# Visualizations
sns.histplot(data=students, x='math_grade', kde=True)
plt.title('Distribution of Math Grades')
plt.show()

sns.boxplot(data=students, x='math_grade')
plt.title('Box Plot of Math Grades')
plt.show()
```

## Parental Jobs Analysis

```python
print('Mother\'s Job:')
print(students['Mjob'].value_counts(normalize=True))

sns.countplot(data=students, x='Mjob')
plt.title("Mother's Job Distribution")
plt.show()
```

## Advanced Features

```python
# Pairplot
sns.pairplot(students, hue='address')
plt.show()

# Correlation
print(students.corr()['math_grade'].sort_values(ascending=False))
```

```python
# Hypothesis Testing: Urban vs Rural Math Grades
urban = students[students['address'] == 'U']['math_grade']
rural = students[students['address'] == 'R']['math_grade']

t_stat, p_val = stats.ttest_ind(urban, rural)
print('T-test p-value:', p_val)
```

## Simulation Section

```python
# Simulate student performance
np.random.seed(42)
sim_grades = np.random.normal(students['math_grade'].mean(), students['math_grade'].std(), 1000)

sns.histplot(sim_grades, kde=True)
plt.title('Simulated Math Grade Distribution')
plt.show()
```
