### Steps 4-8: Baseline Conversion Rate

````python
# Step 4: Get all visitor records
all_visitors = noshmishmosh.customer_visits

# Step 5: Get paying visitor records
paying_visitors = noshmishmosh.purchasing_customers

# Step 6: Count both groups
total_visitor_count = len(all_visitors)
paying_visitor_count = len(paying_visitors)

# Step 7: Calculate baseline percentage
baseline_percent = paying_visitor_count / total_visitor_count * 100

# Step 8: Display result
print(f"Total Visitors: {total_visitor_count}")
print(f"Paying Visitors: {paying_visitor_count}")
print(f"Baseline Conversion Rate: {baseline_percent:.2f}%")
````

**Formula:**

```math
\text{Baseline Conversion Rate} = \left( \frac{\text{Paying Visitors}}{\text{Total Visitors}} \right) \times 100
```

**Why This Matters:**

| Baseline Conversion Rate | Impact on Sample Size |
|-------------------------|--------------------|
| Low (5%) | Harder to detect small changes = LARGER sample needed |
| High (40%) | Easier to detect changes = SMALLER sample needed |

---

### Steps 9-14: Minimum Detectable Effect (MDE)

````python
# Step 9: Get payment history
payment_history = noshmishmosh.money_spent

# Step 10: Calculate average payment
average_payment = np.mean(payment_history)

# Step 11: Find how many NEW customers needed for $1,240
new_customers_needed = np.ceil(1240 / average_payment)

# Step 12: Convert to percentage point increase
percentage_point_increase = new_customers_needed / total_visitor_count * 100

# Step 13: Express as % of baseline (the MDE)
mde = percentage_point_increase / baseline_percent * 100

# Step 14: Display result
print(f"Average Payment: ${average_payment:.2f}")
print(f"New Customers Needed: {int(new_customers_needed)}")
print(f"Percentage Point Increase: {percentage_point_increase:.2f}%")
print(f"Minimum Detectable Effect (MDE): {mde:.2f}%")
````

**MDE Calculation Chain:**

```
$1,240 Revenue Target
        ↓
Divide by Average Payment Amount
        ↓
= Number of New Customers Needed (rounded UP with np.ceil)
        ↓
Divide by Total Visitor Count
        ↓
= Percentage Point Increase (absolute change)
        ↓
Divide by Baseline × 100
        ↓
= MDE (relative lift as percentage of baseline)
```

**Numerical Example:**

| Stage | Example Value | Explanation |
|-------|---------------|-------------|
| Average Payment | $24.80 | From payment_history |
| New Customers Needed | 50 | ceil(1240/24.80) = 50 |
| Percentage Point Increase | 2.5pp | 50/2000 visitors = 2.5% |
| MDE | 12.5% | 2.5pp / 20% baseline = 12.5% lift |

---

### Steps 15-16: Statistical Significance & Final Sample Size

````python
# Step 15: Set significance threshold to 10%
# Not coded but conceptually set
significance_threshold = 0.10

# Step 16: Calculate final sample size
# In practice, this would use an A/B test calculator function
# ab_sample_size = sample_size_calculator(baseline_percent, mde, significance_threshold)
ab_sample_size = 490

print(f"Significance Threshold: {significance_threshold * 100}%")
print(f"Required Sample Size Per Variant: {int(ab_sample_size)}")
print(f"Total Visitors Needed (Both Groups): {int(ab_sample_size * 2)}")
````

**Why 10% Significance Threshold?**

| Threshold | Use Case | Trade-off |
|-----------|----------|-----------|
| 1% | Medical, safety-critical | Very certain, requires VERY large sample |
| 5% | Standard business decisions | Balance of certainty and efficiency |
| 10% | Moderate-risk decisions (this case) | Faster results, accepts some false positives |

**What "490" Means:**

| Group | Visitors | Purpose |
|-------|----------|---------||
| **Control (Group A)** | 490 | Current vegetable photos |
| **Treatment (Group B)** | 490 | Artisanal vegetable photos |
| **Total** | ~980 | Across both variants |

## 🔑 Main Takeaways

1. ✅ **np.ceil matters**: Guarantees we have enough customers to surpass the $1,240 revenue target (rounding down would undershoot)

2. ✅ **Two different calculations**: Percentage points measure absolute change; MDE measures relative change. Both needed because sample size depends on relative lift, not just absolute difference.

3. ✅ **Sample size is sensitive**: Lower baselines (5% vs 40%) and smaller effects (2% vs 20%) dramatically increase required sample sizes.

4. ✅ **Risk tolerance affects certainty**: Choosing 10% significance over 5% speeds up decision-making but accepts higher false-positive risk.

5. ✅ **Business context drives statistics**: The revenue goal ($1,240) directly translates to the MDE, which drives sample size. Clear business objectives enable proper experimental design.
