## 8. Simulation Section — Modify Values & Observe Impact

**Experiment here!** Change the numbers below and re-run to see how metrics change.

Ideas to try:
- Change the discount from 5 to 3 or to 8
- Multiply every `last_week` count by 1.5 (busier week)
- Double the price of the most expensive cut
- Set one popular cut's sales to 0


````r
# === SIMULATION CONTROLS — EDIT THESE ===
DISCOUNT <- 5          # try 3, 5, 8, 10
SALES_MULTIPLIER <- 1.0  # try 1.5 or 0.7
# ========================================

# Rebuild from originals so you can re-run safely
sim_prices <- prices
sim_last_week <- as.integer(round(last_week * SALES_MULTIPLIER))
sim_new_prices <- sim_prices - DISCOUNT

sim_total_revenue <- sum(sim_prices * sim_last_week)
sim_avg_price <- mean(sim_prices)
sim_cuts_under <- hairstyles[sim_new_prices < 30]

cat(sprintf("Discount applied: $%g\n", DISCOUNT))
cat(sprintf("Sales multiplier: %g\n", SALES_MULTIPLIER))
cat(sprintf("Simulated average original price: %.2f\n", sim_avg_price))
cat(sprintf("Simulated total revenue: $%.2f\n", sim_total_revenue))
cat(sprintf("Simulated average daily revenue: $%.2f\n", sim_total_revenue / 7))
cat("Cuts under $30 after discount:", paste(sim_cuts_under, collapse = ", "), "\n")
cat(sprintf("Number of advertised cuts: %d\n", length(sim_cuts_under)))
````
