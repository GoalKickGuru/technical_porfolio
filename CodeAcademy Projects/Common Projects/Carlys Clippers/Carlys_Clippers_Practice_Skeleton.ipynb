## 8. Simulation Section — Modify Values & Observe Impact

**Experiment here!** Change the numbers below and re-run to see how metrics change.

Ideas to try:
- Change the discount from 5 to 3 or to 8
- Multiply every `last_week` count by 1.5 (busier week)
- Double the price of the most expensive cut
- Set one popular cut's sales to 0 (what if mohawk is unavailable?)

Record what happens to average price, total revenue, and cuts under 30.


````python
# === SIMULATION CONTROLS — EDIT THESE ===
DISCOUNT = 5          # try 3, 5, 8, 10
SALES_MULTIPLIER = 1.0  # try 1.5 or 0.7
# ========================================

# Rebuild from originals so you can re-run safely
sim_prices = [p for p in prices]
sim_last_week = [int(c * SALES_MULTIPLIER) for c in last_week]
sim_new_prices = [p - DISCOUNT for p in sim_prices]

sim_total_revenue = sum(p * c for p, c in zip(sim_prices, sim_last_week))
sim_avg_price = sum(sim_prices) / len(sim_prices)
sim_cuts_under = [h for h, np_ in zip(hairstyles, sim_new_prices) if np_ < 30]

print(f"Discount applied: ${DISCOUNT}")
print(f"Sales multiplier: {SALES_MULTIPLIER}")
print(f"Simulated average original price: {sim_avg_price:.2f}")
print(f"Simulated total revenue: ${sim_total_revenue:.2f}")
print(f"Simulated average daily revenue: ${sim_total_revenue/7:.2f}")
print(f"Cuts under $30 after discount: {sim_cuts_under}")
print(f"Number of advertised cuts: {len(sim_cuts_under)}")
````
