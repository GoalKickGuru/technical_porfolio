## 0. Setup & Data


````python
hairstyles = ["bouffant", "pixie", "dreadlocks", "crew", "bowl", "bob", "mohawk", "flattop"]
prices = [30, 25, 40, 20, 20, 35, 50, 35]
last_week = [2, 3, 5, 8, 4, 4, 6, 2]

print("Hairstyles:", hairstyles)
print("Prices   :", prices)
print("Last week:", last_week)
print("Number of styles:", len(hairstyles))
````

## 1. Average Haircut Price


````python
total_price = 0
for price in prices:
    total_price += price

average_price = total_price / len(prices)
print("Average Haircut Price:", average_price)
````

### Alternate 1a — using `sum()`


````python
average_price_alt = sum(prices) / len(prices)
print("Average Haircut Price (sum):", average_price_alt)
````

## 2. New Prices (all reduced by $5)


````python
new_prices = [price - 5 for price in prices]
print(new_prices)
````

## 3. Total Revenue Last Week


````python
total_revenue = 0
for i in range(len(hairstyles)):
    total_revenue += prices[i] * last_week[i]
print("Total Revenue:", total_revenue)
````

### Alternate 3a — zip + generator


````python
total_revenue_alt = sum(p * c for p, c in zip(prices, last_week))
print("Total Revenue (zip):", total_revenue_alt)
````

### Alternate 3b — dictionary of styles


````python
style_data = {h: {"price": p, "sold": c} for h, p, c in zip(hairstyles, prices, last_week)}
total_revenue_dict = sum(d["price"] * d["sold"] for d in style_data.values())
print("Total Revenue (dict):", total_revenue_dict)
print("Example entry:", style_data["mohawk"])
````

## 4. Average Daily Revenue


````python
average_daily_revenue = total_revenue / 7
print("Average Daily Revenue:", average_daily_revenue)
````

## 5. Cuts Under $30 After Discount


````python
cuts_under_30 = [hairstyles[i] for i in range(len(new_prices)) if new_prices[i] < 30]
print(cuts_under_30)
````

### Alternate 5a — zip


````python
cuts_under_30_alt = [h for h, np_ in zip(hairstyles, new_prices) if np_ < 30]
print(cuts_under_30_alt)
````

## 6. More Practice Metrics


````python
# 1. Most popular cut
max_sold = max(last_week)
most_popular_idx = last_week.index(max_sold)
print(f"1. Most popular: {hairstyles[most_popular_idx]} ({max_sold} times)")

# 2. Highest-revenue cut
revenues = [p * c for p, c in zip(prices, last_week)]
max_rev = max(revenues)
max_rev_idx = revenues.index(max_rev)
print(f"2. Highest revenue: {hairstyles[max_rev_idx]} (${max_rev})")

# 3. Percentage under $30 after discount
pct_under = 100 * len(cuts_under_30) / len(hairstyles)
print(f"3. Styles under $30 after discount: {pct_under:.1f}%")

# 4. Total haircuts performed
total_cuts = sum(last_week)
print(f"4. Total haircuts last week: {total_cuts}")

# 5. Summary table
print("\n5. Style summary:")
print(f"{'Style':<12} {'Orig$':>6} {'New$':>6} {'Sold':>5} {'Revenue':>8}")
print("-" * 40)
for h, p, np_, c, r in zip(hairstyles, prices, new_prices, last_week, revenues):
    print(f"{h:<12} {p:>6} {np_:>6} {c:>5} {r:>8}")
````

## 7. Visualization


````python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

x = np.arange(len(hairstyles))
width = 0.35

# Left: prices before / after
axes[0].bar(x - width/2, prices, width, label="Original", color="#4C72B0")
axes[0].bar(x + width/2, new_prices, width, label="After −$5", color="#55A868")
axes[0].set_xticks(x)
axes[0].set_xticklabels(hairstyles, rotation=45, ha="right")
axes[0].set_ylabel("Price ($)")
axes[0].set_title("Prices Before vs After $5 Discount")
axes[0].legend()
axes[0].axhline(30, color="red", linestyle="--", alpha=0.7, label="$30 threshold")
axes[0].legend()

# Right: revenue by style
revenues = [p * c for p, c in zip(prices, last_week)]
colors = ["#C44E52" if r == max(revenues) else "#4C72B0" for r in revenues]
axes[1].bar(hairstyles, revenues, color=colors)
axes[1].set_xticks(range(len(hairstyles)))
axes[1].set_xticklabels(hairstyles, rotation=45, ha="right")
axes[1].set_ylabel("Revenue ($)")
axes[1].set_title("Revenue by Hairstyle (Last Week)")
axes[1].bar_label(axes[1].containers[0], fmt="$%d")

plt.tight_layout()
plt.savefig("/home/workdir/artifacts/carlys_clippers_charts.png", dpi=120, bbox_inches="tight")
plt.show()
print("Chart saved to artifacts/carlys_clippers_charts.png")
````

## 8. Simulation Section — Modify Values & Observe Impact

Change `DISCOUNT` and `SALES_MULTIPLIER` (or edit the lists) and re-run the cell.


````python
# === SIMULATION CONTROLS — EDIT THESE ===
DISCOUNT = 5
SALES_MULTIPLIER = 1.0
# ========================================

sim_prices = list(prices)
sim_last_week = [max(0, int(round(c * SALES_MULTIPLIER))) for c in last_week]
sim_new_prices = [p - DISCOUNT for p in sim_prices]

sim_total_revenue = sum(p * c for p, c in zip(sim_prices, sim_last_week))
sim_avg_price = sum(sim_prices) / len(sim_prices)
sim_cuts_under = [h for h, np_ in zip(hairstyles, sim_new_prices) if np_ < 30]
sim_daily = sim_total_revenue / 7

print(f"=== Simulation Results ===")
print(f"Discount applied          : ${DISCOUNT}")
print(f"Sales multiplier          : {SALES_MULTIPLIER}")
print(f"Average original price    : ${sim_avg_price:.2f}")
print(f"Total revenue             : ${sim_total_revenue:.2f}")
print(f"Average daily revenue     : ${sim_daily:.2f}")
print(f"Cuts under $30            : {sim_cuts_under}")
print(f"# styles advertised       : {len(sim_cuts_under)}")

# Quick sensitivity table
print("\nSensitivity of total revenue to different discounts (sales fixed):")
print(f"{'Discount':>10} {'New Avg Price':>14} {'Cuts < $30':>12} {'Total Rev':>12}")
for d in [0, 3, 5, 8, 10]:
    np_list = [p - d for p in prices]
    under = sum(1 for np_ in np_list if np_ < 30)
    # note: revenue uses ORIGINAL prices; discount is marketing, not immediate price cut on past sales
    print(f"{d:>10} {sum(np_list)/len(np_list):>14.2f} {under:>12} {total_revenue:>12}")
````

### Extra simulation: What if last week’s sales pattern repeats for a full month (4 weeks) and we apply the $5 discount going forward?

Assume the discount does not change demand (conservative). Then projected monthly revenue at new prices = 4 × Σ (new_price × last_week count).


````python
monthly_at_new_prices = 4 * sum(np_ * c for np_, c in zip(new_prices, last_week))
monthly_at_old_prices = 4 * total_revenue
print(f"Projected 4-week revenue at original prices : ${monthly_at_old_prices:,.0f}")
print(f"Projected 4-week revenue at discounted prices: ${monthly_at_new_prices:,.0f}")
print(f"Difference (cost of $5 discount, same volume): ${monthly_at_old_prices - monthly_at_new_prices:,.0f}")
print("\nCarly should weigh this against expected extra volume from advertising the under-$30 cuts.")
````
