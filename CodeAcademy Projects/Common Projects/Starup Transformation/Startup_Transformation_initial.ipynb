````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.preprocessing import StandardScaler
````

````python
# Load Data
financial_data = pd.read_csv('/home/workdir/attachments/financial_data.csv')
expenses = pd.read_csv('/home/workdir/attachments/expenses.csv')
employees = pd.read_csv('/home/workdir/attachments/employees.csv')
print(financial_data.head())
print(expenses)
````

````python
# Financial Plots
plt.figure(figsize=(10,5))
plt.plot(financial_data['Month'], financial_data['Revenue'], label='Revenue')
plt.plot(financial_data['Month'], financial_data['Expenses'], label='Expenses')
plt.xlabel('Month'); plt.ylabel('Amount ($)'); plt.title('Financial Trends'); plt.legend(); plt.show()
````

````python
# Expense Pie (Alternate with seaborn if desired)
categories = ['Salaries', 'Advertising', 'Office Rent', 'Other']
props = [0.62, 0.15, 0.15, 0.08]
plt.pie(props, labels=categories, autopct='%1.1f%%'); plt.title('Expenses'); plt.axis('equal'); plt.show()
````

````python
# Employees
sorted_prod = employees.sort_values('Productivity')
cut_employees = sorted_prod.head(100)
print('Employees to cut:', cut_employees[['Name','Productivity']].head())
````

````python
# Commute
commute = employees['Commute Time']
print(commute.describe())
plt.hist(commute, alpha=0.5, label='Original')
plt.hist(np.log(commute), alpha=0.5, label='Log')
plt.legend(); plt.show()
````

````python
# Simulation
def simulate_cuts(cutoffs=[50,100,150]):
    results = []
    for c in cutoffs:
        remaining = employees.sort_values('Productivity').iloc[c:]['Productivity'].mean()
        results.append({'Cutoff':c, 'Avg_Prod_Remaining':round(remaining,2)})
    return pd.DataFrame(results)

print(simulate_cuts())
````
