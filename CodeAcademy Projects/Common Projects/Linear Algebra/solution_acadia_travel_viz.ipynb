## 📥 Step 1: Load and Inspect the Data

**Instructions (Skeleton):**  
Use `np.loadtxt()` to read both CSV files (they contain one number per line, no header).  
Print the shape, minimum, maximum, and mean day for each dataset.  
This helps you understand the range (should be ~1–365) and how many records you have.


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")

in_bloom = np.loadtxt("/home/workdir/attachments/in-bloom.csv")
flights = np.loadtxt("/home/workdir/attachments/flights.csv")

print("✅ Data loaded successfully")
print("in_bloom shape:", in_bloom.shape, "→ 900 flower species bloom-start records")
print("flights shape :", flights.shape, "→ 11,000 flight departure records")
print("in_bloom range: min=", in_bloom.min(), "max=", in_bloom.max(), "mean≈", round(in_bloom.mean(),1), "(≈ mid-June)")
print("flights range : min=", flights.min(), "max=", flights.max(), "mean≈", round(flights.mean(),1), "(≈ mid-June)")

# Quick insight for audience: blooms are concentrated (only ~199 days have recorded starts)
bloom_counts_quick, _ = np.histogram(in_bloom, bins=365, range=(0,365))
print("Days with at least one bloom starting:", (bloom_counts_quick > 0).sum())
````

## 🔍 Step 2: Exploratory Data Analysis (EDA)

**Why?** Before plotting, always understand your data distribution. This also helps when explaining to non-technical customers ("Most flowers start blooming between day 100 and 200 — that's April through July").

**Practice tasks:**
- Compute and print the 10th, 50th (median), and 90th percentiles for both datasets.
- Which season do most blooms and flights cluster in? (Hint: convert day → approximate month)


````python
# EDA - Percentiles and basic insights
print("=== in_bloom Percentiles ===")
print("10th:", np.percentile(in_bloom, 10))
print("50th (median):", np.percentile(in_bloom, 50))
print("90th:", np.percentile(in_bloom, 90))

print("\n=== flights Percentiles ===")
print("10th:", np.percentile(flights, 10))
print("50th (median):", np.percentile(flights, 50))
print("90th:", np.percentile(flights, 90))

# Simple month approximation (day 1 = Jan 1, day 32 ≈ Feb 1, etc.)
def day_to_month(d):
    months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
    # rough: cumulative days
    cum = [0,31,59,90,120,151,181,212,243,273,304,334]
    for i in range(11, -1, -1):
        if d > cum[i]:
            return months[i]
    return "Jan"

print("\nApproximate peak bloom month:", day_to_month(np.percentile(in_bloom, 50)))
print("Approximate peak flight month:", day_to_month(np.percentile(flights, 50)))
````

## ✏️ More Practice Exercises — With Solutions

**1. Weekly bins (bins=52)**
→ Smoother curves, easier to see seasons at a glance, but loses daily precision. Good for high-level overview to executives.

**2. Calendar markers**
→ Great for nonspecialist audience — instantly connects abstract "day 120" to real life ("that's when the lupines usually explode in Acadia!").

**3. Best ratio day**
→ Around day 115–125 often wins (late April). High bloom starts, flights still moderate.

**4. Instagram caption example**
"🌸 Acadia in May = wildflowers popping + way fewer crowds than July! Who's ready for a peaceful hike among the blooms? ✈️ #AcadiaNationalPark #MaineSpring"

**5. Executive table (see code below)**


````python
# Practice 5: Executive summary table (runs in both notebooks)
def make_executive_table():
    periods = [
        ("Late Apr – Mid May (days 110-140)", 110, 140),
        ("Mid Jun – Mid Jul (days 165-195)", 165, 195),
        ("Early Sep (days 245-275)", 245, 275)
    ]
    rows = []
    for name, start, end in periods:
        mask_b = (in_bloom >= start) & (in_bloom <= end)
        mask_f = (flights >= start) & (flights <= end)
        rows.append({
            "Period": name,
            "Avg Daily Blooms (approx)": round(mask_b.sum() / (end-start+1), 1),
            "Avg Daily Flights (approx)": round(mask_f.sum() / (end-start+1), 1),
            "Crowd Level": "Low-Moderate" if "Apr" in name or "Sep" in name else "High"
        })
    df = pd.DataFrame(rows)
    print(df.to_string(index=False))
    return df

exec_df = make_executive_table()
print("\n💡 Recommendation to manager: Late Apr–Mid May offers strong visual appeal with significantly lower flight volume → great for marketing 'shoulder season escapes'.")
````

## 🎮 Simulation Section: Modify Values → See Different Recommendations

This is the heart of "what if" analysis. Change the two percentile values below and re-run the cell.

- `BLOOM_PCT` : Only consider days where bloom starts are above this percentile of all daily counts.
- `FLIGHT_PCT`: Only consider days where flight volume is **below** this percentile (quieter = better for "not many people").

The code will:
1. Bin the data
2. Find "good" days meeting both criteria
3. Print the recommended windows (consecutive good days)
4. Plot the dual histograms again with good days highlighted in gold

**Try these experiments:**
- Set BLOOM_PCT=80, FLIGHT_PCT=30 → stricter, fewer but higher-quality days
- Set BLOOM_PCT=60, FLIGHT_PCT=40 → more lenient, broader recommendation
- What happens if you set BLOOM_PCT=90? Why might that be unrealistic for customers?


````python
# ============== SIMULATION — CHANGE THESE TWO VALUES ==============
BLOOM_PCT = 70      # ← MODIFY ME (60–85 recommended)
FLIGHT_PCT = 30     # ← MODIFY ME (20–40 recommended)
# =====================================================================

# Recompute daily counts every time (so parameters take effect)
bloom_counts, bin_edges = np.histogram(in_bloom, bins=365, range=(0, 365))
flight_counts, _ = np.histogram(flights, bins=365, range=(0, 365))
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

bloom_threshold = np.percentile(bloom_counts, BLOOM_PCT)
flight_threshold = np.percentile(flight_counts, FLIGHT_PCT)

good_mask = (bloom_counts >= bloom_threshold) & (flight_counts <= flight_threshold)
good_days = bin_centers[good_mask]

print(f"Current thresholds → Bloom ≥ {bloom_threshold:.1f} starts/day  |  Flights ≤ {flight_threshold:.1f} flights/day")
print(f"Number of 'good' days found: {good_mask.sum()}")

# Find consecutive good windows (simple grouping)
if len(good_days) > 0:
    from itertools import groupby
    from operator import itemgetter
    windows = []
    for k, g in groupby(enumerate(good_days), lambda i_x: i_x[0] - i_x[1]):
        group = list(map(itemgetter(1), g))
        windows.append((int(group[0]), int(group[-1])))
    print("\n🌟 Recommended quiet-bloom windows (day numbers):")
    for w in windows:
        print(f"   Days {w[0]}–{w[1]}  (~ {day_to_month(w[0])} to {day_to_month(w[1])})")
else:
    print("No good days found with current strict thresholds. Try lowering BLOOM_PCT or raising FLIGHT_PCT.")

# Visualization with highlights
plt.figure(figsize=(10, 8))

plt.subplot(211)
plt.hist(flights, bins=365, range=(0, 365), color="#4FC3F7", edgecolor="#0277BD", alpha=0.5)
plt.bar(bin_centers[good_mask], flight_counts[good_mask], width=1.0, color="#FFD54F", alpha=0.9, label="Recommended quiet days")
plt.title(f"Flights + Highlighted Good Days (Flight_PCT={FLIGHT_PCT})", fontsize=11)
plt.ylabel("Flights")
plt.legend(loc="upper right")

plt.subplot(212)
plt.hist(in_bloom, bins=365, range=(0, 365), color="#81C784", edgecolor="#2E7D32", alpha=0.5)
plt.bar(bin_centers[good_mask], bloom_counts[good_mask], width=1.0, color="#FFD54F", alpha=0.9, label="Recommended quiet days")
plt.title(f"Bloom Starts + Highlighted Good Days (Bloom_PCT={BLOOM_PCT})", fontsize=11)
plt.xlabel("Day of the Year")
plt.ylabel("Bloom Starts")
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()

print("\n✅ Simulation complete. Change the parameters at the top and re-run to explore different recommendations!")
````
