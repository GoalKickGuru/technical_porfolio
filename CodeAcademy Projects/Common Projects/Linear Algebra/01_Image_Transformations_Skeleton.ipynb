## 6. Linear Algebra: Solving for the 'Secret' Matrix

We generate a random matrix `A` and solve `A @ X = heart_img` for `X`.
Then reconstruct `solved = A @ X` and verify it matches the original heart.

This demonstrates solving systems of linear equations — fundamental in data science (regression, PCA, etc.).

**Practical note**: In real life this idea appears in image compression, denoising, and recommender systems.

```
# TODO: Random matrix and solve
np.random.seed(123)
random_img = np.random.randint(0, 255, size=(7,7)).astype(float)
show_image(random_img, "Random Matrix A")

# Solve A @ X = heart_img  =>  X = A^{-1} @ heart_img
X = np.linalg.solve(random_img, heart_img)
show_image(X, "Solved X matrix")

# Reconstruct
reconstructed = random_img @ X
show_image(reconstructed, "Reconstructed Heart (should look like original)")
print("Reconstruction error (Frobenius norm):", np.linalg.norm(reconstructed - heart_img))
```

## 7. Simulation Section: Modify Parameters & Observe Impact

**This is the core new feature**. Change the values below and re-run the cell to see how transformations affect the image and quantitative metrics.

Metrics we track:
- Mean Absolute Error (MAE) vs original
- Peak Signal-to-Noise Ratio (PSNR) approximation
- Visual side-by-side comparison

```
# ========== SIMULATION CELL - TWEAK THESE PARAMETERS ==========
brightness_factor = 0.7     # 0.5 = darker, 1.5 = brighter
noise_std = 35              # 0 = clean, 50 = quite noisy
apply_blur = True           # True/False
blur_sigma = 0.8
do_invert = False
flip_mode = 'none'          # 'h', 'v', 'none'

# Apply pipeline
sim_img = heart_img.copy().astype(float)
sim_img = np.clip(sim_img * brightness_factor, 0, 255)
if noise_std > 0:
    sim_img += np.random.normal(0, noise_std, sim_img.shape)
    sim_img = np.clip(sim_img, 0, 255)
if apply_blur:
    sim_img = ndimage.gaussian_filter(sim_img, sigma=blur_sigma)
if do_invert:
    sim_img = 255 - sim_img
if flip_mode == 'h':
    sim_img = np.fliplr(sim_img)
elif flip_mode == 'v':
    sim_img = np.flipud(sim_img)

# Metrics
mae = np.mean(np.abs(sim_img - heart_img))
mse = np.mean((sim_img - heart_img)**2)
psnr = 20 * np.log10(255 / np.sqrt(mse)) if mse > 0 else float('inf')
print("=== SIMULATION RESULTS ===")
print(f"Brightness factor : {brightness_factor}")
print(f"Noise std         : {noise_std}")
print(f"Blur sigma        : {blur_sigma if apply_blur else 'N/A'}")
print(f"MAE vs original   : {mae:.2f}")
print(f"MSE vs original   : {mse:.2f}")
print(f"Approx PSNR (dB)  : {psnr:.2f}")

# Visual comparison
fig, axes = plt.subplots(1, 2, figsize=(10,4))
axes[0].imshow(heart_img, cmap='gray')
axes[0].set_title('Original Heart')
axes[0].axis('off')
axes[1].imshow(sim_img, cmap='gray')
axes[1].set_title(f'Transformed (MAE={mae:.1f})')
axes[1].axis('off')
plt.tight_layout()
plt.show()
```
