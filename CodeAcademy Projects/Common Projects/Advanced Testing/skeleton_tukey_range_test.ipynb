## VeryAnts Sales – Identifying Which Stores Differ

After a significant one-way ANOVA, we use **Tukey’s Honestly Significant Difference (HSD)** test to determine *which specific pairs* of groups differ, while controlling the overall Type I error rate.

This exercise expands the original prompt into a complete, interpretable workflow with simulation and audience considerations.


## Flowchart: When and How to Use Tukey’s Range Test
```mermaid
flowchart TD
    A[Run One-way ANOVA] --> B{Is ANOVA p < 0.05?}
    B -->|Yes| C[Perform Tukey HSD<br/>pairwise_tukeyhsd(endog, groups, alpha=0.05)]
    B -->|No| Stop[Stop - No evidence of any differences]
    C --> D[Inspect output table:<br/>meandiff, lower, upper, reject, p-adj]
    D --> E{reject == True?}
    E -->|Yes| F[Significant difference between that pair<br/>Report mean diff + CI + effect size]
    E -->|No| G[No significant difference after correction]
    F --> H[Consider audience when reporting<br/>Which pairs differ? Practical importance?]
    G --> H
    H --> I[Conclusion in data analysis report style]
```
**Key advantage of Tukey:** Controls family-wise error rate for all pairwise comparisons while being more powerful than Bonferroni.

## 6. Simulation: Understanding Tukey’s Behavior

**Goal:** See how Tukey performs when there are real differences vs when all groups are equal.

Modify the parameters below and observe how often Tukey detects differences.


````python
import numpy as np
from statsmodels.stats.multicomp import pairwise_tukeyhsd

np.random.seed(42)

# === MODIFIABLE PARAMETERS ===
true_means = [58, 65, 62]      # change to [60,60,60] to simulate no differences
sigma = 15
n_per_group = 150
n_simulations = 300
alpha = 0.05

# === Simulation skeleton ===
significant_count = 0

for i in range(n_simulations):
    g1 = np.random.normal(true_means[0], sigma, n_per_group)
    g2 = np.random.normal(true_means[1], sigma, n_per_group)
    g3 = np.random.normal(true_means[2], sigma, n_per_group)
    
    # Create a small DataFrame for Tukey
    df_sim = pd.DataFrame({
        'value': np.concatenate([g1, g2, g3]),
        'group': ['A']*n_per_group + ['B']*n_per_group + ['C']*n_per_group
    })
    
    tukey_sim = pairwise_tukeyhsd(df_sim['value'], df_sim['group'], alpha)
    
    # Check if any pair is rejected
    if any(tukey_sim.reject):
        significant_count += 1

detection_rate = significant_count / n_simulations
print(f'Proportion of simulations where Tukey found at least one significant pair: {detection_rate:.3f}')
print('When all means equal → this should be close to alpha (good FWER control)')
print('When real differences exist → this is power')
````
