## 6. Simulation: Impact of Violating Independence

Modify the parameters below to see how p-values behave when observations are independent vs dependent.


````python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

np.random.seed(42)

# === MODIFIABLE PARAMETERS ===
n_subjects = 500
prob_sick_vaccine = 0.10
prob_sick_placebo = 0.25
n_simulations = 300

# Simulate independent groups (correct use of Chi-Square)
sig_independent = 0
for i in range(n_simulations):
    vaccine_sick = np.random.binomial(n_subjects//2, prob_sick_vaccine)
    placebo_sick = np.random.binomial(n_subjects//2, prob_sick_placebo)
    table = [[vaccine_sick, (n_subjects//2 - vaccine_sick)],
             [placebo_sick, (n_subjects//2 - placebo_sick)]]
    p = chi2_contingency(table)[1]
    if p < 0.05:
        sig_independent += 1

print(f"Power with independent groups: {sig_independent / n_simulations:.3f}")
print("Try changing the probabilities to see how power changes.")
````
