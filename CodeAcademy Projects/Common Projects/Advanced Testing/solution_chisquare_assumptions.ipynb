# SOLUTION: Assumptions of the Chi-Square Test – Detailed Explanations & Decision Framework


## Flowchart: Is a Chi-Square Test Appropriate?
```mermaid
flowchart TD
    A[Two Categorical Variables?] --> B{Are observations independent?}
    B -->|No (e.g. same people before/after)| C[Use McNemar's Test or

    D -->|No| E[Recode or use different test]

    F -->|No| G[Use Fisher's Exact Test]

```
**Most common violation in practice:** Using Chi-Square on paired/repeated measures data (same subjects measured multiple times).


## 1. Review of Assumptions (Solution)

**Key insight:** The **independent groups** assumption is the one most frequently violated in practice, especially in studies that measure the same people before and after an intervention (repeated measures / paired design).


## 2. Scenario 1: Vaccine Study (Solution)

**Answer: `checkpoint_1 = True`**

**Explanation:**
- Adults were randomly assigned to vaccine or placebo → groups are independent.
- Each person is measured once (sick or not sick) → categories are mutually exclusive.
- Random sampling assumed.
- Chi-Square test is appropriate to test association between Group and Outcome.


````python
checkpoint_1 = True
print(f"Checkpoint 1: {checkpoint_1} — Chi-Square is appropriate.")
````

## 3. Scenario 2: Yoga & Happiness (Solution)

**Answer: `checkpoint_2 = False`**

**Explanation:**
- Same people are measured **before and after** yoga → observations are **paired/dependent**.
- Chi-Square assumes independent groups. Using it here would be incorrect and could lead to misleading p-values.
- Correct test: **McNemar's test** (for paired binary or categorical data with two time points).


````python
checkpoint_2 = False
print(f"Checkpoint 2: {checkpoint_2} — Chi-Square is NOT appropriate (paired data).")
````

## 5. What to Do When Assumptions Are Violated (Solution)

**When groups are not independent (paired/repeated measures):**
- Use **McNemar's test** (for 2x2 paired data)
- Or **Cochran's Q test** for more than two related samples

**When expected cell counts are too small:**
- Use **Fisher's Exact Test** (especially for 2x2 tables)

**Main risk of misusing Chi-Square on dependent data:** Inflated Type I error rate (false positives) because the test assumes more information than actually exists.


## 6. Simulation (Solution)

With independent groups and the given probabilities, power is high. If we had used paired data incorrectly with Chi-Square, we would get misleading results.


````python
import numpy as np
from scipy.stats import chi2_contingency

np.random.seed(42)

n_subjects = 500
prob_sick_vaccine = 0.10
prob_sick_placebo = 0.25
n_simulations = 300

sig_independent = 0
for i in range(n_simulations):
    vaccine_sick = np.random.binomial(n_subjects//2, prob_sick_vaccine)
    placebo_sick = np.random.binomial(n_subjects//2, prob_sick_placebo)
    table = [[vaccine_sick, (n_subjects//2 - vaccine_sick)],
             [placebo_sick, (n_subjects//2 - placebo_sick)]]
    p = chi2_contingency(table)[1]
    if p < 0.05:
        sig_independent += 1

print(f"Power with independent groups: {sig_independent / n_simulations:.3f}")
print("This shows good power when the design is correct (independent groups).")
````
