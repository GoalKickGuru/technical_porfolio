## 6. Simulation: Impact of Violated Assumptions

Modify parameters and observe how results change when variances differ greatly.


````r
set.seed(42)

# === MODIFIABLE PARAMETERS ===
n <- 100
mean1 <- 18; mean2 <- 12
sd1 <- 3; sd2 <- 5          # try sd2 = 8 or 10 to strongly violate equal variance
n_simulations <- 500
alpha <- 0.05

# === Simulation skeleton ===
sig_count <- 0

for (i in 1:n_simulations) {
  g1 <- rnorm(n, mean1, sd1)
  g2 <- rnorm(n, mean2, sd2)
  p <- t.test(g1, g2, var.equal = TRUE)$p.value   # try var.equal = FALSE
  if (p < alpha) sig_count <- sig_count + 1
}

cat("Proportion of simulations p < 0.05:", round(sig_count / n_simulations, 3), "\n")
cat("Try changing sd2 to a much larger value and compare var.equal = TRUE vs FALSE.\n")
````
