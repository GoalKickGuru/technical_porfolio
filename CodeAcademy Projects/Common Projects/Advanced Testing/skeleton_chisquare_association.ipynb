## Flowchart: Choosing the Right Test for Categorical Associations
```mermaid
flowchart TD
    A[Two Categorical Variables?] --> B{Create Contingency Table}
    B --> C{Expected counts < 5 in >20% of cells?}
    C -->|Yes| D[Use Fisher's Exact Test<br/>(especially 2x2)]
    C -->|No| E[Run Chi-Square Test of Independence]
    E --> F{p < 0.05?}
    F -->|Yes| G[Calculate Cramér's V (effect size)]
    F -->|No| H[No significant association]
    G --> I[Post-hoc: Pairwise Chi-Square with Bonferroni]
    I --> J[Report results with audience in mind]
    H --> J
```
**Practical tip:** Always report both the p-value **and** Cramér's V. A significant p-value with tiny effect size is often not practically meaningful.


## 1. Load Data and Create Contingency Table

**Instructions:**
1. Load `ants_grade.csv`
2. Create a contingency table using `pd.crosstab(ants.Grade, ants.Ant)`
3. Print the table and add margins (totals)


````python
import pandas as pd
from scipy.stats import chi2_contingency

# TODO: Load data
ants = pd.read_csv('ants_grade.csv')

# TODO: Create contingency table
table = pd.crosstab(ants.Grade, ants.Ant)
print(table)

# TODO: Add margins
print(pd.crosstab(ants.Grade, ants.Ant, margins=True))
````

````python
# TODO: Run Chi-Square test
chi2, pval, dof, expected = chi2_contingency(table)
print(f'Chi2 = {chi2:.3f}, p-value = {pval:.4f}, df = {dof}')
print('Expected frequencies:\n', expected)

# TODO: Significance decision
significant = ___
print(f'Significant association? {significant}')
````

## 3. Effect Size: Cramér's V

**Instructions:**
Calculate Cramér's V (effect size for Chi-Square):
`cramers_v = sqrt(chi2 / (n * (min(r, c) - 1)))`
where n = total observations, r = rows, c = columns.

Interpret: 0.1 = small, 0.3 = medium, 0.5 = large.


````python
import numpy as np

# TODO: Calculate Cramér's V
n = table.sum().sum()
r, c = table.shape
cramers_v = np.sqrt(chi2 / (n * (min(r, c) - 1)))
print(f"Cramér's V = {cramers_v:.3f}")
````

## 6. Simulation: Understanding Chi-Square Behavior

Modify parameters to see how p-value and Cramér's V behave under the null vs alternative.


````python
import numpy as np
from scipy.stats import chi2_contingency

np.random.seed(42)

# === MODIFIABLE ===
n_per_grade = 36
prob_leaf_under_null = 0.24        # overall proportion of leaf cutters
prob_leaf_3rd = 0.36               # try increasing this to create association
n_simulations = 300

sig_count = 0
for i in range(n_simulations):
    # Simulate data under alternative
    ants_sim = []
    for grade in ['1st', '2nd', '3rd']:
        p_leaf = prob_leaf_3rd if grade == '3rd' else prob_leaf_under_null
        ants_sim += ['leaf cutter'] * np.random.binomial(n_per_grade, p_leaf)
        ants_sim += ['harvester'] * (n_per_grade - np.random.binomial(n_per_grade, p_leaf))
    
    df_sim = pd.DataFrame({'Grade': ['1st']*n_per_grade + ['2nd']*n_per_grade + ['3rd']*n_per_grade,
                           'Ant': ants_sim})
    tab = pd.crosstab(df_sim.Grade, df_sim.Ant)
    p = chi2_contingency(tab)[1]
    if p < 0.05:
        sig_count += 1

print(f'Proportion of simulations detecting association: {sig_count / n_simulations:.3f}')
print('Increase prob_leaf_3rd to see higher detection rate (power).')
````
