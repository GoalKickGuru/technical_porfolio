````python
import pandas as pd\nimport numpy as np\nfrom scipy.stats import ttest_1samp, binom_test\nheart = pd.read_csv('/home/workdir/artifacts/heart_disease.csv')\nprint('Data loaded successfully. Shape:', heart.shape)\nprint(heart.head())
````

## 1-4. Cholesterol with Heart Disease

````python
yes_hd = heart[heart.heart_disease == 'presence']\nchol_hd = yes_hd.chol\nmean_hd = np.mean(chol_hd)\nprint('Mean chol HD:', mean_hd)\n# Alternate: yes_hd['chol'].mean()\ntstat, pval = ttest_1samp(chol_hd, 240)\none_sided_p = pval / 2\nprint('One-sided p-value:', one_sided_p)\nprint('Significant (>240):', one_sided_p < 0.05)
````

## 5. Cholesterol without Heart Disease

````python
no_hd = heart[heart.heart_disease == 'absence']\nchol_no = no_hd.chol\nprint('Mean chol No HD:', np.mean(chol_no))\ntstat, pval = ttest_1samp(chol_no, 240)\nprint('One-sided p-value:', pval / 2)
````
