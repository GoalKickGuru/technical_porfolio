````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
website = pd.read_csv('/home/workdir/attachments/website.csv')
print(website.head())
print(website.describe())
````

````python
plt.scatter(website.age, website.time_seconds)
plt.xlabel('Age')
plt.ylabel('Time Seconds')
plt.show()
plt.clf()
````

````python
model = sm.OLS.from_formula('time_seconds ~ age', website)
results = model.fit()
print(results.params)

plt.scatter(website.age, website.time_seconds)
plt.plot(website.age, results.params[0] + results.params[1]*website.age, color='red')
plt.show()
plt.clf()
````

````python
fitted = results.predict(website)
residuals = website.time_seconds - fitted
plt.hist(residuals)
plt.show()
plt.clf()

plt.scatter(fitted, residuals)
plt.axhline(0, color='red')
plt.show()
plt.clf()
````

````python
pred40 = results.params[0] + results.params[1]*40
print('Predicted time for 40yo:', pred40)
````

````python
model_b = sm.OLS.from_formula('time_seconds ~ browser', website)
results_b = model_b.fit()
print(results_b.params)

mean_chrome = website[website.browser == 'Chrome']['time_seconds'].mean()
mean_safari = website[website.browser == 'Safari']['time_seconds'].mean()
print('Mean Chrome:', mean_chrome)
print('Mean Safari:', mean_safari)
print('Difference:', mean_chrome - mean_safari)
````

## Simulation & Practice

````python
np.random.seed(42)
sim_age = np.random.normal(40, 15, 100)
sim_time = 300 + 4*sim_age + np.random.normal(0, 50, 100)
sim_df = pd.DataFrame({'age': sim_age, 'time_seconds': sim_time})
print('Simulation model:', sm.OLS.from_formula('time_seconds ~ age', sim_df).fit().params)
````
