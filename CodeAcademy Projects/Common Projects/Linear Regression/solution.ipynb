Full solution for linear regression on score vs hours_studied and breakfast.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

students = pd.read_csv('/home/workdir/attachments/test_data.csv')
print('Data loaded:')
print(students.head())
print(students.describe())
````

## Step 1: Visualize

````python
plt.figure(figsize=(10,6))
plt.scatter(students.hours_studied, students.score, alpha=0.7)
plt.xlabel('Hours Studied')
plt.ylabel('Score (0-100)')
y_manual = 9.85 * students.hours_studied + 43
plt.plot(students.hours_studied, y_manual, color='red', label='Manual Line')
plt.legend()
plt.title('Score vs Hours Studied')
plt.show()

# Alternate: Using seaborn if available, but stick to matplotlib
````

## Step 2: Fit OLS Model

````python
model = sm.OLS.from_formula('score ~ hours_studied', data=students)
results = model.fit()
print('Model Coefficients:')
print(results.params)
print('\nSummary:')
print(results.summary())
````

## Step 3: Predictions

````python
# Manual
pred_3hr = 9.848111 * 3 + 43.016014
print('Predicted score for 3 hours:', pred_3hr)

# Using model
newdata = pd.DataFrame({'hours_studied': [5]})
pred_5hr = results.predict(newdata)
print('Predicted score for 5 hours:', pred_5hr.iloc[0])
````

## Step 4: Residuals

````python
fitted_values = results.predict(students)
residuals = students.score - fitted_values

print('First 5 residuals:')
print(residuals.head())

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(residuals, bins=20)
plt.title('Histogram of Residuals')

plt.subplot(1,2,2)
plt.scatter(fitted_values, residuals)
plt.axhline(0, color='red')
plt.title('Residuals vs Fitted')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.show()
````

## Step 5: Breakfast Analysis

````python
print('Mean score by breakfast:')
print(students.groupby('breakfast')['score'].mean())

model_bf = sm.OLS.from_formula('score ~ breakfast', data=students)
results_bf = model_bf.fit()
print(results_bf.params)

# Difference
mean_no = students[students.breakfast == 0]['score'].mean()
mean_yes = students[students.breakfast == 1]['score'].mean()
print('Difference:', mean_yes - mean_no)

# Plot
plt.scatter(students.breakfast, students.score)
plt.plot([0,1], [mean_no, mean_yes], color='red')
plt.xticks([0,1], ['No Breakfast', 'Breakfast'])
plt.ylabel('Score')
plt.show()
````

## More Practice & Simulation Section
Simulate data with different slopes/intercepts.

````python
# Simulation
np.random.seed(42)
n = 100
hours_sim = np.random.uniform(0.5, 5, n)
score_sim = 10 * hours_sim + 40 + np.random.normal(0, 8, n)  # Modify slope/intercept/noise here
sim_df = pd.DataFrame({'hours_studied': hours_sim, 'score': score_sim})

model_sim = sm.OLS.from_formula('score ~ hours_studied', data=sim_df).fit()
print('Simulated model params:', model_sim.params)

# Plot sim
plt.scatter(sim_df.hours_studied, sim_df.score)
plt.plot(sim_df.hours_studied, model_sim.predict(sim_df), color='red')
plt.show()

# Alternate code: Use np.polyfit for manual regression
slope, intercept = np.polyfit(students.hours_studied, students.score, 1)
print('Polyfit slope:', slope, 'intercept:', intercept)
````
