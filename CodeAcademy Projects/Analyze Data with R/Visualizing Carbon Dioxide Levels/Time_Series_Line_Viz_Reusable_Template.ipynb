# Time-Series Line Visualization Template (ggplot2)

**Reusable for any paleo + modern comparison** (CO₂, temperature, CH₄, sea level, etc.)

1. Point the two `read_csv` paths at your files.
2. Edit the column names inside `aes()`.
3. Adjust titles, the reference-line calculation, and optional filters.
4. Run.


```r
library(readr); library(dplyr); library(ggplot2)
options(scipen = 10000)

# ---- USER INPUTS ----
PALEO_FILE   <- "data/carbon_dioxide_levels.csv"
MODERN_FILE  <- "data/yearly_co2.csv"
PALEO_X      <- "Age_yrBP"      # or your age / year column (older = larger number)
PALEO_Y      <- "CO2_ppmv"
MODERN_X     <- "year"
MODERN_Y     <- "data_mean_global"
REVERSE_X    <- TRUE            # TRUE for BP-style axes
# ---------------------

paleo  <- read_csv(PALEO_FILE,  show_col_types = FALSE)
modern <- read_csv(MODERN_FILE, show_col_types = FALSE)

# Deep-time plot
p1 <- ggplot(paleo, aes(.data[[PALEO_X]], .data[[PALEO_Y]])) +
  geom_line(color = "#2E86AB") +
  {if (REVERSE_X) scale_x_reverse() else NULL} +
  labs(title = "Deep-time series", x = PALEO_X, y = PALEO_Y) +
  theme_minimal()

# Modern + historical peak
hist_max <- max(paleo[[PALEO_Y]], na.rm = TRUE)
p2 <- ggplot(modern, aes(.data[[MODERN_X]], .data[[MODERN_Y]])) +
  geom_line(color = "#E67E22") +
  geom_hline(yintercept = hist_max, linetype = "dashed", color = "red") +
  labs(title = "Modern series + paleo maximum", x = MODERN_X, y = MODERN_Y) +
  theme_minimal()

print(p1); print(p2)
cat("Historical max used for line:", hist_max, "\n")
```
