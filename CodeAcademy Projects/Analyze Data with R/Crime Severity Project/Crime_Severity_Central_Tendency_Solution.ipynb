# Crime Severity Central Tendency Analysis (Mean, Median, Mode) — Solution

**Goal:** Calculate and interpret mean, median and mode of crime severity scores; visualize; adapt messaging for analysts vs. command staff.

**Data:** `data/crimes.csv` (40 records).

**Flowchart:** `crime_severity_central_tendency_flowchart.png`

![Flowchart](crime_severity_central_tendency_flowchart.png)


## Step 0 — Setup & Load Data


```r
crimes <- read.csv("data/crimes.csv")
head(crimes)
str(crimes)
summary(crimes$severity_score)
```

## Step 1 — Extract the Severity Vector


```r
severity <- crimes$severity_score
length(severity)
summary(severity)
```

## Step 2 — Mean


```r
# 2a Manual
total <- sum(severity)
n <- length(severity)
mean_manual <- total / n
print(mean_manual)

# 2b Built-in
mean_builtin <- mean(severity)
print(mean_builtin)
# Expected ~ 5.405
```

## Step 3 — Visualize with Mean Line


```r
hist(severity, breaks = 10, main = "Crime Severity Scores",
     xlab = "Severity Score", col = "lightblue", border = "white")
abline(v = mean_builtin, col = "red", lwd = 2, lty = 1)
legend("topright", legend = paste("Mean =", round(mean_builtin, 2)),
       col = "red", lwd = 2, bty = "n")
```

## Step 4 — Median


```r
# 4a small example
five_scores <- c(3.9, 2.0, 8.1, 4.8, 8.4)
sorted_five <- sort(five_scores)
median_five <- sorted_five[3]   # middle of 5
cat("Sorted:", sorted_five, "\n")
cat("Median of five:", median_five, "\n")

# 4b full data
median_severity <- median(severity)
print(paste("The median severity score is:", median_severity))
# Expected 5.1
```

## Step 5 — Histogram with Mean + Median


```r
hist(severity, breaks = 10, main = "Crime Severity — Mean & Median",
     xlab = "Severity Score", col = "lightblue", border = "white")
abline(v = mean_builtin, col = "red", lwd = 2, lty = 1)
abline(v = median_severity, col = "blue", lwd = 2, lty = 2)
legend("topright",
       legend = c(paste("Mean =", round(mean_builtin, 2)),
                  paste("Median =", round(median_severity, 2))),
       col = c("red", "blue"), lwd = 2, lty = c(1, 2), bty = "n")
```

## Step 6 — Mode


```r
# 6a table
freq <- table(severity)
max_count <- max(freq)
mode_candidates <- names(freq)[freq == max_count]
print(freq)
print(paste("Max count:", max_count))
print(mode_candidates)
# Several values appear twice → multimodal

# 6b custom function
get_mode <- function(x) {
  ux <- unique(x)
  ux[which.max(tabulate(match(x, ux)))]
}
mode_severity <- get_mode(severity)
print(paste("Primary mode severity (first max):", mode_severity))
```

## Step 7 — Full Histogram with Mean, Median, Mode


```r
hist(severity, breaks = 10,
     main = "Crime Severity Scores with Central Tendency",
     xlab = "Severity Score", col = "lightblue", border = "white")
abline(v = mean_builtin, col = "red", lwd = 2, lty = 1)
abline(v = median_severity, col = "blue", lwd = 2, lty = 2)
abline(v = mode_severity, col = "darkgreen", lwd = 2, lty = 1)
legend("topright",
       legend = c(paste("Mean =", round(mean_builtin, 2)),
                  paste("Median =", round(median_severity, 2)),
                  paste("Mode =", mode_severity)),
       col = c("red", "blue", "darkgreen"), lwd = 2,
       lty = c(1, 2, 1), bty = "n")
mtext("Multimodal distribution — several scores appear twice", side = 3, line = 0.2, cex = 0.8)
```

## Step 8 — Audience-Aware Interpretation

### Analyst version (technician / high data literacy)
The severity distribution is slightly right-skewed (mean 5.41 > median 5.1). Multiple values share the highest frequency (mode is not unique). For prioritisation models that are sensitive to the centre, the median is the more robust single number; the mean is pulled upward by the upper tail of high-severity cases (homicides / high-severity assaults). Visual inspection of the histogram confirms a relatively flat central region with several small peaks.

### Commander version (executive / short attention)
**Headline:** Typical crime severity sits around 5.1 (median) on our 1–10 scale; the average is a little higher (5.4) because of a handful of very serious incidents.  
**So-what:** Resource planning should not be driven by the extreme cases alone; the bulk of reported crimes are mid-severity. Focus extra attention on the high-severity tail while keeping baseline capacity sized to the median caseload.


```r
cat("ANALYST VERSION:\n")
cat("The severity distribution is slightly right-skewed (mean 5.41 > median 5.1). Multiple values share the highest frequency (mode is not unique). For prioritisation models that are sensitive to the centre, the median is the more robust single number; the mean is pulled upward by the upper tail of high-severity cases.\n")

cat("\nCOMMANDER VERSION:\n")
cat("Headline: Typical crime severity sits around 5.1 (median) on our 1-10 scale; the average is a little higher (5.4) because of a handful of very serious incidents.\n")
cat("So-what: Resource planning should not be driven by the extreme cases alone; the bulk of reported crimes are mid-severity.\n")
```

```r
# Alternate 1 — mean
mean_alt <- sum(severity) / length(severity)
print(c(mean_builtin = mean(severity), mean_alt = mean_alt))

# Alternate 2 — median for even n
sorted_sev <- sort(severity)
n <- length(sorted_sev)
median_alt <- (sorted_sev[n/2] + sorted_sev[n/2 + 1]) / 2
print(c(median_builtin = median(severity), median_alt = median_alt))

# Alternate 3 — mode
mode_alt <- as.numeric(names(sort(table(severity), decreasing = TRUE))[1])
print(c(mode_custom = mode_severity, mode_alt = mode_alt))
```

## Simulation Section — Solution


```r
set.seed(42)
noise_sd <- 0.5
n_sims   <- 200

sim_means   <- numeric(n_sims)
sim_medians <- numeric(n_sims)
sim_modes   <- numeric(n_sims)

get_mode <- function(x) {
  ux <- unique(x)
  ux[which.max(tabulate(match(x, ux)))]
}

for (i in 1:n_sims) {
  noisy <- severity + rnorm(length(severity), 0, noise_sd)
  sim_means[i]   <- mean(noisy)
  sim_medians[i] <- median(noisy)
  sim_modes[i]   <- get_mode(round(noisy, 1))
}

cat("Summary of simulated means:\n")
print(summary(sim_means))
cat("\nSummary of simulated medians:\n")
print(summary(sim_medians))
cat("\nSummary of simulated modes (rounded):\n")
print(summary(sim_modes))

par(mfrow = c(1, 3))
hist(sim_means, main = "Simulated Means", col = "salmon", border = "white",
     xlab = paste("noise_sd =", noise_sd))
abline(v = mean(severity), col = "darkred", lwd = 2)
hist(sim_medians, main = "Simulated Medians", col = "lightblue", border = "white")
abline(v = median(severity), col = "darkblue", lwd = 2)
hist(sim_modes, main = "Simulated Modes (rounded)", col = "lightgreen", border = "white")
abline(v = mode_severity, col = "darkgreen", lwd = 2)
par(mfrow = c(1, 1))

# Robustness observation
cat("\nSD of sim means:", sd(sim_means),
    " | SD of sim medians:", sd(sim_medians), "\n")
cat("Median is typically more stable under moderate noise for skewed data.\n")
```

## Key Takeaways (Solution)

- Mean ≈ 5.41, Median = 5.1, Mode is multimodal (several values appear twice; primary returned by the function is the first max).
- Slight right skew → mean > median; prefer median for a “typical” severity when communicating to decision makers.
- Audience adaptation matters: analysts get distributional detail; commanders get a headline + actionable so-what.
- Simulation shows the median’s sampling distribution is usually tighter than the mean’s under additive noise for this modestly skewed sample.
