## Provided Function


```r
calculate_annual_growth <- function(year_one, year_two, value_y1, value_y2, entity) {
  annual_growth <- (((value_y2 - value_y1) / value_y1) * 100) / (year_two - year_one)
  message <- paste("From", year_one, "to", year_two, "the loan portfolio of", entity,
                   "grew by approximately", round(annual_growth, 4), "% each year.")
  print(message)
  return(annual_growth)
}
```

## 1. Creating Variables — Solutions


```r
# Task 1
bank_name <- "Apex National Bank"
print(bank_name)
# Expected: [1] "Apex National Bank"
```

```r
# Task 2
loans_year_one <- 28400
print(loans_year_one)
# Expected: [1] 28400
```

```r
# Task 3
loans_year_two <- 142600
print(loans_year_two)
# Expected: [1] 142600
```

## 2. Calculations — Solutions


```r
# Task 4 — absolute change
loan_change <- loans_year_two - loans_year_one
print(loan_change)
# Expected: [1] 114200
```

```r
# Task 5 — percentage growth rate
percentage_gr <- (loan_change / loans_year_one) * 100
print(percentage_gr)
# Expected: [1] 402.1127 (approx)
```

```r
# Task 6 — annual growth rate (arithmetic)
years_elapsed <- 2025 - 2005   # 20
annual_gr <- percentage_gr / years_elapsed
print(annual_gr)
# Expected: [1] 20.10563 (approx)
```

```r
# Task 7 — explanatory print
print(paste(bank_name, "loan book grew by approximately",
            round(annual_gr, 2), "% per year between 2005 and 2025."))
```

## 3. Function Call — Solution


```r
# Task 8
result <- calculate_annual_growth(
  year_one = 2005,
  year_two = 2025,
  value_y1 = loans_year_one,
  value_y2 = loans_year_two,
  entity   = bank_name
)
# Expected console output:
# [1] "From 2005 to 2025 the loan portfolio of Apex National Bank grew by approximately 20.1056 % each year."
print(result)
```

## 4. Alternate Approaches


```r
# Alternate 1 — Vectorised
years <- c(2005, 2025)
loans <- c(28400, 142600)
annual_vec <- (((loans[2] - loans[1]) / loans[1]) * 100) / (years[2] - years[1])
print(annual_vec)
```

```r
# Alternate 2 — Data frame + period rates
apex <- data.frame(
  year  = c(2005, 2012, 2019, 2025),
  loans = c(28400, 61750, 98200, 142600)
)
apex$growth_pct <- c(NA, diff(apex$loans) / apex$loans[-nrow(apex)] * 100)
apex$years_span <- c(NA, diff(apex$year))
apex$annual_pct <- apex$growth_pct / apex$years_span
print(apex)
```

```r
# Alternate 3 — One-liner
annual_one_liner <- (((142600 - 28400) / 28400) * 100) / (2025 - 2005)
print(annual_one_liner)
```

```r
# Alternate 4 — Geometric / CAGR (preferred in banking)
cagr <- (142600 / 28400)^(1 / 20) - 1
print(cagr * 100)   # ≈ 8.40 %
# Arithmetic (~20 %) overstates long-run compounding; risk teams almost always use CAGR.
```

## 5. More Practice — Answers


```r
# Practice A — 2012 → 2019
loans_2012 <- 61750
loans_2019 <- 98200
annual_12_19 <- (((loans_2019 - loans_2012) / loans_2012) * 100) / (2019 - 2012)
print(annual_12_19)
# Expected ≈ 8.43 %
```

```r
# Practice B — 2019 → 2025
loans_2019 <- 98200
loans_2025 <- 142600
annual_19_25 <- (((loans_2025 - loans_2019) / loans_2019) * 100) / (2025 - 2019)
print(annual_19_25)
# Expected ≈ 7.54 %
```

```r
# Practice C
# The full 2005–2025 span shows the highest arithmetic rate (~20 %) because of the low
# starting base. The mid periods (2012–2019 and 2019–2025) are more stable around 7–8.5 %.
# This pattern is typical after the post-GFC recovery and subsequent digital-lending expansion.
# Risk managers therefore rely on CAGR (~8.4 %) rather than the arithmetic average when
# setting long-term growth assumptions for capital planning.
```

## 6. Simulation Section (Ready to Tweak)


```r
# === SIMULATION CONTROLS ===
sim_year_one <- 2005
sim_year_two <- 2025
sim_loans_y1 <- 28400
sim_loans_y2 <- 142600
sim_bank     <- "Apex National Bank"
# ===========================

sim_result <- calculate_annual_growth(
  year_one = sim_year_one,
  year_two = sim_year_two,
  value_y1 = sim_loans_y1,
  value_y2 = sim_loans_y2,
  entity   = sim_bank
)

geo_rate <- ((sim_loans_y2 / sim_loans_y1)^(1 / (sim_year_two - sim_year_one)) - 1) * 100
cat("\nGeometric annual rate / CAGR (%):", round(geo_rate, 4), "\n")
cat("Arithmetic annual rate (%):       ", round(sim_result, 4), "\n")
```

**Key Takeaways from the Solution**

- R uses `<-` (or `=`) for assignment; the left-arrow is idiomatic.
- Arithmetic operators `+ - * /` and parentheses work exactly as in mathematics.
- `paste()` concatenates strings and numbers into readable messages for management reports.
- Functions with named arguments improve clarity and reduce positional errors.
- The arithmetic average growth rate is simple to code but can be misleading over long horizons; banking risk teams almost always prefer the geometric rate (CAGR).
- Variables make code reusable: swap the numbers and the same logic serves any bank, any product portfolio, or any time window.
