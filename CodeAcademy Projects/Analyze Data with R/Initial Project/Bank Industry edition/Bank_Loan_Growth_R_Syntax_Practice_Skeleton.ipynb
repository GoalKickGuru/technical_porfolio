## Provided Function
The notebook already defines `calculate_annual_growth()`. You will call it at the end.
```r
calculate_annual_growth <- function(year_one, year_two, value_y1, value_y2, entity) {
  annual_growth <- (((value_y2 - value_y1) / value_y1) * 100) / (year_two - year_one)
  message <- paste("From", year_one, "to", year_two, "the loan portfolio of", entity,
                   "grew by approximately", round(annual_growth, 4), "% each year.")
  print(message)
  return(annual_growth)
}
```

````r
# === SIMULATION CONTROLS (edit these) ===
sim_year_one <- 2005
sim_year_two <- 2025
sim_loans_y1 <- 28400
sim_loans_y2 <- 142600
sim_bank     <- "Apex National Bank"
# =======================================

# Run the calculation with the current parameters
sim_result <- calculate_annual_growth(
  year_one = sim_year_one,
  year_two = sim_year_two,
  value_y1 = sim_loans_y1,
  value_y2 = sim_loans_y2,
  entity   = sim_bank
)

# Geometric (CAGR) for comparison — preferred in banking
geo_rate <- ((sim_loans_y2 / sim_loans_y1)^(1 / (sim_year_two - sim_year_one)) - 1) * 100
cat("Geometric annual rate / CAGR (%):", round(geo_rate, 4), "\n")
cat("Arithmetic annual rate (%):       ", round(sim_result, 4), "\n")
````
