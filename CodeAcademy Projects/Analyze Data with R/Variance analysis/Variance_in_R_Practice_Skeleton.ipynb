## Step 4 — R's Built-in `var()` (Sample Variance)

R's `var()` uses the **sample** formula (divide by $N-1$). This is the unbiased estimator when the data are a sample from a larger population.

$$
s^2 = \frac{{1}}{{N-1}}\sum_{{i=1}}^{{N}}(X_i - \bar{{x}})^2
$$
