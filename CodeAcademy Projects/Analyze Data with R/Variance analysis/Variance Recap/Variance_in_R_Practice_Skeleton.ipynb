## Part B — Why Standard Deviation?

Variance is in *squared* units (inches², grade-points²). That makes it hard to say “how far from the mean is this value?”  

Standard deviation restores the original units:

$$
\sigma = \sqrt{\sigma^2} = \sqrt{\frac{1}{N}\sum (X_i - \mu)^2}
$$

## Part C — Using Standard Deviation (z-scores)

A z-score tells you *how many standard deviations* a point is from the mean:

$$
z = \frac{x - \mu}{\sigma}
$$

Rough empirical rule (approx. normal data):
- ~68 % of observations lie within ±1 SD of the mean
- ~95 % within ±2 SD
- ~99.7 % within ±3 SD

A point beyond ±3 SD is extremely unusual.
