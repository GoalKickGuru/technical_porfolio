## Part A — Variance Recap (Teachers)
### Step A0 — Load & visualise


```r
teacher_one_grades <- c(83.42, 88.04, 82.12, 85.02, 82.52, 87.47, 84.69, 85.18, 86.29, 85.53, 81.29, 82.54, 83.47, 83.91, 86.83, 88.5, 84.95, 83.79, 84.74, 84.03, 87.62, 81.15, 83.45, 80.24, 82.76, 83.98, 84.95, 83.37, 84.89, 87.29)
teacher_two_grades <- c(85.15, 95.64, 84.73, 71.46, 95.99, 81.61, 86.55, 79.81, 77.06, 92.86, 83.67, 73.63, 90.12, 80.64, 78.46, 76.86, 104.4, 88.53, 74.62, 91.27, 76.53, 94.37, 84.74, 81.84, 97.69, 70.77, 84.44, 88.06, 91.62, 65.82)
variance_pop <- function(x) mean((x - mean(x))^2)

cat("Teacher 1 mean:", round(mean(teacher_one_grades),2),
    " pop var:", round(variance_pop(teacher_one_grades),2), "\n")
cat("Teacher 2 mean:", round(mean(teacher_two_grades),2),
    " pop var:", round(variance_pop(teacher_two_grades),2), "\n")

hist(teacher_one_grades, col=rgb(0,0,1,0.25), xlim=c(65,105),
     main="Teacher Grades", xlab="Grade", breaks=15)
hist(teacher_two_grades, col=rgb(1,0,0,0.25), add=TRUE, breaks=15)
legend("topright", c("Teacher 1","Teacher 2"), fill=c(rgb(0,0,1,0.25),rgb(1,0,0,0.25)))
box()
```

### Step A1–A2 — Variance function & sample comparison


```r
grades <- c(88, 82, 85, 84, 90)
print(variance_pop(grades))
print(var(grades))   # sample
print(variance_pop(grades) * length(grades)/(length(grades)-1))
```

## Part B — Standard Deviation
### Step B1 — From variance to SD


```r
t1_sd_pop <- sqrt(variance_pop(teacher_one_grades))
t2_sd_pop <- sqrt(variance_pop(teacher_two_grades))
cat("Teacher 1 pop SD:", round(t1_sd_pop,2), "  sample sd():", round(sd(teacher_one_grades),2), "\n")
cat("Teacher 2 pop SD:", round(t2_sd_pop,2), "  sample sd():", round(sd(teacher_two_grades),2), "\n")
```

## Part C — z-scores
### Step C1 — LeBron (80 in)


```r
lebron <- 80
nba_mean <- mean(nba_data); okc_mean <- mean(okcupid_data)
nba_sd   <- sd(nba_data);   okc_sd   <- sd(okcupid_data)

nba_diff <- lebron - nba_mean
okc_diff <- lebron - okc_mean
num_nba_deviations <- nba_diff / nba_sd
num_okc_deviations <- okc_diff / okc_sd

cat("LeBron is", round(num_nba_deviations,2), "SD above the NBA mean\n")
cat("LeBron is", round(num_okc_deviations,2), "SD above the OkCupid mean\n")
cat("→ Tall but ordinary among NBA players; extremely tall on OkCupid.\n")
```

### Step C2 — Earl Boykins (65 in)


```r
boykins <- 65
nba_diff_b <- boykins - nba_mean
okc_diff_b <- boykins - okc_mean
num_nba_b <- nba_diff_b / nba_sd
num_okc_b <- okc_diff_b / okc_sd

cat("Boykins is", round(num_nba_b,2), "SD from the NBA mean (very short for NBA)\n")
cat("Boykins is", round(num_okc_b,2), "SD from the OkCupid mean\n")
cat("→ Far more unusual in the NBA than on OkCupid because the OkCupid distribution is wider and centred lower.\n")
```

### Step C3 — Empirical rule check on the data


```r
emp_rule <- function(x) {
  m <- mean(x); s <- sd(x)
  c("within_1sd" = mean(abs(x-m) <= s),
    "within_2sd" = mean(abs(x-m) <= 2*s),
    "within_3sd" = mean(abs(x-m) <= 3*s))
}
print(round(emp_rule(nba_data), 3))
print(round(emp_rule(okcupid_data), 3))
```

## Alternate Implementations


```r
# Population SD
sd_pop <- function(x) sqrt(mean((x-mean(x))^2))
# Vectorised z-scores (sample SD)
z_scores <- function(x) (x - mean(x)) / sd(x)
# Using scale()
z_scale <- function(x) as.numeric(scale(x))

print(c(pop=sd_pop(nba_data), sample=sd(nba_data)))
print(head(z_scores(nba_data), 5))
```

## Simulation — Empirical rule coverage


```r
set.seed(42)
true_mean <- 70; true_sd <- 4; n <- 200; n_sims <- 300

cover <- t(replicate(n_sims, {
  x <- rnorm(n, true_mean, true_sd)
  m <- mean(x); s <- sd(x)
  c(mean(abs(x-m)<=s), mean(abs(x-m)<=2*s), mean(abs(x-m)<=3*s))
}))
colnames(cover) <- c("1SD","2SD","3SD")
print(round(colMeans(cover), 3))
cat("Theoretical targets: 0.68 / 0.95 / 0.997\n")

hist(cover[,"1SD"], breaks=20, col="steelblue", border="white",
     main="Monte-Carlo: empirical coverage within ±1 SD",
     xlab="Proportion")
abline(v=0.68, col="red", lwd=2, lty=2)
```

## Key Takeaways
- Variance units are **squared**; SD restores the **original** units of the data.
- z = (x − μ) / σ measures **how many SDs a point is from the mean**.
- Points beyond ±3 SD are **extremely unusual** (~0.3 % of a normal distribution).
- The same absolute difference can be ordinary in a wide distribution and extreme in a narrow one.
- Prefer SD (or z-scores) over raw variance when communicating spread to any audience.
