## Alternate Implementations (same result)
You do **not** have to rewrite the original function, but try these more idiomatic R approaches and compare the results.

### A. `match()` / `which()`
```r
find_place_match <- function(runner) {
  idx <- match(runner, race_results)
  if (is.na(idx)) length(race_results) + 1 else idx
}
```

### B. Named vector lookup
```r
# create a named place vector once
place_lookup <- setNames(seq_along(race_results), race_results)
find_place_named <- function(runner) {
  p <- place_lookup[runner]
  if (is.na(p)) length(race_results) + 1 else as.integer(p)
}
```
