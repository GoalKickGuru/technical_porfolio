## Alternate Techniques
### A. Bootstrap (resample existing data)
If you already have a vector of observed places, you can resample it with replacement to obtain a sampling distribution of a statistic (e.g. mean place).

### B. Permutation / randomization test idea
Shuffle a treatment label (or a grouping factor) many times to obtain a null distribution of a test statistic.

### C. Parametric bootstrap
Fit a model, then simulate new data from the fitted model and re-estimate.
