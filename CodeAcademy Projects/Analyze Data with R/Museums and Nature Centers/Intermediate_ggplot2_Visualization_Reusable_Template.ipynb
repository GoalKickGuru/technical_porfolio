# Intermediate ggplot2 Visualization — Reusable Template

Use this notebook as a starting point for any project that needs:
- distribution views (histogram / density)
- comparison of groups (box / violin / bar)
- composition views (stacked / filled bars)
- multi-panel comparison (facet_grid / facet_wrap)
- polished axes, dollar/percent formatting, error bars

Replace the placeholder data and column names with your own.

```r
library(dplyr)
library(ggplot2)
library(scales)
library(plotrix)   # optional — std.error()

# ---- 1. Load & quick audit ----
df <- read.csv("path/to/your_data.csv")   # <-- change
glimpse(df)
summary(df$your_numeric_col)              # <-- change
```

## 2. Distribution of a continuous variable

```r
ggplot(df, aes(x = your_numeric_col)) +
  geom_histogram(binwidth = NULL,          # or set an explicit value
                 fill = "#e67e22", colour = "white") +
  scale_x_continuous(labels = dollar_format()) +  # or comma / percent
  labs(title = "Distribution of …", x = "…", y = "Count") +
  theme_minimal()
```

## 3. Group comparison — box / bar of means

```r
# Box plot
ggplot(df, aes(x = factor(group_col), y = your_numeric_col)) +
  geom_boxplot(fill = "#9b59b6", alpha = 0.7) +
  scale_y_continuous(labels = dollar_format()) +
  labs(title = "… by group", x = "Group", y = "Value") +
  theme_minimal()

# Mean bar + error bars
summary_df <- df %>%
  group_by(group_col) %>%
  summarise(Mean = mean(your_numeric_col, na.rm = TRUE),
            SE   = std.error(your_numeric_col),
            .groups = "drop") %>%
  mutate(ymin = Mean - SE, ymax = Mean + SE)

ggplot(summary_df, aes(x = factor(group_col), y = Mean)) +
  geom_col(fill = "#27ae60") +
  geom_errorbar(aes(ymin = ymin, ymax = ymax), width = 0.2) +
  scale_y_continuous(labels = dollar_format()) +
  labs(title = "Mean … by group (with SE)") +
  theme_minimal()
```

## 4. Composition — stacked / filled bars

```r
ggplot(df, aes(x = factor(group_col), fill = factor(category_col))) +
  geom_bar(position = "fill") +                 # or "stack" for counts
  scale_y_continuous(labels = percent_format()) +
  labs(title = "Composition of …", x = "Group", y = "Share", fill = "Category") +
  theme_minimal() +
  theme(axis.text.x = element_text(angle = 30, hjust = 1))
```

## 5. Faceting

```r
ggplot(df, aes(x = factor(category_col))) +
  geom_bar() +
  facet_wrap(~ group_col) +                     # or facet_grid(rows = vars(..), cols = vars(..))
  labs(title = "… by subgroup") +
  theme_minimal()
```
