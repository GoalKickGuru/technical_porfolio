## Audience Adaptation Notes

Before finalising any plot for a real stakeholder, ask:

| Audience | Data literacy | Preferred chart | Label style | Extra guidance |
|----------|---------------|-----------------|-------------|----------------|
| Executive / board | Low–medium | Simple stacked % or mean bar with error bars | Plain English, $M / $B | One headline insight in the title |
| Analyst / data team | High | Box plot, histogram, facet grid | Technical but clean | Keep method notes in caption or appendix |
| General public / press | Low | Fill-bar % or simple counts | Avoid jargon, short labels | Explain “why it matters” in a subtitle |

Refer to the attached PDFs *What to Consider When Considering the Audience* and *Audience and Situation Analysis* for a fuller checklist (data literacy, subject knowledge, mixed audiences, cultural conventions).
