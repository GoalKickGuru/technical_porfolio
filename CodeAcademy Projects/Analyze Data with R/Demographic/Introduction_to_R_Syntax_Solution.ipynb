## Provided Function


```r
calculate_annual_growth <- function(year_one, year_two, pop_y1, pop_y2, city) {
  annual_growth <- (((pop_y2 - pop_y1) / pop_y1) * 100) / (year_two - year_one)
  message <- paste("From", year_one, "to", year_two, "the population of", city,
                   "grew by approximately", annual_growth, "% each year.")
  print(message)
  return(annual_growth)
}
```

## 1. Creating Variables — Solutions


```r
# Task 1
city_name <- "Istanbul, Turkey"
print(city_name)
# Expected: [1] "Istanbul, Turkey"
```

```r
# Task 2
pop_year_one <- 691000
print(pop_year_one)
# Expected: [1] 691000
```

```r
# Task 3
pop_year_two <- 15029231
print(pop_year_two)
# Expected: [1] 15029231
```

## 2. Calculations — Solutions


```r
# Task 4 — absolute change
pop_change <- pop_year_two - pop_year_one
print(pop_change)
# Expected: [1] 14338231
```

```r
# Task 5 — percentage growth rate
percentage_gr <- (pop_change / pop_year_one) * 100
# equivalent: percentage_gr <- (((pop_year_two - pop_year_one) / pop_year_one) * 100)
print(percentage_gr)
# Expected: [1] 2074.997 (approx)
```

```r
# Task 6 — annual growth rate (arithmetic)
years_elapsed <- 2017 - 1927   # 90
annual_gr <- percentage_gr / years_elapsed
print(annual_gr)
# Expected: [1] 23.05553 (approx)
```

```r
# Task 7 — explanatory print
print(paste("Istanbul grew by approximately", round(annual_gr, 2),
            "% per year between 1927 and 2017."))
# or simply:
# print(annual_gr)
```

## 3. Function Call — Solution


```r
# Task 8
result <- calculate_annual_growth(
  year_one = 1927,
  year_two = 2017,
  pop_y1   = pop_year_one,
  pop_y2   = pop_year_two,
  city     = city_name
)
# Expected console output:
# [1] "From 1927 to 2017 the population of Istanbul, Turkey grew by approximately 23.055525 % each year."
print(result)
```

## 4. Alternate Approaches


```r
# Alternate 1 — Vectorised
years <- c(1927, 2017)
pops  <- c(691000, 15029231)
annual_vec <- (((pops[2] - pops[1]) / pops[1]) * 100) / (years[2] - years[1])
print(annual_vec)
```

```r
# Alternate 2 — Data frame + period rates
istanbul <- data.frame(
  year = c(1927, 1950, 2000, 2017),
  pop  = c(691000, 983000, 8831800, 15029231)
)
istanbul$growth_pct <- c(NA, diff(istanbul$pop) / istanbul$pop[-nrow(istanbul)] * 100)
istanbul$years_span <- c(NA, diff(istanbul$year))
istanbul$annual_pct <- istanbul$growth_pct / istanbul$years_span
print(istanbul)
```

```r
# Alternate 3 — One-liner (still readable)
annual_one_liner <- (((15029231 - 691000) / 691000) * 100) / (2017 - 1927)
print(annual_one_liner)
```

```r
# Alternate 4 — Geometric mean (more accurate for compounding)
geo_annual <- (15029231 / 691000)^(1 / 90) - 1
print(geo_annual * 100)   # ≈ 3.48 % per year
# Contrast with the arithmetic ~23 % which overstates long-run compounding.
```

## 5. More Practice — Answers


```r
# Practice A — 1950 → 2000
pop_1950 <- 983000
pop_2000 <- 8831800
annual_50_00 <- (((pop_2000 - pop_1950) / pop_1950) * 100) / (2000 - 1950)
print(annual_50_00)
# Expected ≈ 15.97 %
```

```r
# Practice B — 2000 → 2017
pop_2000 <- 8831800
pop_2017 <- 15029231
annual_00_17 <- (((pop_2017 - pop_2000) / pop_2000) * 100) / (2017 - 2000)
print(annual_00_17)
# Expected ≈ 4.13 %
```

```r
# Practice C
# The 1950–2000 period shows the highest arithmetic annual growth (~16 %),
# reflecting rapid urbanisation after mid-century. The full 1927–2017 span
# is dominated by the early base effect and yields the inflated ~23 % figure.
# Geometric rates tell a smoother story of long-run compounding around 3–4 %.
```

## 6. Simulation Section (Ready to Tweak)


```r
# === SIMULATION CONTROLS ===
sim_year_one <- 1927
sim_year_two <- 2017
sim_pop_y1   <- 691000
sim_pop_y2   <- 15029231
sim_city     <- "Istanbul, Turkey"
# ===========================

sim_result <- calculate_annual_growth(
  year_one = sim_year_one,
  year_two = sim_year_two,
  pop_y1   = sim_pop_y1,
  pop_y2   = sim_pop_y2,
  city     = sim_city
)

geo_rate <- ((sim_pop_y2 / sim_pop_y1)^(1 / (sim_year_two - sim_year_one)) - 1) * 100
cat("\nGeometric annual rate (%):", round(geo_rate, 4), "\n")
cat("Arithmetic annual rate (%):", round(sim_result, 4), "\n")
```

**Key Takeaways from the Solution**

- R uses `<-` (or `=`) for assignment; the left-arrow is idiomatic.
- Arithmetic operators `+ - * /` and parentheses work exactly as in mathematics.
- `paste()` concatenates strings and numbers into readable messages.
- Functions are first-class; calling them with named arguments improves clarity.
- The arithmetic average growth rate is simple to compute but can be misleading over long horizons; always consider the geometric alternative for demographic work.
- Variables make code reusable: swap the numbers and the same logic serves any city or period.
