# NYC One-Bedroom Rents — Central Tendency (Mean, Median, Mode) — Solution

**Goal:** Calculate and interpret the mean, median, and mode of one-bedroom apartment rents in Brooklyn, Manhattan, and Queens. Compare cost of living and practice audience-aware messaging.

**Data:** `data/brooklyn-one-bed.csv`, `data/manhattan-one-bed.csv`, `data/queens-one-bed.csv`.

## Step 0 — Setup & Load Data


```r
brooklyn_one_bed <- read.csv("data/brooklyn-one-bed.csv")
manhattan_one_bed <- read.csv("data/manhattan-one-bed.csv")
queens_one_bed <- read.csv("data/queens-one-bed.csv")

brooklyn_price <- brooklyn_one_bed$rent
manhattan_price <- manhattan_one_bed$rent
queens_price <- queens_one_bed$rent

cat("Brooklyn n =", length(brooklyn_price), "\n")
summary(brooklyn_price)
cat("\nManhattan n =", length(manhattan_price), "\n")
summary(manhattan_price)
cat("\nQueens n =", length(queens_price), "\n")
summary(queens_price)
```

## Step 1 — Mean


```r
brooklyn_mean  <- mean(brooklyn_price)
manhattan_mean <- mean(manhattan_price)
queens_mean    <- mean(queens_price)

cat(sprintf("Brooklyn mean:  $%.2f\n", brooklyn_mean))
cat(sprintf("Manhattan mean: $%.2f\n", manhattan_mean))
cat(sprintf("Queens mean:    $%.2f\n", queens_mean))
```

## Step 2 — Median


```r
brooklyn_median  <- median(brooklyn_price)
manhattan_median <- median(manhattan_price)
queens_median    <- median(queens_price)

cat("Brooklyn median:  $", brooklyn_median, "\n")
cat("Manhattan median: $", manhattan_median, "\n")
cat("Queens median:    $", queens_median, "\n")
```

## Step 3 — Mode


```r
# Using table() (no extra package required)
get_mode <- function(x) {
  freq <- table(x)
  as.numeric(names(freq)[which.max(freq)])
}

brooklyn_mode  <- get_mode(brooklyn_price)
manhattan_mode <- get_mode(manhattan_price)
queens_mode    <- get_mode(queens_price)

cat("Brooklyn mode:  $", brooklyn_mode, "\n")
cat("Manhattan mode: $", manhattan_mode, "\n")
cat("Queens mode:    $", queens_mode, "\n")

# Optional: DescTools::Mode (returns first mode if multimodal)
# library(DescTools)
# Mode(brooklyn_price)
```

## Step 4 — Comparison Table


```r
comparison <- data.frame(
  Borough = c("Brooklyn", "Manhattan", "Queens"),
  Mean    = round(c(brooklyn_mean, manhattan_mean, queens_mean), 2),
  Median  = c(brooklyn_median, manhattan_median, queens_median),
  Mode    = c(brooklyn_mode, manhattan_mode, queens_mode),
  stringsAsFactors = FALSE
)
print(comparison)

cat("\nOrdering check (Mode <= Median <= Mean) for each borough:\n")
for(i in 1:3){
  cat(comparison$Borough[i], ": ",
      comparison$Mode[i], "<=", comparison$Median[i], "<=", comparison$Mean[i], "\n")
}
```

```r
par(mfrow = c(1, 3), mar = c(4, 4, 3, 1))

hist(brooklyn_price, breaks = 20, col = "#3498db", border = "white",
     main = "Brooklyn", xlab = "Rent ($)", xlim = c(1000, 7000))
abline(v = brooklyn_mean,   col = "red",        lwd = 2)
abline(v = brooklyn_median, col = "darkgreen",  lwd = 2, lty = 2)
abline(v = brooklyn_mode,   col = "orange",     lwd = 2, lty = 3)
legend("topright", c("Mean", "Median", "Mode"),
       col = c("red", "darkgreen", "orange"), lty = 1:3, lwd = 2, cex = 0.7)

hist(manhattan_price, breaks = 20, col = "#9b59b6", border = "white",
     main = "Manhattan", xlab = "Rent ($)", xlim = c(1000, 7000))
abline(v = manhattan_mean,   col = "red",       lwd = 2)
abline(v = manhattan_median, col = "darkgreen", lwd = 2, lty = 2)
abline(v = manhattan_mode,   col = "orange",    lwd = 2, lty = 3)
legend("topright", c("Mean", "Median", "Mode"),
       col = c("red", "darkgreen", "orange"), lty = 1:3, lwd = 2, cex = 0.7)

hist(queens_price, breaks = 20, col = "#1abc9c", border = "white",
     main = "Queens", xlab = "Rent ($)", xlim = c(1000, 7000))
abline(v = queens_mean,   col = "red",       lwd = 2)
abline(v = queens_median, col = "darkgreen", lwd = 2, lty = 2)
abline(v = queens_mode,   col = "orange",    lwd = 2, lty = 3)
legend("topright", c("Mean", "Median", "Mode"),
       col = c("red", "darkgreen", "orange"), lty = 1:3, lwd = 2, cex = 0.7)

par(mfrow = c(1, 1))
```

## Simulation Section — Solution


```r
set.seed(42)
noise_sd <- 150
n_sims   <- 200

sim_means   <- numeric(n_sims)
sim_medians <- numeric(n_sims)
sim_modes   <- numeric(n_sims)

# mode helper that bins to nearest $50 for stability under noise
get_mode_binned <- function(x, bin = 50) {
  xb <- round(x / bin) * bin
  freq <- table(xb)
  as.numeric(names(freq)[which.max(freq)])
}

for (i in 1:n_sims) {
  noisy <- brooklyn_price + rnorm(length(brooklyn_price), mean = 0, sd = noise_sd)
  noisy <- pmax(noisy, 1000)
  sim_means[i]   <- mean(noisy)
  sim_medians[i] <- median(noisy)
  sim_modes[i]   <- get_mode_binned(noisy)
}

cat("Simulation (Brooklyn + N(0,", noise_sd, ") noise, ", n_sims, " runs)\n")
cat("Mean of simulated means:  ", round(mean(sim_means), 1), "\n")
cat("SD of simulated means:    ", round(sd(sim_means), 1), "\n")
cat("Mean of simulated medians:", round(mean(sim_medians), 1), "\n")
cat("SD of simulated medians:  ", round(sd(sim_medians), 1), "\n")
cat("Most frequent simulated mode:", names(sort(table(sim_modes), decreasing = TRUE))[1], "\n")

# Quick visual
par(mfrow = c(1, 3))
hist(sim_means,   col = "#e74c3c", border = "white", main = "Sim Means", xlab = "")
abline(v = brooklyn_mean, lwd = 2)
hist(sim_medians, col = "#27ae60", border = "white", main = "Sim Medians", xlab = "")
abline(v = brooklyn_median, lwd = 2)
hist(sim_modes,   col = "#f39c12", border = "white", main = "Sim Modes (binned)", xlab = "")
abline(v = brooklyn_mode, lwd = 2)
par(mfrow = c(1, 1))

cat("\nObservation: median is more stable than mean under additive noise;\n")
cat("raw mode is fragile; binning restores a usable modal estimate.\n")
```

## Key Takeaways

- **Mean** is sensitive to luxury outliers → systematically higher than the typical rent.
- **Median** is the preferred “typical rent” summary for skewed housing data.
- **Mode** tells you the single most common asking price (useful for search filters and market pulse).
- Consistent Mode ≤ Median ≤ Mean ordering signals right skew in all three boroughs.
- Always surface the assumptions (representativeness, quality comparability, time window) when ranking boroughs.
- Re-phrase the same three numbers for analysts (full detail + code), executives (median + implication), and the public (most-common-price story).
- Simulation shows median is robust; mode needs binning once noise is present.
