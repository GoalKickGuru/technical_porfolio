## 1. Vectors — The Building Blocks of Economic Time Series

Vectors are the most basic data structure in R. They hold elements of the **same type** (numeric, character, logical, …) and are one-dimensional.

### Theory quick-ref
- Create with `c(...)` or the colon operator `start:end`
- `length(v)` → number of elements
- `typeof(v)` → underlying type
- Indexing: `v[i]`, `v[c(1,3)]`, `v[1:4]`, logical `v[v > 3]`
- Modification: `v[i] <- new_value` or `v[condition] <- value`

my_list[[i]] <- value # modify
names(my_list) <- c("a","b")  # optional names

## 4. Data Frames — The Workhorse of Economic Tabular Data

A data frame is a special list in which every element is a vector of the **same length**. Columns can have different types. This is the structure that most economic datasets take after import.

### Theory
```r
df <- data.frame(
  col1 = c(...),
  col2 = c(...),
  stringsAsFactors = FALSE   # recommended in modern R
)
df$col1          # extract a column (returns a vector)
df[["col1"]]     # alternative
```
