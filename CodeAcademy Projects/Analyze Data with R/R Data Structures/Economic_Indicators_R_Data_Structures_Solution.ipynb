### Task 1.6 — Floor low GDP growth rates


```r
gdp_q <- c(1.2, 0.3, -0.4, 0.8, 0.1, 2.1, 0.4, 1.5)
# Logical indexing: set every element that is < 0.5 to 0.5
gdp_q[gdp_q < 0.5] <- 0.5
print(gdp_q)
# Expected: 1.2 0.5 0.5 0.8 0.5 2.1 0.5 1.5

# Alternate using pmax (vectorized)
gdp_q2 <- c(1.2, 0.3, -0.4, 0.8, 0.1, 2.1, 0.4, 1.5)
gdp_q2 <- pmax(gdp_q2, 0.5)
print(gdp_q2)
```

## 6. Simulation — Fully Worked Solution


```r
# Parameters (easy to edit for scenario analysis)
base_inflation   <- c(2.1, 2.4, 1.8, 1.2, 4.7, 8.0, 4.1, 3.4, 2.9, 2.5)
shock_year_index <- 6
shock_size       <- 1.5
floor_rate       <- 0.0

# Working copy
sim_inf <- base_inflation

# 1. Apply shock
sim_inf[shock_year_index] <- sim_inf[shock_year_index] + shock_size

# 2. Apply floor
sim_inf[sim_inf < floor_rate] <- floor_rate

# 3. Statistics
orig_mean <- mean(base_inflation)
new_mean  <- mean(sim_inf)
new_max   <- max(sim_inf)
delta     <- new_mean - orig_mean

cat("Original mean inflation :", round(orig_mean, 3), "\n")
cat("Shocked + floored mean  :", round(new_mean, 3), "\n")
cat("New maximum             :", new_max, "\n")
cat("Change in mean (pp)     :", round(delta, 3), "\n")
print(sim_inf)
```

### Simulation function (stretch solution)


```r
simulate_inflation <- function(base, idx, shock, floor = 0) {
  s <- base
  s[idx] <- s[idx] + shock
  s[s < floor] <- floor
  list(
    series   = s,
    mean     = mean(s),
    max      = max(s),
    delta    = mean(s) - mean(base)
  )
}

# Two scenarios
scen_mild   <- simulate_inflation(base_inflation, 6, shock = 1.0)
scen_severe <- simulate_inflation(base_inflation, 6, shock = 3.0)

cat("Mild shock mean  :", round(scen_mild$mean, 3), "  delta =", round(scen_mild$delta, 3), "\n")
cat("Severe shock mean:", round(scen_severe$mean, 3), "  delta =", round(scen_severe$delta, 3), "\n")
```

## Key Takeaways for Economists

1. **Vectors** are the natural home for any single economic time series (inflation, rates, volumes).
2. **Logical indexing** (`x[x < threshold] <- value`) is the cleanest way to impose floors, ceilings or replace outliers.
3. **Matrices** shine when you need homogeneous numeric tables (country × year growth rates) that will later feed linear algebra or simple heatmap visualisations.
4. **Lists** let you keep heterogeneous objects together — perfect for a policy briefing package that mixes text, numbers and flags.
5. **Data frames** are the standard structure for almost every economic dataset you will import or export; master the `$` and `[[` extractors early.
6. Always keep a clean copy of the original series before applying scenario shocks; this makes deltas transparent for decision makers.

These structures form the foundation for later work with packages such as `dplyr`, `tidyr` and `ggplot2`.
