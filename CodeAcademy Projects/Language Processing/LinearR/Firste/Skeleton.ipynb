## Cheat Sheet (also saved as separate .docx)

| Concept | Formula / Code |
|---------|----------------|
| Line | `yhat = m*x + b` |
| SSE / Loss | `sum( (y - yhat)**2 )` |
| Gradient b | `-2/N * sum(y - (m*x+b))` |
| Gradient m | `-2/N * sum(x * (y - (m*x+b)))` |
| Update | `param -= learning_rate * gradient` |
| Convergence | parameters (or loss) change very little |
| Learning rate too small | slow convergence |
| Learning rate too large | divergence / oscillation |
| sklearn | `LinearRegression().fit(X, y)` → `.coef_`, `.intercept_`, `.predict`, `.score` |
| Closed form (simple LR) | `m = cov(x,y)/var(x)`, `b = mean(y)-m*mean(x)` |
| Centering trick | run GD on `X-mean`, `y-mean`; recover `b = y.mean() - m*X.mean()` |
