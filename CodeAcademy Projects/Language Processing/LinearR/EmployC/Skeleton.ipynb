## Cheat Sheet (Employment Campaigns Linear Regression)

| Concept | Formula / Code | Workforce reading |
|---------|----------------|-------------------|
| Line | `yhat = m*x + b` | Predicted placements (or employment rate) |
| Slope `m` | marginal effect | Extra placements per extra $1k campaign spend; or pp gain per training week |
| Intercept `b` | baseline | Placements at zero spend (use cautiously) |
| SSE / Loss | `sum((y-yhat)**2)` | Total squared prediction error |
| Gradient b / m | see solution | Direction of steepest loss increase |
| Update | `param -= α * gradient` | Take a step downhill on the loss surface |
| Convergence | params almost stop changing | Algorithm has (approximately) minimised loss |
| sklearn | `LinearRegression().fit(X,y)` | Fast, reliable reference implementation |
| Closed form | `m = cov(x,y)/var(x)` | Exact OLS solution for simple LR |
| Centering | GD on demeaned data; recover `b` | Numerically stable when scales differ |
