```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
plt.style.use('seaborn-v0_8')
```

## Problem Statement - Healthcare Context

| Age | Annual Healthcare Cost per Person ($1000s) |
|-----|-------------------------------------------|
| 35  | 4.2                                       |
| 45  | 6.8                                       |
| 55  | 9.5                                       |
| 65  | 13.2                                      |
| 75  | 18.0                                      |

```python
x_train = np.array([35, 45, 55, 65, 75])
y_train = np.array([4.2, 6.8, 9.5, 13.2, 18.0])
```

## Core Functions: Cost & Gradient

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0
    for i in range(m):
        f_wb = w * x[i] + b
        cost += (f_wb - y[i])**2
    return cost / (2 * m)

def compute_gradient(x, y, w, b):
    m = x.shape[0]
    dj_dw = 0
    dj_db = 0
    for i in range(m):
        f_wb = w * x[i] + b
        error = f_wb - y[i]
        dj_dw += error * x[i]
        dj_db += error
    return dj_dw / m, dj_db / m
```

## Gradient Descent Implementation

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters):
    w, b = w_in, b_in
    J_history = []
    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(x, y, w, b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        cost = compute_cost(x, y, w, b)
        J_history.append(cost)
        if i % 200 == 0:
            print(f"Iter {i:4d}: Cost {cost:.4f}, w={w:.4f}, b={b:.4f}")
    return w, b, J_history
```

```python
# Run with good learning rate
w_final, b_final, costs = gradient_descent(x_train, y_train, 0, 0, 0.0001, 3000)
print(f"\nOptimal parameters: w = {w_final:.4f}, b = {b_final:.4f}")
```

## Visualizations & Diagnostics

```python
plt.figure(figsize=(10,6))
plt.plot(costs)
plt.title('Cost Reduction During Gradient Descent (Healthcare Costs)')
plt.xlabel('Iteration')
plt.ylabel('Cost J(w,b)')
plt.grid(True)
plt.show()
```

## Feature Scaling (Best Practice)

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x_train.reshape(-1,1)).flatten()

w_s, b_s, costs_s = gradient_descent(x_scaled, y_train, 0, 0, 0.1, 1000)
print("Much faster convergence with scaling!")
```

## Real-World Interpretation

- **Slope (`w`)**: Additional cost per year of age
