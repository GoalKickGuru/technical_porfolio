## Tools

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error
plt.style.use('seaborn-v0_8')
```

## Problem Statement - Climate Crisis Context

Many countries are increasing investment in renewable energy (solar, wind, etc.) to reduce CO₂ emissions.

| Renewable Investment ($ Billion) | CO₂ Emissions Reduction (Million Tons) |
|----------------------------------|----------------------------------------|
| 10                               | 45                                     |
| 25                               | 120                                    |
| 40                               | 180                                    |
| 55                               | 230                                    |

**Goal**: Find the best $w$ and $b$ so our model can advise governments on how much to invest for maximum emissions reduction.

```python
# Training Data
x_train = np.array([10, 25, 40, 55])      # Investment in $ Billion
y_train = np.array([45, 120, 180, 230])   # CO₂ Reduction in Million Tons

print("Investment (x):", x_train)
print("Reduction (y):", y_train)
```

## The Cost Function (Mathematical Heart)

$$ J(w,b) = \frac{1}{2m} \sum_{i=0}^{m-1} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)^2 $$

Where $ f_{w,b}(x) = w \cdot x + b $

```python
def compute_cost(x, y, w, b): 
    m = x.shape[0]
    cost_sum = 0 
    for i in range(m): 
        f_wb = w * x[i] + b
        cost = (f_wb - y[i]) ** 2
        cost_sum += cost
    total_cost = (1 / (2 * m)) * cost_sum
    return total_cost
```

## Experiment with Different Parameters

```python
# Try different values
w = 4.0
b = 5.0
cost = compute_cost(x_train, y_train, w, b)
print(f"For w={w}, b={b} → Cost = {cost:.2f}")
```

**Try yourself**: Change `w` and `b` and rerun to see how cost changes. The goal is to **minimize** the cost.

## Visualization: How Cost Changes

```python
# Generate predictions
def plot_model(w, b):
    x_line = np.linspace(5, 60, 100)
    y_pred = w * x_line + b
    
    plt.figure(figsize=(10, 6))
    plt.scatter(x_train, y_train, marker='o', c='red', s=100, label='Real Data')
    plt.plot(x_line, y_pred, c='blue', linewidth=3, label=f'Model (w={w}, b={b})')
    plt.title('Renewable Investment vs CO₂ Reduction')
    plt.xlabel('Investment ($ Billion)')
    plt.ylabel('CO₂ Reduction (Million Tons)')
    plt.legend()
    plt.grid(True)
    plt.show()

plot_model(4.0, 5.0)
```

## Finding the Optimal Parameters (scikit-learn)

```python
from sklearn.linear_model import LinearRegression

X = x_train.reshape(-1, 1)
model = LinearRegression()
model.fit(X, y_train)

print(f"Optimal slope w     : {model.coef_[0]:.4f}")
print(f"Optimal intercept b : {model.intercept_:.4f}")
print(f"Minimum Cost (MSE)  : {mean_squared_error(y_train, model.predict(X)):.2f}")
```

```python
plot_model(model.coef_[0], model.intercept_)
```

## Real-World Impact Interpretation

- **Slope (`w`)**: For every extra $1 Billion invested in renewables, we reduce CO₂ by approximately `w` million tons.
- **Low Cost** = Reliable model → Better policy decisions.
- **High Cost** = Poor predictions → Wasted money and slower climate action.

**This is how data science directly helps humanity** — optimizing trillions of dollars in green investment.

## Key Takeaways
- The cost function quantifies **how wrong** our model is.
- Squaring the error makes the surface **convex** (soup bowl shape) → always has a minimum.
- Minimizing cost = Finding the best parameters for real impact.
