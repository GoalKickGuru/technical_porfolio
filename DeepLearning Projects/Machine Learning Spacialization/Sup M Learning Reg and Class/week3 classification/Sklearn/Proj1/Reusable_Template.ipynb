- Purely categorical high-cardinality data (consider target encoding or tree models first)
- Need full Bayesian posterior or exact uncertainty quantification

## Responsible checklist before presenting results
1. Residual vs fitted plot – any funnel shape or curvature?
2. Training data range documented – will the model be asked to extrapolate?
3. Audience adaptation – technical coefficients for engineers, plain-language lift for executives.
4. Uncertainty – at least show RMSE / prediction interval estimates.
5. Fairness / bias – check residuals by subgroup if demographic features exist.
