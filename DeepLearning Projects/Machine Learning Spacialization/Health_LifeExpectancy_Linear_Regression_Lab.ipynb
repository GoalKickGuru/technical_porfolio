## Predicting Life Expectancy from Healthcare Spending per Capita

## Goals
- Implement linear regression model
- Apply to Health policy data
- Learn domain-specific interpretation

## Tools

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Problem Statement

| Healthcare Spending per Capita ($1000s) | Life Expectancy (years) |
|------------------------------------------|-------------------------|
| 2.0                                      | 72                      |
| 6.0                                      | 82                      |

Goal: Predict life expectancy for $4,500 spending.

```python
# Training data
x_train = np.array([2.0, 6.0])
y_train = np.array([72.0, 82.0])
print(f"x_train (Spending $1000s) = {x_train}")
print(f"y_train (Life Expectancy) = {y_train}")
```

```python
# Plot data
plt.scatter(x_train, y_train, marker='x', c='r')
plt.title("Healthcare Spending vs Life Expectancy")
plt.ylabel('Life Expectancy (years)')
plt.xlabel('Healthcare Spending per Capita ($1000s)')
plt.show()
```

## Model Function

```python
def compute_model_output(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
# Initial try
w = 2.5
b = 67.0
print(f"w: {w}, b: {b}")

tmp_f_wb = compute_model_output(x_train, w, b)

plt.plot(x_train, tmp_f_wb, c='b', label='Prediction')
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual')
plt.title("Healthcare Spending vs Life Expectancy - Model")
plt.ylabel('Life Expectancy (years)')
plt.xlabel('Healthcare Spending per Capita ($1000s)')
plt.legend()
plt.show()
```

## Prediction

```python
x_new = 4.5
life_pred = w * x_new + b
print(f"Predicted Life Expectancy at ${x_new*1000:.0f} spending: {life_pred:.1f} years")
```

## Interpretation & Next Steps

```python
print("\nKey Insights:")
print("- Slope w ≈ additional years of life per $1000 extra spending")
print("- Important for policy: diminishing returns likely exist in reality")
print("- Expand this: Use scikit-learn LinearRegression on larger datasets")
```
