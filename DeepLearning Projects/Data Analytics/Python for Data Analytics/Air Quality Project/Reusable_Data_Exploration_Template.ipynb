## Step 0 — Configuration (edit these)

Change the names, threshold, and narrative labels to match your domain.

```python
# === DOMAIN CONFIGURATION ===
FEATURE_NAME = "PM 2.5"                    # e.g. "lead_ppb", "avg_temp_F", "daily_traffic"
UNIT = "mcg/m³"                            # e.g. "ppb", "°F", "vehicles"
THRESHOLD = 12.0                           # value that triggers "of concern" messaging
THRESHOLD_LABEL = "moderate risk"          # short phrase for narrative

CATEGORIES = [                             # exactly the groups that appear in the data
    "Chelsea - Clinton",
    "Long Island City - Astoria",
    "Upper West Side"
]

# Short names used for the split lists (must match order of CATEGORIES)
CAT_SHORT = ["CC", "LI", "UWS"]
```

## Step 1 — Load the data

In the original labs a helper `get_list(column)` is supplied.  
For this template we hard-code a small sample so the notebook runs standalone.  
Replace the three lists with your own data (or restore the helper).

```python
# --- SAMPLE DATA (replace with your own) ---
# These values are illustrative; they are not the real EPA series.

geo_place_name = [
    "Chelsea - Clinton", "Chelsea - Clinton", "Chelsea - Clinton",
    "Long Island City - Astoria", "Long Island City - Astoria", "Long Island City - Astoria",
    "Upper West Side", "Upper West Side", "Upper West Side"
]

data_value = [
    14.2, 12.5, 8.4,          # Chelsea - Clinton examples
    11.4, 9.0, 7.2,           # Long Island City - Astoria examples
    12.2, 9.5, 6.8            # Upper West Side examples
]

year = [
    2009, 2014, 2021,
    2009, 2015, 2021,
    2009, 2015, 2021
]

num_observations = len(year)
print(f"Loaded {num_observations} observations.")
```

### Exercise 1 — Types of variables
Print the type of each list and the type of its first element.

```python
print("Variable geo_place_name is of type", type(geo_place_name),
      ", and its elements are of type", type(geo_place_name[0]))

print("Variable data_value is of type", type(data_value),
      ", and its elements are of type", type(data_value[0]))

print("Variable year is of type", type(year),
      ", and its elements are of type", type(year[0]))
```

### Exercise 2 — Length of the lists
Confirm that the three parallel lists have the same length.

```python
print("Variable geo_place_name has length", len(geo_place_name))
print("Variable data_value has length", len(data_value))
print("Variable year has length", len(year))

# Quick sanity check
assert len(geo_place_name) == len(data_value) == len(year), "Lists are not parallel!"
```

### Exercise 3 — Inspect a single observation
Retrieve the observation at a chosen index (remember Python is 0-based).

```python
idx = 0   # change this to explore other rows
observation = [geo_place_name[idx], data_value[idx], year[idx]]
print(observation)
```

## Step 2 — Overall aggregates

### Exercise 4 — Sum of all values

```python
sum_values = sum(data_value)
print(f"The sum of the {FEATURE_NAME} values is {sum_values}")
```

### Exercise 5 — Overall average

```python
average_value = sum(data_value) / len(data_value)
print(f"The average {FEATURE_NAME} is: {average_value:.3f} {UNIT}")
```

## Step 3 — Adding new data

Demonstrate `.append()` on each parallel list.

```python
# Example: append one new observation (edit values as needed)
geo_place_name.append("Chelsea - Clinton")
data_value.append(8.1)
year.append(2022)

print("Updated last elements:")
print("  place :", geo_place_name[-1])
print("  value :", data_value[-1])
print("  year  :", year[-1])
print("New length:", len(data_value))
```

## Step 4 — Maximum observation

```python
max_value = max(data_value)

# Find the first index of the maximum (simple approach for beginners)
idx_max = data_value.index(max_value)

region_max = geo_place_name[idx_max]
year_max = year[idx_max]

print(f"The maximum {FEATURE_NAME} was {max_value} {UNIT}")
print(f"It occurred at {region_max} in the year {year_max}")
```

## Step 5 — Filter observations above a threshold

Use a `for` loop with `range` to visit every index.

```python
print(f"Observations where {FEATURE_NAME} > {THRESHOLD} ({THRESHOLD_LABEL}):")
print("-" * 50)

for i in range(len(data_value)):
    if data_value[i] > THRESHOLD:
        print(geo_place_name[i], year[i], "→", data_value[i], UNIT)
```

## Step 6 — Split data by category

Create one list per category and populate them inside a single loop.

```python
# Create empty lists for each category
CC_values = []   # Chelsea - Clinton
LI_values = []   # Long Island City - Astoria
UWS_values = []  # Upper West Side

for i in range(len(geo_place_name)):
    if geo_place_name[i] == "Chelsea - Clinton":
        CC_values.append(data_value[i])
    elif geo_place_name[i] == "Long Island City - Astoria":
        LI_values.append(data_value[i])
    else:
        # remaining observations belong to Upper West Side
        UWS_values.append(data_value[i])

print("There are", len(CC_values), "observations for Chelsea - Clinton")
print("There are", len(LI_values), "observations for Long Island City - Astoria")
print("There are", len(UWS_values), "observations for Upper West Side")
```

### Exercise — Average per category

```python
average_CC = sum(CC_values) / len(CC_values) if CC_values else float("nan")
average_LI = sum(LI_values) / len(LI_values) if LI_values else float("nan")
average_UWS = sum(UWS_values) / len(UWS_values) if UWS_values else float("nan")

print(f"Chelsea - Clinton average {FEATURE_NAME}: {average_CC:.3f} {UNIT}")
print(f"Long Island City - Astoria average {FEATURE_NAME}: {average_LI:.3f} {UNIT}")
print(f"Upper West Side average {FEATURE_NAME}: {average_UWS:.3f} {UNIT}")
```
