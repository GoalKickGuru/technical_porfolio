## What is Player Efficiency?

The official **Player Efficiency Rating (PER)** by John Hollinger is complex.  
For this lab we use a widely adopted **box-score efficiency** formula that works with the columns available in most game logs:

$$
\text{Efficiency} = \text{PTS} + \text{REB} + \text{AST} + \text{STL} + \text{BLK} - \text{TOV} - (\text{FGA} - \text{FGM}) - (\text{FTA} - \text{FTM})
$$

This rewards positive contributions and penalizes missed shots and turnovers.

```python
eff_avg_query = """
SELECT 
    player,
    team,
    COUNT(*) AS games,
    ROUND(AVG(PTS), 1) AS avg_pts,
    ROUND(AVG(REB), 1) AS avg_reb,
    ROUND(AVG(AST), 1) AS avg_ast,
    ROUND(AVG(
        PTS + REB + AST + STL + BLK 
        - TOV 
        - (FGA - FGM) 
        - (FTA - FTM)
    ), 2) AS avg_efficiency
FROM player_stats
GROUP BY player, team
HAVING games >= 5
ORDER BY avg_efficiency DESC
LIMIT 20;
"""

eff_avg_df = pd.read_sql_query(eff_avg_query, connection)
```

<summary style="background-color: #FDBFC7; padding: 10px; border-radius: 3px; cursor: pointer; font-weight: bold;">Click here to see the solution</summary>

```python
top_per_team_query = """
WITH player_avg AS (
    SELECT 
        player,
        team,
        COUNT(*) AS games,
        ROUND(AVG(
            PTS + REB + AST + STL + BLK 
            - TOV 
            - (FGA - FGM) 
            - (FTA - FTM)
        ), 2) AS avg_efficiency
    FROM player_stats
    GROUP BY player, team
    HAVING games >= 5
),
ranked AS (
    SELECT *,
           RANK() OVER (PARTITION BY team ORDER BY avg_efficiency DESC) AS rnk
    FROM player_avg
)
SELECT player, team, games, avg_efficiency
FROM ranked
WHERE rnk = 1
ORDER BY avg_efficiency DESC;
"""

top_per_team_df = pd.read_sql_query(top_per_team_query, connection)
```

If window functions are not available, you can solve it with a correlated subquery or by joining the max efficiency back to the averages.
