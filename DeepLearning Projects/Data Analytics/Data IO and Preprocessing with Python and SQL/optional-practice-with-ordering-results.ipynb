## SOLUTIONS

## Exercise 1:

```sql
SELECT FirstName, LastName, SupportRepId 
FROM customers 
ORDER BY SupportRepId ASC;
```

## Exercise 2:

```sql
SELECT CustomerId, InvoiceDate, BillingCountry, Total 
FROM invoices 
ORDER BY Total DESC;
```

## Exercise 3:

```sql
SELECT CustomerId, InvoiceDate, BillingCountry, Total 
FROM invoices 
ORDER BY CustomerId ASC, Total DESC;
```

## Exercise 4:

```sql
SELECT
    CustomerId AS "Customer ID",
    InvoiceDate AS Date,
    BillingCountry AS "Billing Country",
    Total
FROM invoices
ORDER BY "Customer ID" ASC, Total DESC;
```
