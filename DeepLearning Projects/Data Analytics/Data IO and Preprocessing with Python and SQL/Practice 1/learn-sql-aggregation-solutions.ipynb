## What are aggregation functions?

Aggregation functions calculate a single value from a set of rows:

| Function | Purpose | Example |
|----------|---------|---------|
| `COUNT()` | Number of rows | `COUNT(*)`, `COUNT(column)` |
| `SUM()` | Total of numeric values | `SUM(Total)` |
| `AVG()` | Average of numeric values | `AVG(UnitPrice)` |
| `MIN()` | Smallest value | `MIN(HireDate)` |
| `MAX()` | Largest value | `MAX(Total)` |

## Key clauses you will use

- **`GROUP BY`** – groups rows that share the same value(s) so aggregates can be calculated *per group*
- **`HAVING`** – filters *groups* (works like `WHERE`, but after aggregation)
- **`ORDER BY`** – sorts the final result (can use aggregate expressions or aliases)

## Important rules

1. Any column in the `SELECT` that is **not** inside an aggregate function **must** appear in the `GROUP BY`.
2. `WHERE` filters rows *before* grouping; `HAVING` filters groups *after* aggregation.
3. You can give aggregates aliases with `AS` for clearer output.

## Exercise 1 – Whole-table aggregates (no GROUP BY)

**Business question:**  
How many invoices exist, and what are the total, average, minimum and maximum invoice amounts?

### Instructions

Write a query that returns:

- Total number of invoices (`COUNT`)
- Sum of all invoice totals (`SUM`)
- Average invoice total (`AVG`)
- Smallest invoice total (`MIN`)
- Largest invoice total (`MAX`)

from the `invoices` table.


````python
SELECT
    COUNT(*) AS InvoiceCount,
    SUM(Total) AS TotalRevenue,
    AVG(Total) AS AvgInvoice,
    MIN(Total) AS MinInvoice,
    MAX(Total) AS MaxInvoice
FROM invoices;
````

**Solution**

```sql
SELECT
    COUNT(*) AS InvoiceCount,
    SUM(Total) AS TotalRevenue,
    AVG(Total) AS AvgInvoice,
    MIN(Total) AS MinInvoice,
    MAX(Total) AS MaxInvoice
FROM invoices;
```

**Hints**

- You can select multiple aggregate functions in one query.
- No `GROUP BY` is needed when you want results for the entire table.
- Example shape: `SELECT COUNT(*), SUM(col), AVG(col), ... FROM table;`


## Exercise 2 – COUNT with GROUP BY

**Business question:**  
How many customers live in each country?

### Instructions

- Select `Country` and the number of customers in that country.
- Group by `Country`.
- Order the results so the countries with the most customers appear first.


````python
SELECT
    Country,
    COUNT(*) AS CustomerCount
FROM customers
GROUP BY Country
ORDER BY CustomerCount DESC;
````

**Solution**

```sql
SELECT
    Country,
    COUNT(*) AS CustomerCount
FROM customers
GROUP BY Country
ORDER BY CustomerCount DESC;
```

**Hints**

- `COUNT(*)` or `COUNT(CustomerId)` both work here.
- `GROUP BY Country`
- `ORDER BY COUNT(*) DESC` (or give the count an alias and order by the alias).

## Exercise 4 – AVG and aliases

**Business question:**  
What is the average track length (in milliseconds) and average unit price for each genre?

### Instructions

You will need to join `tracks` and `genres`:

```sql
FROM tracks t
JOIN genres g ON t.GenreId = g.GenreId
```

Then:

- Select the genre name
- Calculate `AVG(Milliseconds)` and `AVG(UnitPrice)`
- Group by genre name
- Order by average duration descending


````python
SELECT
    g.Name AS Genre,
    AVG(t.Milliseconds) AS "Avg Duration (ms)",
    AVG(t.UnitPrice) AS "Avg Price"
FROM tracks t
JOIN genres g ON t.GenreId = g.GenreId
GROUP BY g.Name
ORDER BY "Avg Duration (ms)" DESC;
````

**Solution**

```sql
SELECT
    g.Name AS Genre,
    AVG(t.Milliseconds) AS "Avg Duration (ms)",
    AVG(t.UnitPrice) AS "Avg Price"
FROM tracks t
JOIN genres g ON t.GenreId = g.GenreId
GROUP BY g.Name
ORDER BY "Avg Duration (ms)" DESC;
```

**Hints**

- Give clear aliases such as `"Avg Duration (ms)"` and `"Avg Price"`.
- Because you are selecting `g.Name`, it must appear in the `GROUP BY`.

## Exercise 6 – Filtering groups with HAVING

**Business question:**  
Which countries have more than 3 customers?

### Instructions

- Select `Country` and the customer count
- Group by `Country`
- Keep only groups where the count is greater than 3
- Order by the count descending


````python
SELECT
    Country,
    COUNT(*) AS CustomerCount
FROM customers
GROUP BY Country
HAVING COUNT(*) > 3
ORDER BY CustomerCount DESC;
````

**Solution**

```sql
SELECT
    Country,
    COUNT(*) AS CustomerCount
FROM customers
GROUP BY Country
HAVING COUNT(*) > 3
ORDER BY CustomerCount DESC;
```

**Hints**

- Use `HAVING COUNT(*) > 3` (not `WHERE`).
- `WHERE` filters individual rows; `HAVING` filters after the groups are formed.


## Exercise 7 – HAVING with SUM

**Business question:**  
Which billing countries generated more than $100 in total revenue?

From `invoices`:

- Select `BillingCountry` and the sum of `Total`
- Group by `BillingCountry`
- Keep only countries whose total revenue exceeds 100
- Order by revenue descending

## Exercise 8 – GROUP BY multiple columns

**Business question:**  
How many invoices were issued per country *and* per year?

### Instructions

SQLite lets you extract the year with `strftime('%Y', InvoiceDate)`.

Write a query that:

- Selects the billing country, the year, and the number of invoices
- Groups by both country and year
- Orders by country, then by year


````python
SELECT
    BillingCountry,
    strftime('%Y', InvoiceDate) AS Year,
    COUNT(*) AS InvoiceCount
FROM invoices
GROUP BY BillingCountry, strftime('%Y', InvoiceDate)
ORDER BY BillingCountry, Year;
````

```sql
    BillingCountry,

ORDER BY BillingCountry, Year;
```

## Exercise 9 – Combining WHERE + GROUP BY + HAVING + ORDER BY

**Business question:**  
Looking only at invoices from 2012 and 2013, which countries had an average invoice total greater than $6?

### Instructions

- Filter rows to years 2012 and 2013 first (`WHERE`)
- Group by `BillingCountry`
- Calculate the average `Total`
- Keep only groups whose average is greater than 6 (`HAVING`)
- Order by the average descending


````python
SELECT
    BillingCountry,
    AVG(Total) AS AvgInvoice
FROM invoices
WHERE strftime('%Y', InvoiceDate) IN ('2012', '2013')
GROUP BY BillingCountry
HAVING AVG(Total) > 6
ORDER BY AvgInvoice DESC;
````

**Solution**

```sql
SELECT
    BillingCountry,
    AVG(Total) AS AvgInvoice
FROM invoices
WHERE strftime('%Y', InvoiceDate) IN ('2012', '2013')
GROUP BY BillingCountry
HAVING AVG(Total) > 6
ORDER BY AvgInvoice DESC;
```

**Hints**

- `WHERE strftime('%Y', InvoiceDate) IN ('2012', '2013')`
- Then `GROUP BY BillingCountry`
- `HAVING AVG(Total) > 6`


## Exercise 10 – Challenge: Support rep performance

**Business question:**  
For each support representative, how many customers do they manage and what is the total revenue generated by those customers’ invoices?

### Instructions

You will need to join three tables:

```sql
customers c
JOIN invoices i ON c.CustomerId = i.CustomerId
```

(The support rep is stored as `c.SupportRepId`.)

Write a query that returns:

- SupportRepId
- Number of distinct customers they support
- Total revenue (`SUM(i.Total)`)
- Average invoice value

Group by SupportRepId, and order by total revenue descending.


````python
SELECT
    c.SupportRepId,
    COUNT(DISTINCT c.CustomerId) AS CustomerCount,
    SUM(i.Total) AS TotalRevenue,
    AVG(i.Total) AS AvgInvoice
FROM customers c
JOIN invoices i ON c.CustomerId = i.CustomerId
GROUP BY c.SupportRepId
ORDER BY TotalRevenue DESC;
````

**Solution**

```sql
SELECT
    c.SupportRepId,
    COUNT(DISTINCT c.CustomerId) AS CustomerCount,
    SUM(i.Total) AS TotalRevenue,
    AVG(i.Total) AS AvgInvoice
FROM customers c
JOIN invoices i ON c.CustomerId = i.CustomerId
GROUP BY c.SupportRepId
ORDER BY TotalRevenue DESC;
```

**Hints**

- `COUNT(DISTINCT c.CustomerId)` gives the number of unique customers.
- `SUM(i.Total)` and `AVG(i.Total)`
- `GROUP BY c.SupportRepId`

## Mental model summary

```

FROM     tables

GROUP BY columns

ORDER BY columns / aggregates
```
