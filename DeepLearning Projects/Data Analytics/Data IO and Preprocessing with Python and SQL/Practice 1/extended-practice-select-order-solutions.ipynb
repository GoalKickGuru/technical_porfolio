## Exercise 5 – Invoice totals with readable column names (aliases)

**Business context:**  
Finance wants a report of invoices sorted from largest to smallest amount, with clear column headers.

### Instructions

Write a SQLite query that:

- Selects `InvoiceId`, `CustomerId`, `InvoiceDate`, and `Total` from `invoices`
- Uses the following aliases:
  - `InvoiceId` → `"Invoice #"`
  - `CustomerId` → `"Customer ID"`
  - `InvoiceDate` → `"Date"`
  - `Total` → `"Amount"`
- Orders by `Total` **descending**


````python
SELECT
    InvoiceId AS "Invoice #",
    CustomerId AS "Customer ID",
    InvoiceDate AS "Date",
    Total AS "Amount"
FROM invoices
ORDER BY Total DESC;
````

**Solution explanation**

```sql
SELECT
    InvoiceId AS "Invoice #",
    CustomerId AS "Customer ID",
    InvoiceDate AS "Date",
    Total AS "Amount"
FROM invoices
ORDER BY Total DESC;
```

**Hints**

- Alias syntax: `column AS "New Name"` (quotes are required when the name contains spaces or special characters).
- You can order by the original column name or by the alias (both work in SQLite).

## Exercise 10 – Challenge: Combined skills

**Business context:**  
Create a customer support view that shows who each customer is assigned to, sorted so that support representatives can easily scan their own clients.

### Instructions

Write a SQLite query that:

- Selects `CustomerId`, `FirstName`, `LastName`, `Country`, `Email`, and `SupportRepId` from `customers`
- Uses these aliases:
  - `CustomerId` → `"ID"`
  - `FirstName` → `"First Name"`
  - `LastName` → `"Last Name"`
  - `SupportRepId` → `"Rep ID"`
- Orders first by `SupportRepId` **ascending**, then by `LastName` **ascending**


````python
SELECT
    CustomerId AS "ID",
    FirstName AS "First Name",
    LastName AS "Last Name",
    Country,
    Email,
    SupportRepId AS "Rep ID"
FROM customers
ORDER BY SupportRepId ASC, LastName ASC;
````

**Solution explanation**

```sql
SELECT
    CustomerId AS "ID",
    FirstName AS "First Name",
    LastName AS "Last Name",
    Country,
    Email,
    SupportRepId AS "Rep ID"
FROM customers
ORDER BY SupportRepId ASC, LastName ASC;
```

**Hints**

- Combine aliases and multi-column ordering.
- `ORDER BY SupportRepId ASC, LastName ASC` (or use the alias `"Rep ID"` if you prefer).
