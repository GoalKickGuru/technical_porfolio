## Step 3: Data Validation

### Exercise 1: Unique Records
Count unique entities vs total rows to detect potential duplicates.


```python
# TEMPLATE CELL: Unique Records
### START CODE HERE ###

query_unique = """
SELECT COUNT(DISTINCT your_id_column) AS num_unique
FROM your_main_table;
"""

query_total = """
SELECT COUNT(*) AS num_rows
FROM your_main_table;
"""

df_unique = pd.read_sql_query(query_unique, connection)
df_total  = pd.read_sql_query(query_total, connection)

### END CODE HERE ###

print(df_unique)
print(df_total)
```

### Exercise 2: Missing Data
Identify rows with critical NULL values (e.g., missing release year, missing rating).


```python
# TEMPLATE CELL: Missing Data
### START CODE HERE ###

query_missing = """
SELECT your_id_column, title_column, other_key_columns
FROM your_main_table
WHERE critical_year_or_date_column IS NULL;
"""

df_missing = pd.read_sql_query(query_missing, connection)

### END CODE HERE ###

df_missing
```

## Step 4: Deeper Analysis

### Exercise 3: Binning / Categorization
Create discrete categories (e.g., High / Medium / Low) using a CASE statement.


```python
# TEMPLATE CELL: Binning
### START CODE HERE ###

query_binning = """
SELECT
    CASE
        WHEN score_column > 3.5 THEN 'High'
        WHEN score_column >= 2.5 AND score_column <= 3.5 THEN 'Medium'
        ELSE 'Low'
    END AS category,
    COUNT(*) AS item_count
FROM your_main_table
GROUP BY category
ORDER BY item_count DESC;
"""

df_binning = pd.read_sql_query(query_binning, connection)

### END CODE HERE ###

df_binning
```

### Exercise 4: Filtered High-Value Items
Find recent / high-scoring items that meet specific business criteria.


```python
# TEMPLATE CELL: High-Rated / Recent
### START CODE HERE ###

query_high = """
SELECT title_column, score_column, year_column
FROM your_main_table
WHERE year_column > 2015
  AND score_column BETWEEN 4.5 AND 4.9
ORDER BY score_column DESC, year_column DESC;
"""

df_high = pd.read_sql_query(query_high, connection)

### END CODE HERE ###

df_high
```

### Exercise 5: Pattern Matching (LIKE)
Search for special editions, variants, or keywords appearing anywhere in titles.


```python
# TEMPLATE CELL: Pattern Matching
### START CODE HERE ###

query_pattern = """
SELECT title_column, director_or_creator, score_column
FROM your_main_table
WHERE title_column LIKE '%keyword%'   -- e.g., '%director''s cut%'
  AND score_column IS NOT NULL
ORDER BY score_column DESC;
"""

df_pattern = pd.read_sql_query(query_pattern, connection)

### END CODE HERE ###

df_pattern
```

### Exercise 6: Aggregations by Time
Compute average scores and maximum popularity/engagement per year (or other period).


```python
# TEMPLATE CELL: Time-based Aggregates
### START CODE HERE ###

query_agg = """
SELECT 
    year_column AS year,
    AVG(score_column) AS avg_score,
    MAX(popularity_column) AS max_popularity
FROM your_main_table
GROUP BY year_column
ORDER BY year DESC
LIMIT 10;
"""

df_agg = pd.read_sql_query(query_agg, connection)

### END CODE HERE ###

df_agg
```

### Exercise 7: Joins for Related Data
Join the main entity table with a feedback/reviews table to surface top content by engagement.


```python
# TEMPLATE CELL: Top Critiques / Reviews via JOIN
### START CODE HERE ###

query_join = """
SELECT 
    m.title_column,
    m.year_column,
    r.critique_or_review_text,
    r.likes_or_engagement
FROM your_main_table m
INNER JOIN your_ratings_or_reviews_table r
    ON m.id_column = r.foreign_key_column
ORDER BY r.likes_or_engagement DESC
LIMIT 10;
"""

df_top = pd.read_sql_query(query_join, connection)

### END CODE HERE ###

df_top
```

### Exercise 8: Identifying Gaps (LEFT JOIN)
Find entities that have no associated ratings or reviews.


```python
# TEMPLATE CELL: Unrated / Missing Feedback
### START CODE HERE ###

query_gaps = """
SELECT COUNT(*) AS num_without_feedback
FROM your_main_table m
LEFT JOIN your_ratings_or_reviews_table r
    ON m.id_column = r.foreign_key_column
WHERE r.foreign_key_column IS NULL;
"""

df_gaps = pd.read_sql_query(query_gaps, connection)

### END CODE HERE ###

df_gaps
```
