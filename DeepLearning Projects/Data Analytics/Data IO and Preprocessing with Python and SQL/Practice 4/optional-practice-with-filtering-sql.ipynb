## **Exercise 6:**

The HR department wants to find all employees working on sales

## Instructions

Write a SQLite query that:

- Selects `LastName`, `FirstName`, and `Title` from the `employees` table
- Filters the employees whose `Title` begins with "Sales".

**Expected output:**

```
+----------+-----------+---------------------+
| LastName | FirstName | Title               |
+----------+-----------+---------------------+
| Edwards  | Nancy     | Sales Manager       |
| Peacock  | Jane      | Sales Support Agent |
| Park     | Margaret  | Sales Support Agent |
| Johnson  | Steve     | Sales Support Agent |
+----------+-----------+---------------------+
```

**Hints:**

- Choose the desired columns with the `SELECT` statement followed by the name of the columns, separated by a comma (,)

## SOLUTIONS:

## Exercise 1:

```sql
SELECT Name, Milliseconds 
FROM tracks 
WHERE Milliseconds > 300000;
```

## Exercise 2:

```sql
SELECT FirstName, LastName, City 
FROM employees 
WHERE City IN ("Edmonton", "Calgary");
```

## Exercise 3:

```sql
SELECT InvoiceId, InvoiceDate, Total 
FROM invoices 
WHERE InvoiceDate BETWEEN "2010-01-01" AND "2010-03-31";
```

## Exercise 4:

```sql
SELECT Name, Milliseconds, UnitPrice 
FROM tracks 
WHERE Milliseconds > 240000 AND UnitPrice < 1.99;
```

## Exercise 5:

```sql
SELECT InvoiceId, BillingCountry, Total 
FROM invoices 
WHERE BillingCountry IN ("USA", "Canada") AND Total > 50;
```

## Exercise 6:

```sql
SELECT LastName, FirstName, Title 
FROM employees 
WHERE Title LIKE "Sales%";
```
