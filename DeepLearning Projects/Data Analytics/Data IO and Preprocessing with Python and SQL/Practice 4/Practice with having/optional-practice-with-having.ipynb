## **Exercise 1 : Minimum and maximum track prices per album**

In the previous exercise, the sales department asked you about the minimum and maximum track prices for each album. It turns out many albums have both their lowest and highest track prices equal to 99 cents. For the marketing purposes, you want to refine this data to show only albums where the maximum price is higher than that.

## Instructions

Write a SQLite query that:

- Groups tracks by `AlbumId`.
- Returns the `AlbumId`, the lowest track price (`MIN`), and the highest track price (`MAX`) for the column UnitPrice.
- Filters the data to only return the albums where the highest track price is above `0.99`.
- Uses the aliases LowestTrackPrice for the minimum price and HighestTrackPrice for the maximum price.

**Expected output:**

```
+---------+------------------+-------------------+
| AlbumId | LowestTrackPrice | HighestTrackPrice |
+---------+------------------+-------------------+
| 226     | 1.99             | 1.99              |
| 227     | 1.99             | 1.99              |
| 228     | 1.99             | 1.99              |
| 229     | 1.99             | 1.99              |
| 230     | 1.99             | 1.99              |
| 231     | 1.99             | 1.99              |
| 249     | 1.99             | 1.99              |
| 250     | 1.99             | 1.99              |
| 251     | 1.99             | 1.99              |
| 253     | 1.99             | 1.99              |
| 254     | 1.99             | 1.99              |
| 261     | 1.99             | 1.99              |
+---------+------------------+-------------------+
```

**Hints:**

- Use the `MIN` statement to find the lowest track price in each album.
- Use the `MAX` statement to find the highest track price in each album.
- Since you are categorizing by album, use `GROUP BY` AlbumId to perform calculations per album.
- Use `HAVING` to select the albums where `HighestTrackPrice` is larger than 0.99.
- To set the alias remember to use the `AS` statement.

## SOLUTION:

## Exercise 1:

```sql
SELECT AlbumId, 
       MIN(UnitPrice) AS LowestTrackPrice, 
       MAX(UnitPrice) AS HighestTrackPrice 
FROM tracks 
GROUP BY AlbumId 
HAVING HighestTrackPrice > 0.99;
```
