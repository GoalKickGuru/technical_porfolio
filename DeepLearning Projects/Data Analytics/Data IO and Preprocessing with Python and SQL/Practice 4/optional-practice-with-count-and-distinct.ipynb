## **Exercise 1 : Number of observations**

The sales department wants to know the total number of customers. You decide to count the rows in the `customers` table.

## Instructions

Write a SQLite query that:

- Counts the number of rows in the `customers` table
- Under the alias `num_customers`.

**Expected output:**

```
+---------------+
| num_customers |
+---------------+
| 59            |
+---------------+
```

**Hints:**

- Use the `COUNT` statement to count the number of rows in the table.
- Since you want the number of rows, regardless of any specific column, you can use the asterisk (*) as the argument for `COUNT`.
- To set the alias remember to use the `AS` statement.

## **Exercise 2: Number of unique clients**

To verify that there are no repeated clients in the dataset, you can compare the previous result with the number of **distinct** rows.

## Instructions

Write a SQLite query that:

- Returns the count of distinct rows in the `customers` table.
- Under the alias `num_unique_customer`.

**Expected output:**

```
+----------------------+
| num_unique_customers |
+----------------------+
| 59                   |
+----------------------+
```

**Hints:**

- Use statements `COUNT` and `DISTINCT` to solve the desired query.
- To set the alias remember to use the `AS` statement.

## SOLUTIONS:

## Exercise 1:

```sql
SELECT COUNT(*) AS num_customers 
FROM CUSTOMERS;
```

## Exercise 2:

```sql
SELECT COUNT(DISTINCT CustomerId) AS num_unique_customers 
FROM CUSTOMERS;
```
