## **Exercise 2: Top spending customers**

Management wants to identify the highest-spending customers to offer them exclusive deals and personalized promotions.

## Instructions

Write a SQLite query that:

- Gets the CustomerId, FirstName, LastName, and total spent by that customer.
- Returns the 5 top spending customers.

**Expected output:**

```
+-----------+------------+------------+--------------------+
| FirstName | LastName   | CustomerId | Total_Spent        |
+-----------+------------+------------+--------------------+
| Helena    | Holý       | 6          | 49.620000000000005 |
| Richard   | Cunningham | 26         | 47.620000000000005 |
| Luis      | Rojas      | 57         | 46.62              |
| Ladislav  | Kovács     | 45         | 45.62              |
| Hugh      | O'Reilly   | 46         | 45.62              |
+-----------+------------+------------+--------------------+
```

**Hints:**

- To include the data from the `customers` table, you can use an INNER JOIN. Join should be made on the `CustomerId`.
- To calculate total spending, use the `SUM` function on the `Total` column from the `invoices` table.
- Group the results by `CustomerID` to ensure totals are calculated per customer.
- Use `ORDER BY` to sort the customers based on their total spending in descending order.
- To get the top 5 spending customers you can use the `LIMIT` statement.
