## Quick Reference – Common Patterns

```sql
-- Two tables
SELECT a.col1, b.col2
FROM table_a a
INNER JOIN table_b b ON a.key = b.key;

-- Three tables
SELECT a.col, b.col, c.col
FROM table_a a
INNER JOIN table_b b ON a.key = b.key
INNER JOIN table_c c ON b.key2 = c.key2;

-- Aggregation after join
SELECT g.Name, SUM(ii.UnitPrice * ii.Quantity) AS Revenue
FROM genres g
INNER JOIN tracks t ON g.GenreId = t.GenreId
INNER JOIN invoice_items ii ON t.TrackId = ii.TrackId
GROUP BY g.Name
ORDER BY Revenue DESC
LIMIT 10;
```
