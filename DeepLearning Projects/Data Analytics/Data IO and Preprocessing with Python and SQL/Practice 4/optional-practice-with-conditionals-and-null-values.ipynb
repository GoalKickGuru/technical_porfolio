## **Exercise 1:**

The HR team wants to identify the area where each employee works at. Employees with "IT" in their title should be categorized under "IT," those with "Sales" under "Sales," and the rest should fall under "Management."

## Instructions

Write a SQLite query that:

- Identifies the area where employees work.
  - If Title starts with the word "IT" in it, then the area is "IT".
  - If Title starts with the word "Sales" in it, then the area is "Sales".
  - Otherwise, area is "Management".
- Uses the alias `Area` for this new column.
- Returns the columns `EmployeeId`, `FirstName`, `LastName`, and `Area` from the `employees` table.

**Expected output:**

```
+------------+-----------+----------+------------+
| EmployeeId | FirstName | LastName | Area       |
+------------+-----------+----------+------------+
| 1          | Andrew    | Adams    | Management |
| 2          | Nancy     | Edwards  | Sales      |
| 3          | Jane      | Peacock  | Sales      |
| 4          | Margaret  | Park     | Sales      |
| 5          | Steve     | Johnson  | Sales      |
| 6          | Michael   | Mitchell | IT         |
| 7          | Robert    | King     | IT         |
| 8          | Laura     | Callahan | IT         |
+------------+-----------+----------+------------+
```

**Hints:**

- Use the `SELECT` statement to retrieve the `EmployeeId`, `FirstName`, and `LastName`, and columns.
- Use the `CASE` statement to categorize employees based on the `Title` column.
- You must have a comma (,) before the `CASE` statement, because you're creating a new column.
- Remember that each condition has syntax `WHEN <condition> THEN <value>`.
- The `ELSE` clause is useful to handle cases that don't fit the given categories, in this case IT and Sales.
- Use the `LIKE` operator to check if the title contains specific words such as "IT" or "Sales."
- Assign an alias to the new column using the `AS` keyword.

## **Exercise 2:**

The marketing team wants to investigate its independent customers. These are the customers where there is no information on their company (`Company` is NULL).

## Instructions

- Filters the `customers` table to only show the independent customers (where `Company` is NULL)
- Returns the columns CustomerId, FirstName and LastName from the `customers` table.

**Hints:**

## SOLUTION:

## Exercise 1:

```sql
SELECT EmployeeId, FirstName, LastName,
CASE 
    WHEN Title LIKE "%IT%" THEN "IT"
    WHEN Title LIKE "%sales%" THEN "Sales"
    ELSE "Management"
END AS Area
FROM employees;
```

## Exercise 2:

```sql
SELECT CustomerId, FirstName, LastName
FROM customers
WHERE Company IS NULL
```
