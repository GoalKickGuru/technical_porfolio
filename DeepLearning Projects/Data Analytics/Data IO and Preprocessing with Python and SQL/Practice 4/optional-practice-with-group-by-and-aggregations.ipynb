## Exercise 2: Spending by invoice categories

Your team wants to analyze customer spending patterns by categorizing invoices based on their total amount. You need to classify invoices into three categories:

- "Low" if the total amount is less than 2.
- "Medium" if the total amount is between 2 and 5.
- "High" if the total amount is greater than 5.
- "NA" if none of the above (missing value).

## Instructions

Write a SQLite query that:

- Categorizes the Total column in invoices in three categories: Low, Medium, and High.
- Groups the data by the category and counts the number of invoices in each category.
- Returns the category and the number of invoices in each category.
- Uses the aliases SpendingCategory for the category, and NumInvoices for the count of invoices.

**Expected output:**

```
+------------------+-------------+
| SpendingCategory | NumInvoices |
+------------------+-------------+
| High             | 179         |
| Low              | 170         |
| Medium           | 63          |
+------------------+-------------+
```

**Hints:**

- Use the `CASE` statement to create spending categories based on invoice totals.
- `BETWEEN` helps define a range (use `Total BETWEEN 2 AND 5` for "Medium").
- When counting invoices, use `COUNT(*)`, as it counts all rows.
- Use `GROUP BY` to ensure each spending category has its own row in the output.
- To set aliases, remember to use the `AS` statement for clarity.

## SOLUTIONS:

## Exercise 1:

```sql
SELECT AlbumId, 
       MIN(UnitPrice) AS LowestTrackPrice, 
       MAX(UnitPrice) AS HighestTrackPrice 
FROM tracks 
GROUP BY AlbumId;
```

## Exercise 2:

```sql
SELECT 
    CASE 
        WHEN Total < 2 THEN 'Low'
        WHEN Total BETWEEN 2 AND 5 THEN 'Medium'
        WHEN Total > 5 THEN 'High'
        ELSE 'NA'
    END AS SpendingCategory,
    COUNT(*) AS NumInvoices
FROM invoices
GROUP BY SpendingCategory;
```
