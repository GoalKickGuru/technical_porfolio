## Predicting GDP per Capita from Average Years of Schooling

**Context**: Governments invest in education expecting economic returns. Linear regression helps quantify this relationship.

## Goals
- Learn to implement the linear model $f_{w,b}$ for regression
- Apply it to real-world Economic data
- Understand interpretation in policy context

## Tools

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Problem Statement

| Avg Years of Schooling | GDP per Capita ($1000s) |
|------------------------|-------------------------|
| 8.0                    | 25                      |
| 13.0                   | 55                      |

Goal: Predict GDP for a country with 11 years of schooling.

```python
# Create training data
x_train = np.array([8.0, 13.0])
y_train = np.array([25.0, 55.0])
print(f"x_train (Schooling) = {x_train}")
print(f"y_train (GDP) = {y_train}")

m = x_train.shape[0]
print(f"Number of training examples: {m}")
```

```python
# Plot the data
plt.scatter(x_train, y_train, marker='x', c='r')
plt.title("Education vs GDP per Capita")
plt.ylabel('GDP per Capita ($1000s)')
plt.xlabel('Average Years of Schooling')
plt.show()
```

## Model Function

```python
def compute_model_output(x, w, b):
    """Compute linear model predictions"""
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
# Try initial parameters
w = 4.0
b = -10.0
print(f"w: {w}, b: {b}")

tmp_f_wb = compute_model_output(x_train, w, b)

# Plot model
plt.plot(x_train, tmp_f_wb, c='b', label='Prediction')
plt.scatter(x_train, y_train, marker='x', c='r', label='Actual')
plt.title("Education vs GDP - Model Prediction")
plt.ylabel('GDP per Capita ($1000s)')
plt.xlabel('Average Years of Schooling')
plt.legend()
plt.show()
```

## Prediction

```python
w = 6.0
b = -23.0
x_new = 11.0  # 11 years schooling
gdp_pred = w * x_new + b
print(f"Predicted GDP per Capita for {x_new} years schooling: ${gdp_pred:.1f} thousand")
```

## Interpretation

```python
print("\nInterpretation:")
print(f"- Slope w={w}: Each additional year of schooling associated with ${w*1000:.0f} increase in GDP per capita")
print("- This is a simplified model. Real analysis uses more data and controls for confounders.")
```
