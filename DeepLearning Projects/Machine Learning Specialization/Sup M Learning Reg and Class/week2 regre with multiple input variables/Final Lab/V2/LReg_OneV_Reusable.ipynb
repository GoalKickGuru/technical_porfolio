## 3. Cost function (loop + optional vectorised)

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost_sum = 0.0
    for i in range(m):
        f_wb = w * x[i] + b
        cost_sum += (f_wb - y[i]) ** 2
    return cost_sum / (2 * m)

# Optional vectorised version
def compute_cost_vec(x, y, w, b):
    return np.mean((w * x + b - y) ** 2) / 2
```

## 4. Gradient

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    dj_dw = dj_db = 0.0
    for i in range(m):
        err = (w * x[i] + b) - y[i]
        dj_dw += err * x[i]
        dj_db += err
    return dj_dw / m, dj_db / m
```

## 5. Batch gradient descent driver

```python
def gradient_descent(x, y, w_in, b_in, cost_fn, grad_fn, alpha, num_iters):
    w, b = copy.deepcopy(w_in), b_in
    J_hist = []
    for i in range(num_iters):
        dj_dw, dj_db = grad_fn(x, y, w, b)
        w -= alpha * dj_dw
        b -= alpha * dj_db
        J_hist.append(cost_fn(x, y, w, b))
        if i % max(1, num_iters // 10) == 0:
            print(f"Iter {i:4d}  cost={J_hist[-1]:.4f}")
    return w, b, J_hist
```
