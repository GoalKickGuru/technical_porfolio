## 5. Performance Check: Vectorized vs. Loop
Whenever your dataset could realistically scale up (more goods, more periods, more countries, simulation draws), confirm that your vectorized approach beats a naive loop. Reuse this cell as a template.

```python
def loop_version(a, b):
    """Reference loop implementation — replace with your own operation."""
    total = 0.0
    for i in range(a.shape[0]):
        total += a[i] * b[i]
    return total

n = 1_000_000
a_test = np.random.rand(n)
b_test = np.random.rand(n)

tic = time.time(); vec_result = np.dot(a_test, b_test); toc = time.time()
print(f"Vectorized: {vec_result:.4f} in {1000*(toc-tic):.3f} ms")

tic = time.time(); loop_result = loop_version(a_test, b_test); toc = time.time()
print(f"Loop:       {loop_result:.4f} in {1000*(toc-tic):.3f} ms")

del a_test, b_test
```
