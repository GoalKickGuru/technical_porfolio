### Core Formulas
| Name | Formula | Notes |
|------|---------|-------|
| Sigmoid / Logistic | \( g(z) = \frac{1}{1+e^{-z}} \) | Maps ℝ → (0,1) |
| Logistic model | \( f_{\mathbf{w},b}(\mathbf{x}) = g(\mathbf{w}\cdot\mathbf{x}+b) \) | Probability \( P(y=1\mid x) \) |
| Decision | \( \hat{y} = 1 \) if \( f \ge \tau \) else 0 | Default \( \tau = 0.5 \) |
| Log-odds / logit | \( \log\frac{p}{1-p} = \mathbf{w}\cdot\mathbf{x}+b \) | Linear in features |

### NumPy essentials
```python
import numpy as np
np.exp(z)          # e^z (vectorized)
1 / (1 + np.exp(-z))  # sigmoid
np.clip(z, -500, 500) # avoid overflow in exp for large |z|
```

### Quick checks
- \( g(0) = 0.5 \)
- \( g(z) \to 1 \) as \( z \to +\infty \)
- \( g(z) \to 0 \) as \( z \to -\infty \)
- Derivative: \( g'(z) = g(z)(1-g(z)) \) (useful later for back-prop / GD)

### Common pitfalls
- Forgetting the minus sign in `exp(-z)`
- Overflow when `|z|` is huge → use `np.clip` or `scipy.special.expit`
- Treating the raw probability as a hard label without a threshold
- Assuming the decision boundary is linear in the original feature space without checking

## 7. Simulation Section — Modify Parameters & Observe

You can change the values in the **parameter box** below and re-run to see different behaviour.

**Ideas to explore:**
- Change true underlying `w_true`, `b_true` (steeper or shifted sigmoid)
- Change decision threshold
- Add Gaussian noise to the linear predictor or flip labels
- Change sample size `n`
- Observe how accuracy and the probability curve react

This section helps build intuition about sensitivity and robustness.


````python
# === PARAMETER BOX (edit these) ===
np.random.seed(42)
n               = 150          # sample size
w_true          = 1.5          # true weight
b_true          = -6.0         # true bias
threshold       = 0.5
label_noise_frac= 0.05         # fraction of labels to flip
z_noise_std     = 0.3          # noise added to z before sigmoid

# Generate synthetic data from the true logistic model
x_sim = np.random.uniform(0, 10, n)
z_clean = w_true * x_sim + b_true
z_noisy = z_clean + np.random.normal(0, z_noise_std, n)
p_true  = 1 / (1 + np.exp(-z_noisy))
y_sim   = (np.random.rand(n) < p_true).astype(int)
# label noise
flip_idx = np.random.choice(n, size=int(label_noise_frac * n), replace=False)
y_sim[flip_idx] = 1 - y_sim[flip_idx]

# "Learned" parameters (here we pretend we recovered the true ones; 
# in a full lab you would run gradient descent)
w_hat, b_hat = w_true, b_true

probs = 1 / (1 + np.exp(-(w_hat * x_sim + b_hat)))
preds = (probs >= threshold).astype(int)
acc   = (preds == y_sim).mean()

print(f"Accuracy at τ={threshold}: {acc:.3f}")
print(f"Positive rate (true): {y_sim.mean():.3f}")

# Plot
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].scatter(x_sim[y_sim==0], y_sim[y_sim==0], c='C0', alpha=0.6, label='y=0')
axes[0].scatter(x_sim[y_sim==1], y_sim[y_sim==1], c='C3', alpha=0.6, label='y=1')
x_line = np.linspace(0, 10, 200)
axes[0].plot(x_line, 1/(1+np.exp(-(w_hat*x_line+b_hat))), 'k-', lw=2, label='f(x)')
axes[0].axhline(threshold, color='orange', ls='--')
axes[0].set_title(f'Simulated data + model (acc={acc:.2f})')
axes[0].legend()
axes[0].set_xlabel('x'); axes[0].set_ylabel('y / p')

# Threshold sweep
ths = np.linspace(0.1, 0.9, 17)
accs = [((1/(1+np.exp(-(w_hat*x_sim+b_hat))) >= t).astype(int) == y_sim).mean() for t in ths]
axes[1].plot(ths, accs, 'o-', color='purple')
axes[1].axvline(threshold, color='orange', ls='--')
axes[1].set_xlabel('Threshold'); axes[1].set_ylabel('Accuracy')
axes[1].set_title('Accuracy vs Threshold (this draw)')
plt.tight_layout()
plt.show()
````

## 8. What It Can & Cannot Predict — Responsible Use Notes

**Can predict / produce:**
- A calibrated-ish probability \( P(y=1\mid x) \) under the logistic assumption
- A hard binary decision after choosing a threshold
- Odds ratios (how features multiply the odds) when the model is well-specified

**Cannot (with the basic model):**
- Multi-class probabilities (need softmax / multinomial logistic)
- Continuous numeric targets (use linear / other regression)
- Causal effects (“this treatment causes …”) without experimental design or strong assumptions
- Complex non-linear decision surfaces without feature engineering, kernels, or switching to other models
- Reliable predictions far outside the range of training `x` (extrapolation)

**Responsible practices:**
- Always examine class balance and choose threshold according to the cost of FP vs FN
- Validate on held-out data; report confidence intervals or calibration plots when possible
- Document the population the model was trained on; do not claim universal validity
- For high-stakes domains (medical, credit, criminal justice) require human oversight and fairness audits
- Prefer simpler, interpretable models when stakeholders need to understand *why*
