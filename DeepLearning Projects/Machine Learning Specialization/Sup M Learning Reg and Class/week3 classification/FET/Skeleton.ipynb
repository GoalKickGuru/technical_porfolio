## Cheat Sheet (quick reference)

| Transform | Formula / code | Typical use |
|-----------|----------------|-------------|
| Centering | `x - mean` | interpret deviations |
| Standardization | `(x - μ) / σ` or `StandardScaler` | PCA, KNN, regularized LR |
| Min-Max | `(x - min)/(max - min)` or `MinMaxScaler` | neural nets, [0,1] bounds |
| Log | `np.log1p(x)` | right-skewed positive data |
| Binning | `pd.cut` / `pd.qcut` | non-linear effects, ordinal |

Always **fit on train, transform on test**.
