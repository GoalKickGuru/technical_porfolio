## Inline cheat-sheet (keep this cell visible)

See also **`LogReg_GD_Sklearn_Cheatsheet.docx`**.

| Item | Formula / code |
|------|----------------|
| Sigmoid | $g(z)=1/(1+e^{-z})$ · clip $z$ to $[-50,50]$ |
| Model | $f=g(w\cdot x+b)$ |
| Cost | $J=\frac{1}{m}\sum -[y\log f+(1-y)\log(1-f)]$ |
| $\partial J/\partial w_j$ | $\frac{1}{m}\sum_i (f^{(i)}-y^{(i)})x_j^{(i)}$ |
| $\partial J/\partial b$ | $\frac{1}{m}\sum_i (f^{(i)}-y^{(i)})$ |
| Update | $w \leftarrow w-\alpha\,\partial J/\partial w$, same for $b$ |
| Lab06 check | $w=(2,3),\,b=1$ → `dj_db≈0.49862`, `dj_dw≈[0.49833, 0.49884]` |
| sklearn | `LogisticRegression(C=np.inf)` ≈ unregularized; default `C=1` is L2 |
| Decision | predict 1 if $f \ge 0.5$ (or another risk threshold) |

**Flow:** data → $z$ → sigmoid $f$ → error $(f-y)$ → gradient → GD step → boundary / sklearn compare → simulate.

## 2. Logistic gradient descent

Recall the simultaneous updates

$$
w_j \leftarrow w_j - \alpha \frac{\partial J}{\partial w_j},\qquad
b \leftarrow b - \alpha \frac{\partial J}{\partial b}
$$

with

$$
\frac{\partial J}{\partial w_j} = \frac{1}{m}\sum_{i=0}^{m-1}(f_{w,b}(x^{(i)})-y^{(i)})x_j^{(i)},\qquad
\frac{\partial J}{\partial b} = \frac{1}{m}\sum_{i=0}^{m-1}(f_{w,b}(x^{(i)})-y^{(i)}).
$$

$f_{w,b}(x)=g(w\cdot x+b)$ and $g$ is the sigmoid.
