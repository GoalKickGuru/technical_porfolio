### 1.6 Gradient Descent
Use the functions you wrote to minimise J. Start with small α and increase iterations until the cost plateaus.


```python
def gradient_descent(X, y, w_in, b_in, cost_fn, grad_fn, alpha, num_iters, lambda_=0):
    """
    Batch gradient descent for logistic regression
    """
    w = copy.deepcopy(w_in)
    b = b_in
    J_history = []
    for i in range(num_iters):
        dj_db, dj_dw = grad_fn(X, y, w, b, lambda_)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        if i < 100000:
            J_history.append(cost_fn(X, y, w, b, lambda_))
        if i % math.ceil(num_iters/10) == 0:
            print(f"Iteration {i:5d}: Cost {J_history[-1]:.4f}")
    return w, b, J_history

# TODO: run GD
# initial_w = np.zeros(2)
# initial_b = 0.
# w, b, J_hist = gradient_descent(X_train, y_train, initial_w, initial_b,
#                                 compute_cost, compute_gradient, alpha=0.001, num_iters=100000)

# YOUR CODE HERE

print("Learned parameters w, b =", w, b)
```
